## Project Overview

This project presents an end-to-end environmental data science study
of air quality patterns across Indian cities.

It investigates pollutant distributions, environmental trends,
city-level differences, seasonal variation, statistical relationships,
pollution exposure and environmental risk indicators.

The project emphasizes exploratory data analysis, statistical
inference, temporal analysis, environmental risk assessment and
statistical time-series methods.

## Project Objectives

1. Analyze air quality and pollutant distributions.
2. Investigate temporal and seasonal pollution patterns.
3. Compare environmental conditions across cities.
4. Examine pollutant–AQI statistical relationships.
5. Identify persistent and extreme pollution risks.
6. Develop composite environmental risk indicators.
7. Apply statistical robustness and time-series diagnostics.
8. Communicate findings through visualizations and reports.

## Methodological Scope

This is a **Data Science and statistical analysis project**.
It does not use machine learning, deep learning or generative AI.

## Dataset Information

The project uses the Indian Air Quality dataset containing
city-level daily observations and pollutant measurements.

The analysis preserves missing-data information and treats
extreme AQI observations carefully rather than automatically
removing or imputing them.

```python
# create a professional missing-value report

missing_report = pd.DataFrame({
    "Missing_Count": air_df.isnull().sum(),
    "Missing_Percentage": air_df.isnull().mean() * 100,
    "Available_Count": air_df.notnull().sum()
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_report
```

```python

plt.figure(figsize=(14, 7))

sns.heatmap(
    air_df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Value Distribution")
plt.xlabel("Dataset Columns")
plt.ylabel("Observations")

plt.tight_layout()
plt.show()
```

```python
# Understanding your heatmap
# Purple/dark areas: Values are present.
# Yellow areas: Values are missing.
# Horizontal patterns indicate missing values across observations.
# Some columns have more visible missing-value patterns than others.
# This tells us that missing data is not necessarily distributed equally across all variables.
```

```python
# missing value severity

# Create a missing-value severity classification

missing_report = pd.DataFrame({
    "Missing_Count": air_df.isnull().sum(),
    "Missing_Percentage": air_df.isnull().mean() * 100
})

missing_report["Data_Status"] = np.where(
    missing_report["Missing_Percentage"] == 0,
    "Complete",
    np.where(
        missing_report["Missing_Percentage"] <= 5,
        "Low Missingness",
        np.where(
            missing_report["Missing_Percentage"] <= 30,
            "Moderate Missingness",
            "High Missingness"
        )
    )
)

missing_report = missing_report.sort_values(
    "Missing_Percentage",
    ascending=False
)

missing_report.round(2)
```

```python
# identify columns with most missing data

print("Top 5 columns with highest missing values:")

missing_report.head(5)
```

```python
# calculate overall data
total_cells = air_df.size
missing_cells = air_df.isnull().sum().sum()

complete_cells = total_cells - missing_cells

print("Total cells:", total_cells)
print("Missing cells:", missing_cells)
print("Complete cells:", complete_cells)

print(
    "Overall completeness:",
    round((complete_cells / total_cells) * 100, 2),
    "%"
)
```

```python
# Missingness for selected pollutands

pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

city_pollutant_missing = (
    air_df.groupby("City")[pollutant_columns]
    .apply(lambda group: group.isnull().mean() * 100)
)

city_pollutant_missing.round(2)
```

```python
# What your heatmap tells us

# Your visualization shows that missing pollutant measurements vary across cities and pollutants.
# Some visible patterns:
# Ahmedabad: Several pollutants have noticeable missingness.
# Chennai: PM10 has a particularly strong missingness pattern.
#Bhopal: Some of the later pollutant columns show substantial missingness.
# Mumbai: Missingness appears across several pollutant measurements.
# Thiruvananthapuram: Some later pollutant measurements have substantial missingness.
# These are visual observations, not yet quantified rankings. We need numerical results before making conclusions.
```

```python
# next -rank cities by missingness

# Calculate overall pollutant missingness by city

city_missing_summary = pd.DataFrame({
    "Missing_Percentage":
        air_df.groupby("City")[pollutant_columns]
        .apply(lambda x: x.isnull().mean().mean() * 100),

    "Total_Records":
        air_df.groupby("City").size()
})

city_missing_summary = city_missing_summary.sort_values(
    by="Missing_Percentage",
    ascending=False
)

city_missing_summary.round(2)
```

```python
# visualize the top 10 cities

top_10_missing = city_missing_summary.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_missing.reset_index(),
    x="Missing_Percentage",
    y="City"
)

plt.title("Top 10 Cities by Average Pollutant Missingness")
plt.xlabel("Average Missing Values (%)")
plt.ylabel("City")

plt.tight_layout()
plt.show()
```

```python
# Create the working

# Reload original CSV to preserve an untouched source

air_raw = pd.read_csv(file_path)

# Create a separate working copy
air_df = air_raw.copy()

print("Original dataset:", air_raw.shape)
print("Working dataset:", air_df.shape)
print("Original data preserved:", air_raw.equals(air_df))
```

```python
# validate data types

dtype_report = pd.DataFrame({
    "Column": air_df.columns,
    "Data_Type": air_df.dtypes.astype(str).values,
    "Missing_Count": air_df.isnull().sum().values,
    "Unique_Values": air_df.nunique().values
})

dtype_report
```

```python
# convert and validate dates

air_df["Date"] = pd.to_datetime(
    air_df["Date"],
    errors="coerce"
)

print("Date data type:", air_df["Date"].dtype)
print("Invalid dates:", air_df["Date"].isnull().sum())
print("Earliest date:", air_df["Date"].min())
print("Latest date:", air_df["Date"].max())
```

```python
# identify the numercial values

numeric_columns = air_df.select_dtypes(
    include=["number"]
).columns

invalid_value_report = pd.DataFrame({
    "Negative_Count": (air_df[numeric_columns] < 0).sum(),
    "Zero_Count": (air_df[numeric_columns] == 0).sum(),
    "Missing_Count": air_df[numeric_columns].isnull().sum()
})

invalid_value_report
```

```python
# numerical statistics assessment

# Identify numerical columns
numeric_columns = air_df.select_dtypes(
    include=["number"]
).columns

# Generate descriptive statistics
numerical_summary = air_df[numeric_columns].describe().T

numerical_summary["Missing_Count"] = air_df[numeric_columns].isnull().sum()

numerical_summary["Zero_Count"] = (
    air_df[numeric_columns] == 0
).sum()

numerical_summary["Zero_Percentage"] = (
    numerical_summary["Zero_Count"] / len(air_df) * 100
).round(2)

numerical_summary
```

```python
# save the statistical report

numerical_summary.to_csv(
    report_dir + r"\numerical_statistical_summary.csv"
)

print("Numerical statistical report saved successfully!")
```

```python
# calculate the IQR boundries

# Identify numerical columns
numeric_columns = air_df.select_dtypes(
    include=["number"]
).columns

# Calculate quartiles
Q1 = air_df[numeric_columns].quantile(0.25)
Q3 = air_df[numeric_columns].quantile(0.75)

# Calculate interquartile range
IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("IQR boundaries calculated successfully!")
```

```python
# count potential outliers

outlier_report = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound": lower_bound,
    "Upper_Bound": upper_bound,
    "Outlier_Count": (
        (air_df[numeric_columns] < lower_bound) |
        (air_df[numeric_columns] > upper_bound)
    ).sum()
})

outlier_report["Outlier_Percentage"] = (
    outlier_report["Outlier_Count"] / len(air_df) * 100
).round(2)

outlier_report = outlier_report.sort_values(
    "Outlier_Percentage",
    ascending=False
)

outlier_report
```

```python
# save the outliers report

outlier_report.to_csv(
    report_dir + r"\outlier_detection_report.csv"
)

print("Outlier detection report saved successfully!")
```

```python
# Run the outlier detection analysis
import pandas as pd
import numpy as np

pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

outlier_results = []

for column in pollutant_columns:

    data = air_df[column].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_limit) |
        (data > upper_limit)
    ]

    outlier_results.append({
        "Pollutant": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (
            len(outliers) / len(data) * 100
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

outlier_summary.round(2)
```

```python
#save the results

outlier_summary.to_csv(
    os.path.join(report_dir, "pollutant_outlier_analysis.csv"),
    index=False
)

print("Outlier analysis report saved successfully!")
```

```python
# Calculate Spearman correlation with p-values

from scipy.stats import spearmanr
import pandas as pd
import numpy as np

# Store statistical test results
correlation_test_results = []

for column in pollutant_columns:

    # Select valid paired observations
    temp = air_df[[column, "AQI"]].dropna()

    correlation, p_value = spearmanr(
        temp[column],
        temp["AQI"]
    )

    correlation_test_results.append({
        "Pollutant": column,
        "Sample_Size": len(temp),
        "Spearman_Correlation": correlation,
        "P_Value": p_value
    })

correlation_test = pd.DataFrame(
    correlation_test_results
)

correlation_test.round(5)
```

```python

# Interpret statistical significance

correlation_test["Significance"] = np.where(
    correlation_test["P_Value"] < 0.05,
    "Statistically Significant",
    "Not Statistically Significant"
)

correlation_test["Relationship"] = np.where(
    correlation_test["Spearman_Correlation"] > 0,
    "Positive",
    "Negative"
)

correlation_test
```

```python
### Interpretation:
- p-value below 0.05: evidence against the null hypothesis of zero population rank correlation.
- p-value at or above 0.05: insufficient evidence to reject that null hypothesis.
- Positive correlation: higher pollutant measurements tend to accompany higher AQI.
- Negative correlation: higher pollutant measurements tend to accompany lower AQI.
A statistically significant result does not automatically mean the relationship is strong or causal.
```

```python


# Save the statistical test report
correlation_test.to_csv(
    os.path.join(
        report_dir,
        "spearman_statistical_significance.csv"
    ),
    index=False
)

print("Statistical significance report saved!")
```

```python
# Multiple Hypothesis Testing Correction.
```

```python
# Apply Benjamini–Hochberg correction

from statsmodels.stats.multitest import multipletests
import numpy as np
import pandas as pd

# Correct the p-values from our Spearman tests
rejected, adjusted_pvalues, _, _ = multipletests(
    correlation_test["P_Value"],
    alpha=0.05,
    method="fdr_bh"
)

correlation_test["Adjusted_P_Value"] = adjusted_pvalues

correlation_test["FDR_Significant"] = np.where(
    rejected,
    "Significant",
    "Not Significant"
)

correlation_test.round(5)
```

```python
# Compare original and adjusted p-values

pvalue_comparison = correlation_test[
    [
        "Pollutant",
        "Spearman_Correlation",
        "P_Value",
        "Adjusted_P_Value",
        "FDR_Significant"
    ]
].sort_values("Adjusted_P_Value")

pvalue_comparison
```

```python
#Save the corrected statistical report

pvalue_comparison.to_csv(
    os.path.join(
        report_dir,
        "multiple_testing_correction_report.csv"
    ),
    index=False
)

print("Multiple testing correction report saved!")
```

```python
# Prepare a readable statistical report
statistical_report = correlation_test.copy()

statistical_report["P_Value_Display"] = statistical_report["P_Value"].apply(
    lambda p: "< 1e-300" if p == 0 else f"{p:.3e}"
)

statistical_report["Adjusted_P_Value_Display"] = statistical_report[
    "Adjusted_P_Value"
].apply(
    lambda p: "< 1e-300" if p == 0 else f"{p:.3e}"
)

# Save report
statistical_report.to_csv(
    os.path.join(report_dir, "spearman_correlation_fdr_report.csv"),
    index=False
)

print("Statistical report saved successfully.")
print("Shape:", statistical_report.shape)

statistical_report[
    ["Pollutant", "Spearman_Correlation",
     "P_Value_Display", "Adjusted_P_Value_Display", "FDR_Significant"]
]
```

```python
# Bootstrap confidence intervals

import numpy as np
import pandas as pd
from scipy.stats import spearmanr

# Pollutants to analyse
pollutants = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

np.random.seed(42)

bootstrap_results = []SE

for pollutant in pollutants:

    temp = air_df[[pollutant, "AQI"]].dropna()

    x = temp[pollutant].to_numpy()
    y = temp["AQI"].to_numpy()

    correlations = []

    for _ in range(1000):
        indices = np.random.randint(0, len(temp), len(temp))

        corr, _ = spearmanr(x[indices], y[indices])

        if np.isfinite(corr):
            correlations.append(corr)

    lower = np.percentile(correlations, 2.5)
    upper = np.percentile(correlations, 97.5)

    bootstrap_results.append({
        "Pollutant": pollutant,
        "Sample_Size": len(temp),
        "Spearman_Correlation": spearmanr(x, y)[0],
        "CI_Lower_95": lower,
        "CI_Upper_95": upper
    })

bootstrap_ci_df = pd.DataFrame(bootstrap_results)

bootstrap_ci_df.round(5)
```

```python
# Prepare temporal features

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Create a separate temporal analysis dataset
air_temporal = air_df.copy()

# Convert Date into datetime format
air_temporal["Date"] = pd.to_datetime(
    air_temporal["Date"],
    errors="coerce"
)

# Extract time-based features
air_temporal["Year"] = air_temporal["Date"].dt.year
air_temporal["Month"] = air_temporal["Date"].dt.month
air_temporal["Month_Name"] = air_temporal["Date"].dt.month_name()

# Define Indian seasonal periods
def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Pre-Monsoon"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

air_temporal["Season"] = air_temporal["Month"].apply(assign_season)

print("Temporal features created successfully.")
print("Dataset shape:", air_temporal.shape)

air_temporal[
    ["Date", "City", "AQI", "Year", "Month",
     "Month_Name", "Season"]
].head(10)
```

```python
What this creates:
- Year — annual pollution analysis.
- Month and Month_Name — monthly patterns.
- Season — seasonal comparisons.
These seasonal labels are analytical groupings, not meteorological observations for each city.
```

```python
# ruskal–Wallis test to determine whether the AQI distributions differ significantly across Winter, Post-Monsoon, Pre-Monsoon and Monsoon
```

```python
# Prepare seasonal samples

from scipy.stats import kruskal

season_groups = {
    season: group["AQI"].dropna().values
    for season, group in air_temporal.groupby("Season", observed=True)
}

# Check sample sizes
for season, values in season_groups.items():
    print(f"{season}: {len(values)} observations")
```

```python
# Run Kruskal–Wallis test

# Statistical test across all four seasons

kruskal_stat, kruskal_p = kruskal(
    *season_groups.values()
)

print("Kruskal-Wallis Statistical Analysis")
print("-----------------------------------")
print("H-statistic:", round(kruskal_stat, 4))
print("P-value:", kruskal_p)

alpha = 0.05

if kruskal_p < alpha:
    print("Conclusion: Statistically significant seasonal differences.")
else:
    print("Conclusion: No statistically significant seasonal differences.")
    
```

```python
xcellent! Your statistical test executed successfully.
1. Verified results
Kruskal–Wallis H-statistic
4136.0776

P-value
Below 0.05
Statistically significant

Interpretation: There is strong evidence that AQI distributions are not identical across all four seasons. At least one seasonal distribution differs from another.

However, this test does not identify which seasons differ.
```

```python
# Pairwise seasonal comparisons

from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
from itertools import combinations
import pandas as pd
import numpy as np

season_names = [
    "Winter",
    "Post-Monsoon",
    "Pre-Monsoon",
    "Monsoon"
]

pairwise_results = []

for season1, season2 in combinations(season_names, 2):

    group1 = air_temporal.loc[
        air_temporal["Season"] == season1, "AQI"
    ].dropna()

    group2 = air_temporal.loc[
        air_temporal["Season"] == season2, "AQI"
    ].dropna()

    statistic, p_value = mannwhitneyu(
        group1,
        group2,
        alternative="two-sided"
    )

    pairwise_results.append({
        "Season_1": season1,
        "Season_2": season2,
        "U_Statistic": statistic,
        "P_Value": p_value
    })

pairwise_df = pd.DataFrame(pairwise_results)

# Multiple-testing correction
reject, adjusted_p, _, _ = multipletests(
    pairwise_df["P_Value"],
    alpha=0.05,
    method="fdr_bh"
)

pairwise_df["Adjusted_P_Value"] = adjusted_p
pairwise_df["Significant"] = np.where(
    reject, "Significant", "Not Significant"
)

display(pairwise_df.round(6))
```

```python
# Measure the size of seasonal differences
Statistical significance does not tell us whether the differences are practically large. We'll calculate:
- Median AQI for each season.
- Difference between seasonal medians.
- Rank-biserial effect size for each pair.
Run this in a new cell:
```

```python
# Measure the size of seasonal differences

# Add practical effect sizes to seasonal comparisons

effect_results = []

for season1, season2 in combinations(season_names, 2):

    group1 = air_temporal.loc[
        air_temporal["Season"] == season1, "AQI"
    ].dropna()

    group2 = air_temporal.loc[
        air_temporal["Season"] == season2, "AQI"
    ].dropna()

    u_stat, p_value = mannwhitneyu(
        group1, group2, alternative="two-sided"
    )

    n1 = len(group1)
    n2 = len(group2)

    # Rank-biserial correlation
    effect_size = (2 * u_stat / (n1 * n2)) - 1

    effect_results.append({
        "Season_1": season1,
        "Season_2": season2,
        "Median_1": group1.median(),
        "Median_2": group2.median(),
        "Median_Difference": group1.median() - group2.median(),
        "Rank_Biserial_Effect": effect_size
    })

effect_size_df = pd.DataFrame(effect_results)

display(effect_size_df.round(4))
```

```python
#save all seasonal analysis reports

import os

# Ensure report directory exists
os.makedirs(report_dir, exist_ok=True)

# Save seasonal summary
seasonal_summary.to_csv(
    os.path.join(report_dir, "seasonal_aqi_summary.csv"),
    index=False
)

# Save Kruskal-Wallis test
kruskal_report = pd.DataFrame([{
    "Test": "Kruskal-Wallis",
    "H_Statistic": kruskal_stat,
    "P_Value": kruskal_p,
    "Conclusion": "Significant" if kruskal_p < 0.05
                  else "Not Significant"
}])

kruskal_report.to_csv(
    os.path.join(report_dir, "seasonal_kruskal_wallis.csv"),
    index=False
)

# Save pairwise comparisons
pairwise_df.to_csv(
    os.path.join(report_dir, "seasonal_pairwise_comparisons.csv"),
    index=False
)

# Save effect sizes
effect_size_df.to_csv(
    os.path.join(report_dir, "seasonal_effect_sizes.csv"),
    index=False
)

print("All seasonal analysis reports saved successfully.")
```

```python
# Missing-value treatment strategy

# MISSING VALUE TREATMENT STRATEGY

treatment_strategy = pollutant_coverage.copy()

def recommend_treatment(missing_percentage):
    if missing_percentage <= 10:
        return "Evaluate limited imputation"
    elif missing_percentage <= 20:
        return "Conditional imputation"
    elif missing_percentage <= 40:
        return "City/year-specific assessment"
    else:
        return "Retain missing; avoid blind imputation"

treatment_strategy["Recommended_Strategy"] = (
    treatment_strategy["Missing_Percentage"]
    .apply(recommend_treatment)
)

display(treatment_strategy)
```

```python
# CITY-WISE POLLUTANT DATA AVAILABILITY

key_pollutants = [
    "PM2.5", "PM10", "NO2", "NOx",
    "CO", "SO2", "O3"
]

coverage_records = []

for city, group in air_df.groupby("City"):

    for pollutant in key_pollutants:

        total_records = len(group)
        valid_records = group[pollutant].notna().sum()

        coverage_records.append({
            "City": city,
            "Pollutant": pollutant,
            "Total_Records": total_records,
            "Valid_Records": valid_records,
            "Missing_Records": total_records - valid_records,
            "Coverage_Percentage": round(
                valid_records / total_records * 100, 2
            )
        })

city_pollutant_coverage = pd.DataFrame(coverage_records)

print("Report shape:", city_pollutant_coverage.shape)

display(city_pollutant_coverage.head(20))
```

```python
# AQI AND POLLUTANT CORRELATION ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

pollutants = [
    "PM2.5", "PM10", "NO", "NO2",
    "NOx", "NH3", "CO", "SO2",
    "O3", "Benzene", "Toluene", "Xylene"
]

correlation_data = air_df[
    ["AQI"] + pollutants
].copy()

pearson_corr = correlation_data.corr(
    method="pearson",
    min_periods=100
)

aqi_correlation = (
    pearson_corr["AQI"]
    .drop("AQI")
    .sort_values(ascending=False)
)

display(
    aqi_correlation.to_frame(
        name="Pearson_Correlation_with_AQI"
    ).round(3)
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    pearson_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title(
    "Pearson Correlation: AQI and Pollutants",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "aqi_pollutant_pearson_correlation.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# SPEARMAN CORRELATION ANALYSIS

spearman_corr = correlation_data.corr(
    method="spearman",
    min_periods=100
)

spearman_aqi = (
    spearman_corr["AQI"]
    .drop("AQI")
    .sort_values(ascending=False)
)

display(
    spearman_aqi.to_frame(
        name="Spearman_Correlation_with_AQI"
    ).round(3)
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    spearman_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title(
    "Spearman Correlation: AQI and Pollutants",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "aqi_pollutant_spearman_correlation.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# PEARSON VS SPEARMAN COMPARISON

correlation_comparison = pd.DataFrame({
    "Pearson": pearson_corr["AQI"].drop("AQI"),
    "Spearman": spearman_corr["AQI"].drop("AQI")
})

correlation_comparison["Difference"] = (
    correlation_comparison["Spearman"]
    - correlation_comparison["Pearson"]
)

correlation_comparison["Absolute_Difference"] = (
    correlation_comparison["Difference"].abs()
)

correlation_comparison = correlation_comparison.sort_values(
    "Absolute_Difference",
    ascending=False
)

display(correlation_comparison.round(3))

# Visualization
correlation_comparison[
    ["Pearson", "Spearman"]
].plot(
    kind="bar",
    figsize=(13, 6),
    width=0.8
)

plt.title(
    "Pearson vs Spearman Correlation with AQI",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Pollutants")
plt.ylabel("Correlation Coefficient")
plt.xticks(rotation=45)
plt.axhline(0, color="black", linewidth=0.8)
plt.grid(axis="y", alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "pearson_vs_spearman_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Save statistical comparison
correlation_comparison.to_csv(
    os.path.join(
        report_dir,
        "pearson_vs_spearman_comparison.csv"
    )
)

print("Correlation comparison report saved successfully.")
```

```python
# Statistical significance testing

from scipy.stats import spearmanr
from statsmodels.stats.multitest import multipletests

# STATISTICAL SIGNIFICANCE TESTING

significance_results = []

for pollutant in pollutants:

    temp = air_df[["AQI", pollutant]].dropna()

    correlation, p_value = spearmanr(
        temp["AQI"],
        temp[pollutant]
    )

    significance_results.append({
        "Pollutant": pollutant,
        "Sample_Size": len(temp),
        "Spearman_Correlation": correlation,
        "P_Value": p_value
    })

significance_df = pd.DataFrame(significance_results)

# Multiple testing correction
significance_df["Adjusted_P_Value"] = multipletests(
    significance_df["P_Value"],
    method="fdr_bh"
)[1]

significance_df["Significant_After_FDR"] = (
    significance_df["Adjusted_P_Value"] < 0.05
)

significance_df = significance_df.sort_values(
    "Adjusted_P_Value"
)

display(significance_df.round(6))

# Save report
significance_df.to_csv(
    os.path.join(
        report_dir,
        "aqi_correlation_significance_testing.csv"
    ),
    index=False
)

print("Statistical significance report saved successfully.")
```

```python
# City-wise AQI trend direction

from scipy.stats import spearmanr

city_trend_results = []

for city, group in city_yearly.groupby("City"):

    group = group.dropna(
        subset=["Year", "Average_AQI"]
    ).sort_values("Year")

    if len(group) >= 3:
        correlation, p_value = spearmanr(
            group["Year"],
            group["Average_AQI"]
        )

        city_trend_results.append({
            "City": city,
            "Available_Years": len(group),
            "Trend_Correlation": correlation,
            "P_Value": p_value
        })

city_trend_df = pd.DataFrame(city_trend_results)

display(
    city_trend_df.sort_values(
        "Trend_Correlation",
        ascending=False
    ).round(4)
)
```

```python
# City-wise Mann–Kendall-style trend testing

import numpy as np
import pandas as pd
from scipy.stats import kendalltau
from statsmodels.stats.multitest import multipletests

trend_test_results = []

for city, group in city_yearly.groupby("City"):

    group = (
        group.dropna(subset=["Year", "Average_AQI"])
        .sort_values("Year")
    )

    if len(group) >= 4:

        tau, p_value = kendalltau(
            group["Year"],
            group["Average_AQI"]
        )

        trend_test_results.append({
            "City": city,
            "Years_Observed": len(group),
            "Kendall_Tau": tau,
            "P_Value": p_value
        })

trend_test_df = pd.DataFrame(trend_test_results)

# Correct for multiple city comparisons
reject, adjusted_p, _, _ = multipletests(
    trend_test_df["P_Value"],
    method="fdr_bh"
)

trend_test_df["Adjusted_P_Value"] = adjusted_p
trend_test_df["Significant"] = reject

trend_test_df["Trend"] = np.select(
    [
        trend_test_df["Kendall_Tau"] > 0,
        trend_test_df["Kendall_Tau"] < 0
    ],
    [
        "Increasing",
        "Decreasing"
    ],
    default="No trend"
)

display(
    trend_test_df.sort_values(
        "Kendall_Tau"
    ).round(4)
)
```

```python
trend_test_df.to_csv(
    os.path.join(
        report_dir,
        "city_wise_kendall_trend_test.csv"
    ),
    index=False
)

print("City-wise trend test report saved.")
```

```python
# ============================================================
# TOPIC: MISSING-VALUE TREATMENT DECISION FRAMEWORK
# PURPOSE: Decide how each column's missing values should be handled.
# METHOD: Evaluate missing percentages and assign treatment rules.
# OUTPUT: A documented missing-value treatment plan.
# ============================================================

import pandas as pd
import numpy as np

# Work from the untouched original dataset
missing_plan = pd.DataFrame({
    "Column": air_raw.columns,
    "Missing_Count": air_raw.isna().sum().values,
    "Missing_Percentage": (
        air_raw.isna().mean().values * 100
    ).round(2)
})

def treatment_decision(row):
    column = row["Column"]
    missing_pct = row["Missing_Percentage"]

    if missing_pct == 0:
        return "Retain as-is"

    elif column in ["AQI", "AQI_Bucket"]:
        return "Retain missing; do not invent AQI values"

    elif missing_pct <= 10:
        return "Evaluate limited imputation with validation"

    elif missing_pct <= 20:
        return "Investigate city/time patterns before imputation"

    elif missing_pct <= 40:
        return "Retain unless a justified method is available"

    else:
        return "Retain missing; avoid blanket imputation"

missing_plan["Treatment_Decision"] = missing_plan.apply(
    treatment_decision,
    axis=1
)

print("MISSING-VALUE TREATMENT PLAN")
display(
    missing_plan.sort_values(
        "Missing_Percentage",
        ascending=False
    ).reset_index(drop=True)
)

# Save the treatment plan
missing_plan.to_csv(
    os.path.join(report_dir, "missing_value_treatment_plan.csv"),
    index=False
)

print("Treatment plan saved successfully.")
```

```python
Overall pollutant outlier detection
We will use the Interquartile Range (IQR) method:
\[
IQR = Q3-Q1
\]

\[
Lower\ Bound=Q1-1.5(IQR)
\]

\[
Upper\ Bound=Q3+1.5(IQR)
\]
```

```python
# ============================================================
# TOPIC: POLLUTANT OUTLIER DETECTION USING IQR
# PURPOSE: Identify unusually low and high pollutant measurements.
# METHOD: Calculate quartiles, IQR bounds and outlier counts.
# OUTPUT: Comprehensive pollutant outlier investigation report.
# ============================================================

import numpy as np
import pandas as pd
import os

pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene",
    "Toluene", "Xylene"
]

outlier_results = []

for column in pollutant_columns:

    values = air_clean[column].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    lower_outliers = (values < lower_bound).sum()
    upper_outliers = (values > upper_bound).sum()

    outlier_results.append({
        "Pollutant": column,
        "Valid_Observations": len(values),
        "Q1": Q1,
        "Median": values.median(),
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Lower_Outliers": lower_outliers,
        "Upper_Outliers": upper_outliers,
        "Total_Outliers": lower_outliers + upper_outliers,
        "Outlier_Percentage": (
            (lower_outliers + upper_outliers)
            / len(values) * 100
        ) if len(values) else np.nan
    })

outlier_report = pd.DataFrame(outlier_results)

print("POLLUTANT OUTLIER DETECTION REPORT")
display(
    outlier_report.sort_values(
        "Outlier_Percentage",
        ascending=False
    ).round(3)
)

outlier_report.to_csv(
    os.path.join(report_dir, "pollutant_iqr_outlier_report.csv"),
    index=False
)

print("Outlier report saved successfully.")
```

```python
# ============================================================
# TOPIC: POLLUTANT OUTLIER VISUALIZATION
# PURPOSE: Compare pollutant distributions and extreme values.
# METHOD: Generate boxplots for all 12 pollutant variables.
# OUTPUT: Saved multi-panel pollutant boxplot visualization.
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(4, 3, figsize=(18, 18))

for ax, column in zip(axes.flatten(), pollutant_columns):

    sns.boxplot(
        y=air_clean[column],
        ax=ax
    )

    ax.set_title(f"{column} Distribution")
    ax.set_ylabel("Concentration")

plt.suptitle(
    "Outlier Investigation Across Air Pollutants",
    fontsize=18,
    y=1.01
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "pollutant_outlier_boxplots.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: CITY-WISE POLLUTANT OUTLIER INVESTIGATION
# PURPOSE: Identify unusual measurements within individual cities.
# METHOD: Apply IQR limits separately to each city and pollutant.
# OUTPUT: City-wise outlier counts and percentages.
# ============================================================

city_outlier_results = []

for city, group in air_clean.groupby("City"):

    for column in pollutant_columns:

        values = group[column].dropna()

        if len(values) < 10:
            continue

        Q1 = values.quantile(0.25)
        Q3 = values.quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        count = (
            (values < lower_bound) |
            (values > upper_bound)
        ).sum()

        city_outlier_results.append({
            "City": city,
            "Pollutant": column,
            "Valid_Observations": len(values),
            "Outlier_Count": count,
            "Outlier_Percentage": count / len(values) * 100
        })

city_outlier_report = pd.DataFrame(city_outlier_results)

print("CITY-WISE OUTLIER REPORT")
display(
    city_outlier_report.sort_values(
        "Outlier_Percentage",
        ascending=False
    ).head(20).round(2)
)

city_outlier_report.to_csv(
    os.path.join(report_dir, "city_wise_pollutant_outliers.csv"),
    index=False
)

print("City-wise outlier report saved.")
```

```python
# ============================================================
# TOPIC: POLLUTANT VALUE CONSISTENCY VALIDATION
# PURPOSE: Identify invalid or suspicious pollutant measurements.
# METHOD: Check negative, zero, infinite and missing values.
# OUTPUT: Pollutant-level consistency report.
# ============================================================

import numpy as np
import pandas as pd
import os

pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene",
    "Toluene", "Xylene"
]

consistency_results = []

for column in pollutant_columns:

    values = air_clean[column]

    consistency_results.append({
        "Pollutant": column,
        "Total_Records": len(values),
        "Missing_Values": values.isna().sum(),
        "Negative_Values": (values < 0).sum(),
        "Zero_Values": (values == 0).sum(),
        "Infinite_Values": np.isinf(values).sum(),
        "Valid_Values": values.notna().sum()
    })

pollutant_consistency = pd.DataFrame(consistency_results)

print("POLLUTANT CONSISTENCY REPORT")
display(pollutant_consistency)

pollutant_consistency.to_csv(
    os.path.join(report_dir, "pollutant_consistency_report.csv"),
    index=False
)
```

```python
# ============================================================
# TOPIC: AQI AND AQI CATEGORY VALIDATION
# PURPOSE: Check the integrity of AQI measurements and labels.
# METHOD: Inspect numeric ranges and category distributions.
# OUTPUT: AQI validation report and category frequency table.
# ============================================================

# AQI numerical validation
aqi_values = air_clean["AQI"]

aqi_validation = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Missing AQI",
        "Negative AQI",
        "Zero AQI",
        "Minimum AQI",
        "Maximum AQI",
        "Unique AQI Values"
    ],
    "Value": [
        len(aqi_values),
        aqi_values.isna().sum(),
        (aqi_values < 0).sum(),
        (aqi_values == 0).sum(),
        aqi_values.min(),
        aqi_values.max(),
        aqi_values.nunique()
    ]
})

print("AQI VALIDATION REPORT")
display(aqi_validation)

# AQI category validation
print("\nAQI CATEGORY DISTRIBUTION")

aqi_category_distribution = (
    air_clean["AQI_Bucket"]
    .value_counts(dropna=False)
    .rename_axis("AQI_Category")
    .reset_index(name="Record_Count")
)

display(aqi_category_distribution)

aqi_category_distribution.to_csv(
    os.path.join(report_dir, "aqi_category_validation.csv"),
    index=False
)
```

```python
# ============================================================
# TOPIC: AQI CATEGORY LOGICAL CONSISTENCY
# PURPOSE: Compare existing AQI labels against AQI numerical bands.
# METHOD: Apply CPCB category thresholds and identify mismatches.
# OUTPUT: AQI category consistency report.
# ============================================================

def classify_aqi(value):

    if pd.isna(value):
        return np.nan

    if value < 0 or value > 500:
        return "Outside defined range"

    elif value <= 50:
        return "Good"

    elif value <= 100:
        return "Satisfactory"

    elif value <= 200:
        return "Moderate"

    elif value <= 300:
        return "Poor"

    elif value <= 400:
        return "Very Poor"

    else:
        return "Severe"


aqi_check = air_clean[
    ["City", "Date", "AQI", "AQI_Bucket"]
].copy()

aqi_check["Calculated_Category"] = (
    aqi_check["AQI"].apply(classify_aqi)
)

aqi_check["Category_Mismatch"] = (
    aqi_check["AQI_Bucket"].notna()
    & aqi_check["Calculated_Category"].notna()
    & (aqi_check["AQI_Bucket"] != aqi_check["Calculated_Category"])
)

print("CATEGORY MISMATCH COUNT:",
      aqi_check["Category_Mismatch"].sum())

print("\nCATEGORY MISMATCH EXAMPLES")

display(
    aqi_check.loc[
        aqi_check["Category_Mismatch"]
    ].head(15)
)

aqi_check.to_csv(
    os.path.join(report_dir, "aqi_category_consistency.csv"),
    index=False
)
```

```python
# ============================================================
# TOPIC: AQI CATEGORY MISMATCH DIAGNOSTICS
# PURPOSE: Quantify and investigate AQI values above 500.
# METHOD: Compare original labels, calculated labels, cities
#         and years.
# OUTPUT: Detailed AQI mismatch investigation reports.
# ============================================================

# 1. Count mismatches
print("TOTAL CATEGORY MISMATCHES:",
      aqi_check["Category_Mismatch"].sum())

# 2. Original versus calculated category
print("\nORIGINAL VS CALCULATED CATEGORY")

display(
    pd.crosstab(
        aqi_check["AQI_Bucket"],
        aqi_check["Calculated_Category"],
        dropna=False
    )
)

# 3. Distribution by city
print("\nMISMATCHES BY CITY")

mismatch_city = (
    aqi_check[aqi_check["Category_Mismatch"]]
    .groupby("City")
    .agg(
        Mismatch_Count=("AQI", "count"),
        Maximum_AQI=("AQI", "max")
    )
    .sort_values("Mismatch_Count", ascending=False)
)

display(mismatch_city)

# 4. Distribution by year
mismatch_year_data = aqi_check[
    aqi_check["Category_Mismatch"]
].copy()

mismatch_year_data["Year"] = (
    mismatch_year_data["Date"].dt.year
)

mismatch_year = (
    mismatch_year_data.groupby("Year")
    .agg(
        Mismatch_Count=("AQI", "count"),
        Maximum_AQI=("AQI", "max")
    )
)

print("\nMISMATCHES BY YEAR")
display(mismatch_year)

# 5. Save diagnostic reports
mismatch_city.to_csv(
    os.path.join(report_dir, "aqi_mismatch_by_city.csv")
)

mismatch_year.to_csv(
    os.path.join(report_dir, "aqi_mismatch_by_year.csv")
)
```

```python
# ============================================================
# TOPIC: AQI CATEGORY MISMATCH DIAGNOSTICS
# PURPOSE: Quantify and investigate AQI values above 500.
# METHOD: Compare original labels, calculated labels, cities
#         and years.
# OUTPUT: Detailed AQI mismatch investigation reports.
# ============================================================

# 1. Count mismatches
print("TOTAL CATEGORY MISMATCHES:",
      aqi_check["Category_Mismatch"].sum())

# 2. Original versus calculated category
print("\nORIGINAL VS CALCULATED CATEGORY")

display(
    pd.crosstab(
        aqi_check["AQI_Bucket"],
        aqi_check["Calculated_Category"],
        dropna=False
    )
)

# 3. Distribution by city
print("\nMISMATCHES BY CITY")

mismatch_city = (
    aqi_check[aqi_check["Category_Mismatch"]]
    .groupby("City")
    .agg(
        Mismatch_Count=("AQI", "count"),
        Maximum_AQI=("AQI", "max")
    )
    .sort_values("Mismatch_Count", ascending=False)
)

display(mismatch_city)

# 4. Distribution by year
mismatch_year_data = aqi_check[
    aqi_check["Category_Mismatch"]
].copy()

mismatch_year_data["Year"] = (
    mismatch_year_data["Date"].dt.year
)

mismatch_year = (
    mismatch_year_data.groupby("Year")
    .agg(
        Mismatch_Count=("AQI", "count"),
        Maximum_AQI=("AQI", "max")
    )
)

print("\nMISMATCHES BY YEAR")
display(mismatch_year)

# 5. Save diagnostic reports
mismatch_city.to_csv(
    os.path.join(report_dir, "aqi_mismatch_by_city.csv")
)

mismatch_year.to_csv(
    os.path.join(report_dir, "aqi_mismatch_by_year.csv")
)
```

```python
# ============================================================
# TOPIC: ROOT-CAUSE ANALYSIS OF AQI CATEGORY MISMATCHES
# PURPOSE: Determine why the 543 AQI category mismatches occur.
# METHOD: Compare AQI ranges, original labels and calculated labels.
# OUTPUT: Root-cause summary and AQI range distribution.
# ============================================================

# 1. Extract mismatch records
mismatch_records = aqi_check[
    aqi_check["Category_Mismatch"]
].copy()

# 2. Check AQI values above 500
above_500 = mismatch_records[
    mismatch_records["AQI"] > 500
]

print("ROOT-CAUSE SUMMARY")

print("Total mismatches:", len(mismatch_records))
print("Mismatches above 500:", len(above_500))
print(
    "Mismatches at or below 500:",
    len(mismatch_records) - len(above_500)
)

# 3. Original category distribution
print("\nORIGINAL CATEGORY DISTRIBUTION")

display(
    mismatch_records["AQI_Bucket"]
    .value_counts(dropna=False)
)

# 4. AQI range distribution
print("\nAQI RANGE DISTRIBUTION")

aqi_ranges = pd.cut(
    mismatch_records["AQI"],
    bins=[0, 500, 1000, 1500, 2000, float("inf")],
    labels=[
        "0-500",
        "501-1000",
        "1001-1500",
        "1501-2000",
        "Above 2000"
    ]
)

display(
    aqi_ranges.value_counts()
    .sort_index()
    .rename_axis("AQI_Range")
    .reset_index(name="Record_Count")
)

# 5. Save root-cause report
mismatch_records.to_csv(
    os.path.join(report_dir, "aqi_mismatch_root_cause_records.csv"),
    index=False
)

print("\nRoot-cause records saved.")
```

```python
# ============================================================
# TOPIC: AQI VALIDATION FINDINGS DOCUMENTATION
# PURPOSE: Record the confirmed AQI category mismatch findings.
# METHOD: Summarize the diagnostic results without modifying data.
# OUTPUT: Saved AQI validation findings report.
# ============================================================

aqi_findings = pd.DataFrame({
    "Finding": [
        "Total category mismatches",
        "Mismatches above AQI 500",
        "Mismatches at or below AQI 500",
        "Original category of mismatches",
        "Maximum AQI observed",
        "AQI validation threshold",
        "Treatment decision"
    ],
    "Result": [
        543,
        543,
        0,
        "Severe",
        2049,
        500,
        "Preserve and flag; do not automatically modify"
    ]
})

display(aqi_findings)

aqi_findings.to_csv(
    os.path.join(report_dir, "aqi_validation_findings.csv"),
    index=False
)

print("AQI validation findings saved successfully.")
```

```python
# ============================================================
# TOPIC: FINAL CITY-DATE DUPLICATE VALIDATION
# PURPOSE: Confirm whether each city-date combination is unique.
# METHOD: Identify duplicate city-date keys and inspect records.
# OUTPUT: City-date validation report.
# ============================================================

city_date_duplicates = air_clean.duplicated(
    subset=["City", "Date"],
    keep=False
)

print("CITY-DATE VALIDATION")
print("Total records:", len(air_clean))
print("Duplicate city-date records:", city_date_duplicates.sum())
print(
    "Unique city-date combinations:",
    air_clean[["City", "Date"]].drop_duplicates().shape[0]
)

if city_date_duplicates.any():
    print("\nDuplicate examples:")
    display(
        air_clean.loc[city_date_duplicates]
        .sort_values(["City", "Date"])
        .head(20)
    )
else:
    print("No duplicate city-date combinations found.")

city_date_report.to_csv(
    os.path.join(report_dir, "city_date_validation_report.csv"),
    index=False
)
```

```python
# ============================================================
# TOPIC: CITY-DATE VALIDATION REPORT EXPORT
# PURPOSE: Create and save the final city-date validation report.
# METHOD: Summarize duplicate and unique city-date combinations.
# OUTPUT: city_date_validation_report.csv
# ============================================================

city_date_report = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Duplicate City-Date Records",
        "Unique City-Date Combinations",
        "Unique Cities",
        "Unique Dates"
    ],
    "Value": [
        len(air_clean),
        air_clean.duplicated(
            subset=["City", "Date"]
        ).sum(),
        air_clean[["City", "Date"]].drop_duplicates().shape[0],
        air_clean["City"].nunique(),
        air_clean["Date"].nunique()
    ]
})

display(city_date_report)

city_date_report.to_csv(
    os.path.join(
        report_dir,
        "city_date_validation_report.csv"
    ),
    index=False
)

print("City-date validation report saved successfully.")
```

```python
# ============================================================
# TOPIC: FINAL ANALYTICAL DATASET PREPARATION
# PURPOSE: Prepare a validated dataset for advanced analysis.
# METHOD: Preserve original measurements and add quality flags.
# OUTPUT: air_quality_final.csv
# ============================================================

import pandas as pd
import numpy as np
import os

# Start from the cleaned working dataset
air_final = air_clean.copy()

# 1. Flag AQI values outside the CPCB 0-500 range
air_final["AQI_Above_500_Flag"] = (
    air_final["AQI"] > 500
)

# 2. Flag negative pollutant measurements
pollutant_columns = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene",
    "Toluene", "Xylene"
]

air_final["Negative_Pollutant_Flag"] = (
    air_final[pollutant_columns].lt(0).any(axis=1)
)

# 3. Flag infinite pollutant measurements
air_final["Infinite_Pollutant_Flag"] = (
    np.isinf(air_final[pollutant_columns])
    .any(axis=1)
)

# 4. Flag missing AQI
air_final["Missing_AQI_Flag"] = (
    air_final["AQI"].isna()
)

# 5. Flag missing pollutant measurements
air_final["Missing_Pollutant_Flag"] = (
    air_final[pollutant_columns].isna().any(axis=1)
)

# Save final analytical dataset
final_dataset_path = os.path.join(
    processed_dir,
    "air_quality_final.csv"
)

air_final.to_csv(
    final_dataset_path,
    index=False
)

print("FINAL ANALYTICAL DATASET CREATED")
print("Dataset shape:", air_final.shape)
print("Saved location:", final_dataset_path)
```

```python
# ============================================================
# TOPIC: FINAL DATASET QUALITY SUMMARY
# PURPOSE: Summarize the quality of the final analytical dataset.
# METHOD: Calculate structural, missingness and anomaly indicators.
# OUTPUT: Final data-quality summary CSV.
# ============================================================

final_quality_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Columns",
        "Exact Duplicate Rows",
        "Duplicate City-Date Records",
        "Missing Cells",
        "Missing AQI Records",
        "AQI Above 500",
        "Rows With Negative Pollutants",
        "Rows With Infinite Pollutants",
        "Rows With Missing Pollutants",
        "Unique Cities"
    ],
    "Value": [
        len(air_final),
        air_final.shape[1],
        air_final.duplicated().sum(),
        air_final.duplicated(
            subset=["City", "Date"]
        ).sum(),
        air_final.isna().sum().sum(),
        air_final["AQI"].isna().sum(),
        air_final["AQI_Above_500_Flag"].sum(),
        air_final["Negative_Pollutant_Flag"].sum(),
        air_final["Infinite_Pollutant_Flag"].sum(),
        air_final["Missing_Pollutant_Flag"].sum(),
        air_final["City"].nunique()
    ]
})

display(final_quality_summary)

final_quality_summary.to_csv(
    os.path.join(
        report_dir,
        "final_data_quality_summary.csv"
    ),
    index=False
)

print("Final quality summary saved.")
```

```python
# ============================================================
# TOPIC: FINAL DATASET RELOAD VERIFICATION
# PURPOSE: Confirm that the exported dataset is readable.
# METHOD: Reload the CSV and compare its dimensions and columns.
# OUTPUT: Export verification result.
# ============================================================

air_final_check = pd.read_csv(final_dataset_path)

print("ORIGINAL FINAL DATASET SHAPE:", air_final.shape)
print("RELOADED DATASET SHAPE:", air_final_check.shape)

print(
    "COLUMN STRUCTURE MATCH:",
    air_final.columns.tolist() == air_final_check.columns.tolist()
)

print(
    "EXPORTED FILE EXISTS:",
    os.path.isfile(final_dataset_path)
)
```

```python

# ============================================================
# TOPIC: POLLUTION EPISODE IDENTIFICATION
# PURPOSE: Identify consecutive high-pollution periods by city.
# METHOD: Apply AQI > 200 threshold and group consecutive dates.
# OUTPUT: City-wise pollution episode records.
# ============================================================

import pandas as pd
import numpy as np
import os

# Work from the final analytical dataset
episode_data = air_final[
    ["City", "Date", "AQI", "AQI_Bucket"]
].copy()

episode_data["Date"] = pd.to_datetime(
    episode_data["Date"],
    errors="coerce"
)

episode_data = episode_data.dropna(
    subset=["City", "Date", "AQI"]
).copy()

# Initial pollution episode threshold
episode_threshold = 200

episode_data["High_Pollution_Day"] = (
    episode_data["AQI"] > episode_threshold
)

# Sort observations by city and date
episode_data = episode_data.sort_values(
    ["City", "Date"]
).reset_index(drop=True)

# Identify consecutive high-pollution runs separately by city
episode_data["Date_Gap"] = (
    episode_data.groupby("City")["Date"].diff().dt.days
)

episode_data["Previous_High_Day"] = (
    episode_data.groupby("City")["High_Pollution_Day"]
    .shift()
    .fillna(False)
    .astype(bool)
)

episode_data["New_Episode"] = (
    episode_data["High_Pollution_Day"]
    & (
        (~episode_data["Previous_High_Day"])
        | (episode_data["Date_Gap"] != 1)
    )
)

episode_data["Episode_ID"] = (
    episode_data.groupby("City")["New_Episode"]
    .cumsum()
)

# Extract only high-pollution observations
high_pollution_records = episode_data[
    episode_data["High_Pollution_Day"]
].copy()

# Summarize each episode
pollution_episodes = (
    high_pollution_records
    .groupby(["City", "Episode_ID"])
    .agg(
        Start_Date=("Date", "min"),
        End_Date=("Date", "max"),
        Duration_Days=("Date", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        Minimum_AQI=("AQI", "min")
    )
    .reset_index()
)

# Identify episodes containing AQI values above 500
pollution_episodes["Contains_AQI_Above_500"] = (
    high_pollution_records
    .groupby(["City", "Episode_ID"])["AQI"]
    .max()
    .reset_index(drop=True) > 500
)

print("POLLUTION EPISODE IDENTIFICATION")
print("Threshold: AQI > 200")
print("Total pollution episodes:", len(pollution_episodes))

display(
    pollution_episodes.sort_values(
        "Maximum_AQI",
        ascending=False
    ).head(20)
)

pollution_episodes.to_csv(
    os.path.join(
        processed_dir,
        "pollution_episode_records.csv"
    ),
    index=False
)

print("Pollution episode records saved.")
```

```python
# ============================================================
# TOPIC: CITY-WISE POLLUTION EPISODE PERFORMANCE
# PURPOSE: Compare pollution-event frequency and intensity by city.
# METHOD: Aggregate episode records into city-level indicators.
# OUTPUT: City-wise pollution risk comparison report.
# ============================================================

city_episode_summary = (
    pollution_episodes
    .groupby("City")
    .agg(
        Total_Episodes=("Episode_ID", "count"),
        Total_Pollution_Days=("Duration_Days", "sum"),
        Average_Episode_Duration=("Duration_Days", "mean"),
        Longest_Episode=("Duration_Days", "max"),
        Average_Episode_AQI=("Average_AQI", "mean"),
        Maximum_Recorded_AQI=("Maximum_AQI", "max"),
        Episodes_With_AQI_Above_500=(
            "Contains_AQI_Above_500", "sum"
        )
    )
    .reset_index()
)

city_episode_summary = city_episode_summary.sort_values(
    "Total_Episodes",
    ascending=False
)

print("CITY-WISE POLLUTION EPISODE SUMMARY")
display(city_episode_summary.round(2))

city_episode_summary.to_csv(
    os.path.join(
        report_dir,
        "city_wise_pollution_episode_summary.csv"
    ),
    index=False
)

print("City-wise episode summary saved.")
```

```python
# ============================================================
# TOPIC: CITY-WISE POLLUTION EPISODE VISUALIZATION
# PURPOSE: Compare the frequency of high-pollution episodes.
# METHOD: Rank cities by their number of identified episodes.
# OUTPUT: Saved city-wise pollution episode chart.
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

plot_data = city_episode_summary.sort_values(
    "Total_Episodes",
    ascending=True
)

plt.figure(figsize=(12, 10))

sns.barplot(
    data=plot_data,
    x="Total_Episodes",
    y="City"
)

plt.title(
    "City-wise Frequency of High-Pollution Episodes",
    fontsize=16
)

plt.xlabel("Number of Pollution Episodes")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_wise_pollution_episode_frequency.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: POLLUTION EPISODE DURATION CLASSIFICATION
# PURPOSE: Classify pollution episodes by their duration.
# METHOD: Categorize consecutive AQI > 200 observations.
# OUTPUT: Duration-classified pollution episode dataset.
# ============================================================

import pandas as pd
import numpy as np
import os

episode_duration = pollution_episodes.copy()

def classify_duration(days):
    if days == 1:
        return "Isolated Event"
    elif days <= 3:
        return "Short Episode"
    elif days <= 7:
        return "Medium Episode"
    elif days <= 14:
        return "Long Episode"
    else:
        return "Prolonged Episode"

episode_duration["Duration_Category"] = (
    episode_duration["Duration_Days"]
    .apply(classify_duration)
)

duration_summary = (
    episode_duration.groupby("Duration_Category")
    .agg(
        Episode_Count=("Episode_ID", "count"),
        Average_Duration=("Duration_Days", "mean"),
        Maximum_Duration=("Duration_Days", "max"),
        Average_AQI=("Average_AQI", "mean"),
        Maximum_AQI=("Maximum_AQI", "max")
    )
    .reset_index()
)

duration_order = [
    "Isolated Event",
    "Short Episode",
    "Medium Episode",
    "Long Episode",
    "Prolonged Episode"
]

duration_summary["Duration_Category"] = pd.Categorical(
    duration_summary["Duration_Category"],
    categories=duration_order,
    ordered=True
)

duration_summary = duration_summary.sort_values(
    "Duration_Category"
)

print("POLLUTION EPISODE DURATION SUMMARY")
display(duration_summary.round(2))

episode_duration.to_csv(
    os.path.join(
        processed_dir,
        "pollution_episodes_with_duration_category.csv"
    ),
    index=False
)

duration_summary.to_csv(
    os.path.join(
        report_dir,
        "pollution_episode_duration_summary.csv"
    ),
    index=False
)

print("Duration classification reports saved.")
```

```python
# ============================================================
# TOPIC: MONTHLY POLLUTION EPISODE ANALYSIS
# PURPOSE: Identify months with the greatest pollution-event burden.
# METHOD: Aggregate high-pollution days and episode starts by month.
# OUTPUT: Monthly pollution-event summary.
# ============================================================

# Extract monthly information from daily high-pollution records
monthly_events = high_pollution_records.copy()

monthly_events["Year"] = monthly_events["Date"].dt.year
monthly_events["Month"] = monthly_events["Date"].dt.month
monthly_events["Month_Name"] = monthly_events["Date"].dt.month_name()

# Monthly high-pollution days
monthly_day_summary = (
    monthly_events.groupby(["Month", "Month_Name"])
    .agg(
        High_Pollution_Days=("Date", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        Cities_Affected=("City", "nunique")
    )
    .reset_index()
)

# Monthly episode starts
episode_starts = episode_duration.copy()

episode_starts["Month"] = (
    pd.to_datetime(episode_starts["Start_Date"]).dt.month
)

monthly_start_summary = (
    episode_starts.groupby("Month")
    .agg(
        Episode_Starts=("Episode_ID", "count")
    )
    .reset_index()
)

monthly_event_summary = monthly_day_summary.merge(
    monthly_start_summary,
    on="Month",
    how="left"
)

monthly_event_summary = monthly_event_summary.sort_values(
    "Month"
)

print("MONTHLY POLLUTION EVENT SUMMARY")
display(monthly_event_summary.round(2))

monthly_event_summary.to_csv(
    os.path.join(
        report_dir,
        "monthly_pollution_event_summary.csv"
    ),
    index=False
)

print("Monthly event summary saved.")
```

```python
# ============================================================
# TOPIC: SEASONAL POLLUTION EPISODE ANALYSIS
# PURPOSE: Compare pollution-event burden across four seasons.
# METHOD: Assign seasons using calendar months and aggregate
#         event frequency, duration and AQI intensity.
# OUTPUT: Seasonal pollution-event comparison report.
# ============================================================

def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Pre-Monsoon"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

# Seasonal daily pollution records
seasonal_events = high_pollution_records.copy()

seasonal_events["Season"] = (
    seasonal_events["Date"].dt.month.apply(assign_season)
)

seasonal_day_summary = (
    seasonal_events.groupby("Season")
    .agg(
        High_Pollution_Days=("Date", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        Cities_Affected=("City", "nunique")
    )
    .reset_index()
)

# Assign season based on episode start date
seasonal_episode_data = episode_duration.copy()

seasonal_episode_data["Season"] = (
    pd.to_datetime(
        seasonal_episode_data["Start_Date"]
    ).dt.month.apply(assign_season)
)

seasonal_episode_summary = (
    seasonal_episode_data.groupby("Season")
    .agg(
        Episode_Count=("Episode_ID", "count"),
        Average_Duration=("Duration_Days", "mean"),
        Longest_Duration=("Duration_Days", "max"),
        Average_Episode_AQI=("Average_AQI", "mean")
    )
    .reset_index()
)

seasonal_event_summary = seasonal_day_summary.merge(
    seasonal_episode_summary,
    on="Season",
    how="outer"
)

season_order = [
    "Winter",
    "Pre-Monsoon",
    "Monsoon",
    "Post-Monsoon"
]

seasonal_event_summary["Season"] = pd.Categorical(
    seasonal_event_summary["Season"],
    categories=season_order,
    ordered=True
)

seasonal_event_summary = seasonal_event_summary.sort_values(
    "Season"
)

print("SEASONAL POLLUTION EVENT SUMMARY")
display(seasonal_event_summary.round(2))

seasonal_event_summary.to_csv(
    os.path.join(
        report_dir,
        "seasonal_pollution_event_summary.csv"
    ),
    index=False
)

print("Seasonal event summary saved.")
```

```python
# ============================================================
# TOPIC: YEAR-WISE POLLUTION EVENT TREND ANALYSIS
# PURPOSE: Examine annual changes in pollution-event frequency.
# METHOD: Aggregate high-pollution days and episodes by year.
# OUTPUT: Annual pollution-event trend report.
# ============================================================

annual_events = high_pollution_records.copy()

annual_events["Year"] = annual_events["Date"].dt.year

annual_day_summary = (
    annual_events.groupby("Year")
    .agg(
        High_Pollution_Days=("Date", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        Cities_Affected=("City", "nunique")
    )
    .reset_index()
)

annual_episode_data = episode_duration.copy()

annual_episode_data["Year"] = (
    pd.to_datetime(annual_episode_data["Start_Date"]).dt.year
)

annual_episode_summary = (
    annual_episode_data.groupby("Year")
    .agg(
        Episode_Count=("Episode_ID", "count"),
        Average_Episode_Duration=("Duration_Days", "mean"),
        Longest_Episode=("Duration_Days", "max")
    )
    .reset_index()
)

annual_event_summary = annual_day_summary.merge(
    annual_episode_summary,
    on="Year",
    how="outer"
).sort_values("Year")

print("ANNUAL POLLUTION EVENT SUMMARY")
display(annual_event_summary.round(2))

annual_event_summary.to_csv(
    os.path.join(
        report_dir,
        "annual_pollution_event_summary.csv"
    ),
    index=False
)

print("Annual event summary saved.")
```

```python
City-wise pollution exposure analysis
We will calculate the percentage of valid AQI observation days exceeding 200 for each city.
\[
\text{High Pollution Rate} =
\frac{\text{Days with AQI}>200}
{\text{Valid AQI observation days}}\times100
\]
```

```python
# ============================================================
# TOPIC: CITY-WISE POLLUTION EXPOSURE ANALYSIS
# PURPOSE: Measure the proportion of observed days with AQI > 200.
# METHOD: Compare high-pollution days against valid AQI days
#         separately for each city.
# OUTPUT: City-wise pollution exposure report.
# ============================================================

# Total valid AQI observation days by city
city_observations = (
    episode_data.groupby("City")
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        AQI_Above_500_Days=("AQI", lambda x: (x > 500).sum())
    )
    .reset_index()
)

# High-pollution days by city
city_high_days = (
    high_pollution_records.groupby("City")
    .agg(
        High_Pollution_Days=("AQI", "count")
    )
    .reset_index()
)

# Combine observation and pollution counts
city_risk_profile = city_observations.merge(
    city_high_days,
    on="City",
    how="left"
)

city_risk_profile["High_Pollution_Days"] = (
    city_risk_profile["High_Pollution_Days"].fillna(0)
)

# Calculate normalized exposure percentage
city_risk_profile["High_Pollution_Rate_Percent"] = (
    city_risk_profile["High_Pollution_Days"]
    / city_risk_profile["Valid_AQI_Days"] * 100
)

# Add episode information
city_risk_profile = city_risk_profile.merge(
    city_episode_summary[
        [
            "City",
            "Total_Episodes",
            "Average_Episode_Duration",
            "Longest_Episode",
            "Episodes_With_AQI_Above_500"
        ]
    ],
    on="City",
    how="left"
)

city_risk_profile = city_risk_profile.fillna({
    "Total_Episodes": 0,
    "Average_Episode_Duration": 0,
    "Longest_Episode": 0,
    "Episodes_With_AQI_Above_500": 0
})

city_risk_profile = city_risk_profile.sort_values(
    "High_Pollution_Rate_Percent",
    ascending=False
).reset_index(drop=True)

print("CITY-WISE POLLUTION RISK PROFILE")

display(city_risk_profile.round(2))

city_risk_profile.to_csv(
    os.path.join(
        report_dir,
        "city_wise_pollution_risk_profile.csv"
    ),
    index=False
)

print("City-wise pollution risk report saved.")
```

```python
# ============================================================
# TOPIC: CITY-WISE POLLUTION HOTSPOT CLASSIFICATION
# PURPOSE: Categorize cities by their observed high-pollution rate.
# METHOD: Apply project-defined descriptive percentage bands.
# OUTPUT: City hotspot classification report.
# ============================================================

def classify_city_risk(rate):

    if rate < 10:
        return "Low"

    elif rate < 25:
        return "Moderate"

    elif rate < 50:
        return "High"

    else:
        return "Very High"


city_risk_profile["Pollution_Risk_Class"] = (
    city_risk_profile["High_Pollution_Rate_Percent"]
    .apply(classify_city_risk)
)

hotspot_summary = (
    city_risk_profile.groupby("Pollution_Risk_Class")
    .agg(
        City_Count=("City", "count"),
        Average_Exposure_Rate=(
            "High_Pollution_Rate_Percent", "mean"
        )
    )
    .reset_index()
)

print("POLLUTION HOTSPOT CLASSIFICATION")

display(
    city_risk_profile[
        [
            "City",
            "Valid_AQI_Days",
            "High_Pollution_Days",
            "High_Pollution_Rate_Percent",
            "Pollution_Risk_Class"
        ]
    ].round(2)
)

print("\nHOTSPOT CLASS SUMMARY")
display(hotspot_summary.round(2))

city_risk_profile.to_csv(
    os.path.join(
        report_dir,
        "city_pollution_hotspot_classification.csv"
    ),
    index=False
)
```

```python
Calculate city-season pollution exposure rates
For each city, we calculate:
\[
\text{Exposure Rate} =
\frac{\text{Days with AQI}>200}
{\text{Valid AQI days}}\times100
\]

We will compare Winter, Pre-Monsoon, Monsoon and Post-Monsoon.
```

```python
# ============================================================
# TOPIC: CITY-SEASON POLLUTION EXPOSURE ANALYSIS
# PURPOSE: Calculate comparable seasonal pollution rates by city.
# METHOD: Calculate high-pollution days / valid AQI days.
# OUTPUT: City-season exposure table and CSV report.
# ============================================================

from scipy.stats import friedmanchisquare, wilcoxon
from statsmodels.stats.multitest import multipletests
import pandas as pd
import numpy as np
import os

season_order = [
    "Winter",
    "Pre-Monsoon",
    "Monsoon",
    "Post-Monsoon"
]

season_data = episode_data.copy()

def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Pre-Monsoon"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    return "Post-Monsoon"

season_data["Season"] = (
    season_data["Date"].dt.month.map(assign_season)
)

city_season = (
    season_data.groupby(["City", "Season"])
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        High_Pollution_Days=("High_Pollution_Day", "sum")
    )
    .reset_index()
)

city_season["Exposure_Rate_Percent"] = (
    city_season["High_Pollution_Days"]
    / city_season["Valid_AQI_Days"] * 100
)

city_season.to_csv(
    os.path.join(report_dir, "city_season_exposure_rates.csv"),
    index=False
)

print("CITY-SEASON EXPOSURE RATES")
display(
    city_season.sort_values(
        ["City", "Season"]
    ).round(2)
)

# Create one row per city and one column per season
season_pivot = city_season.pivot(
    index="City",
    columns="Season",
    values="Exposure_Rate_Percent"
)

season_pivot = season_pivot.reindex(columns=season_order)

# Use only cities with all four seasons represented
complete_season_pivot = season_pivot.dropna()

print("\nComplete cities available for paired testing:",
      len(complete_season_pivot))

display(complete_season_pivot.round(2))
```

```python
# ============================================================
# TOPIC: FRIEDMAN TEST FOR SEASONAL POLLUTION EXPOSURE
# PURPOSE: Test whether seasonal exposure rates differ across cities.
# METHOD: Apply the Friedman repeated-measures non-parametric test.
# OUTPUT: Test statistic, p-value and interpretation.
# ============================================================

if len(complete_season_pivot) < 3:
    raise ValueError(
        "Insufficient complete cities for the Friedman test."
    )

friedman_stat, friedman_p = friedmanchisquare(
    *[
        complete_season_pivot[season].values
        for season in season_order
    ]
)

friedman_result = pd.DataFrame({
    "Test": ["Friedman Test"],
    "Cities_Compared": [len(complete_season_pivot)],
    "Seasons": [", ".join(season_order)],
    "Statistic": [friedman_stat],
    "P_Value": [friedman_p],
    "Significant_At_0_05": [friedman_p < 0.05]
})

display(friedman_result)

friedman_result.to_csv(
    os.path.join(report_dir, "seasonal_friedman_test.csv"),
    index=False
)

print(
    "Interpretation:",
    "Evidence of seasonal differences."
    if friedman_p < 0.05
    else "Insufficient evidence of seasonal differences."
)
```

```python
# ============================================================
# TOPIC: PAIRWISE SEASONAL WILCOXON TESTS
# PURPOSE: Identify which seasonal exposure rates differ.
# METHOD: Paired Wilcoxon tests with FDR correction.
# OUTPUT: Pairwise statistics, adjusted p-values and effect sizes.
# ============================================================

from itertools import combinations
from scipy.stats import rankdata

pairwise_results = []

for season_a, season_b in combinations(season_order, 2):

    x = complete_season_pivot[season_a].values
    y = complete_season_pivot[season_b].values

    differences = x - y
    nonzero = differences != 0

    if nonzero.sum() == 0:
        statistic = 0.0
        p_value = 1.0
        effect_size = 0.0
    else:
        statistic, p_value = wilcoxon(
            x,
            y,
            alternative="two-sided",
            zero_method="wilcox"
        )

        # Matched-pairs rank-biserial correlation
        ranks = rankdata(np.abs(differences[nonzero]))

        positive_rank_sum = ranks[
            differences[nonzero] > 0
        ].sum()

        negative_rank_sum = ranks[
            differences[nonzero] < 0
        ].sum()

        total_rank_sum = positive_rank_sum + negative_rank_sum

        effect_size = (
            (positive_rank_sum - negative_rank_sum)
            / total_rank_sum
            if total_rank_sum > 0 else 0.0
        )

    pairwise_results.append({
        "Season_A": season_a,
        "Season_B": season_b,
        "Cities_Compared": len(x),
        "Median_Difference_A_minus_B": np.median(differences),
        "Wilcoxon_Statistic": statistic,
        "P_Value": p_value,
        "Rank_Biserial_Correlation": effect_size
    })

pairwise_season_results = pd.DataFrame(pairwise_results)

# Correct for six pairwise comparisons
reject, adjusted_p, _, _ = multipletests(
    pairwise_season_results["P_Value"],
    method="fdr_bh"
)

pairwise_season_results["Adjusted_P_Value"] = adjusted_p
pairwise_season_results["Significant_After_FDR"] = reject

print("PAIRWISE SEASONAL COMPARISONS")

display(
    pairwise_season_results.round(5)
)

pairwise_season_results.to_csv(
    os.path.join(
        report_dir,
        "seasonal_exposure_pairwise_tests.csv"
    ),
    index=False
)

print("Pairwise seasonal report saved.")
```

```python
# ============================================================
# TOPIC: SEASONAL POLLUTION EXPOSURE VISUALIZATION
# PURPOSE: Compare city-level exposure rates across seasons.
# METHOD: Plot the distribution of city-season exposure rates.
# OUTPUT: Saved seasonal exposure comparison chart.
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(11, 7))

sns.boxplot(
    data=city_season,
    x="Season",
    y="Exposure_Rate_Percent",
    order=season_order
)

sns.stripplot(
    data=city_season,
    x="Season",
    y="Exposure_Rate_Percent",
    order=season_order,
    color="black",
    alpha=0.45,
    size=4
)

plt.title(
    "City-Level Seasonal High-Pollution Exposure Rates",
    fontsize=15
)

plt.xlabel("Season")
plt.ylabel("High-Pollution Days (%)")
plt.xticks(rotation=15)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "seasonal_city_exposure_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: EXTREME AQI SENSITIVITY ANALYSIS
# PURPOSE: Examine how AQI values above 500 affect city statistics.
# METHOD: Compare full observations with observations restricted
#         to the 0-500 AQI range.
# OUTPUT: City-wise sensitivity comparison report.
# ============================================================

import pandas as pd
import numpy as np
import os

sensitivity_data = air_final[
    ["City", "Date", "AQI"]
].dropna().copy()

# Dataset containing all recorded AQI observations
full_city_stats = (
    sensitivity_data.groupby("City")
    .agg(
        Full_Observation_Count=("AQI", "count"),
        Full_Mean_AQI=("AQI", "mean"),
        Full_Median_AQI=("AQI", "median"),
        Full_Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

# Sensitivity dataset: exclude values outside 0-500
standard_range_data = sensitivity_data[
    sensitivity_data["AQI"].between(0, 500)
].copy()

standard_city_stats = (
    standard_range_data.groupby("City")
    .agg(
        Standard_Range_Count=("AQI", "count"),
        Standard_Range_Mean_AQI=("AQI", "mean"),
        Standard_Range_Median_AQI=("AQI", "median"),
        Standard_Range_Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

# Combine both comparisons
aqi_sensitivity = full_city_stats.merge(
    standard_city_stats,
    on="City",
    how="left"
)

aqi_sensitivity["Mean_AQI_Difference"] = (
    aqi_sensitivity["Full_Mean_AQI"]
    - aqi_sensitivity["Standard_Range_Mean_AQI"]
)

print("AQI EXTREME-VALUE SENSITIVITY ANALYSIS")
display(aqi_sensitivity.round(2))

aqi_sensitivity.to_csv(
    os.path.join(
        report_dir,
        "aqi_extreme_value_sensitivity.csv"
    ),
    index=False
)

print("Sensitivity report saved.")
```

```python
# ============================================================
# TOPIC: POLLUTION THRESHOLD SENSITIVITY ANALYSIS
# PURPOSE: Test how pollution exposure changes under different
#          AQI thresholds.
# METHOD: Calculate city-wise exposure rates at three thresholds.
# OUTPUT: Comparative threshold sensitivity report.
# ============================================================

thresholds = [100, 200, 300]

threshold_results = []

for threshold in thresholds:

    for city, group in sensitivity_data.groupby("City"):

        valid_count = len(group)

        high_count = (
            group["AQI"] > threshold
        ).sum()

        exposure_rate = (
            high_count / valid_count * 100
            if valid_count else np.nan
        )

        threshold_results.append({
            "City": city,
            "AQI_Threshold": threshold,
            "Valid_Observations": valid_count,
            "High_Pollution_Days": high_count,
            "Exposure_Rate_Percent": exposure_rate
        })

threshold_sensitivity = pd.DataFrame(threshold_results)

print("POLLUTION THRESHOLD SENSITIVITY")
display(
    threshold_sensitivity.sort_values(
        ["AQI_Threshold", "Exposure_Rate_Percent"],
        ascending=[True, False]
    ).round(2)
)

threshold_sensitivity.to_csv(
    os.path.join(
        report_dir,
        "pollution_threshold_sensitivity.csv"
    ),
    index=False
)

print("Threshold sensitivity report saved.")
```

```python
# ============================================================
# TOPIC: MONITORING COVERAGE SENSITIVITY ANALYSIS
# PURPOSE: Examine how city exposure rankings change when
#          cities with limited observations are excluded.
# METHOD: Apply minimum observation requirements of 100, 300
#         and 500 valid AQI observations.
# OUTPUT: Monitoring coverage sensitivity report.
# ============================================================

coverage_thresholds = [100, 300, 500]

coverage_results = []

for minimum_observations in coverage_thresholds:

    for city, group in sensitivity_data.groupby("City"):

        valid_count = len(group)

        if valid_count < minimum_observations:
            continue

        high_days = (group["AQI"] > 200).sum()

        exposure_rate = high_days / valid_count * 100

        coverage_results.append({
            "Minimum_Observations": minimum_observations,
            "City": city,
            "Valid_Observations": valid_count,
            "High_Pollution_Days": high_days,
            "Exposure_Rate_Percent": exposure_rate
        })

coverage_sensitivity = pd.DataFrame(coverage_results)

print("MONITORING COVERAGE SENSITIVITY")
display(
    coverage_sensitivity.sort_values(
        ["Minimum_Observations", "Exposure_Rate_Percent"],
        ascending=[True, False]
    ).round(2)
)

coverage_sensitivity.to_csv(
    os.path.join(
        report_dir,
        "monitoring_coverage_sensitivity.csv"
    ),
    index=False
)

print("Coverage sensitivity report saved.")
```

```python
# ============================================================
# TOPIC: SEASONAL EXTREME-VALUE SENSITIVITY
# PURPOSE: Evaluate whether extreme AQI values influence
#          seasonal pollution exposure rates.
# METHOD: Compare seasonal exposure with all observations
#         and with AQI restricted to 0-500.
# OUTPUT: Seasonal sensitivity comparison.
# ============================================================

seasonal_sensitivity_data = sensitivity_data.copy()

seasonal_sensitivity_data["Season"] = (
    seasonal_sensitivity_data["Date"]
    .dt.month.map(assign_season)
)

seasonal_sensitivity_data["Above_200"] = (
    seasonal_sensitivity_data["AQI"] > 200
)

seasonal_full = (
    seasonal_sensitivity_data.groupby("Season")
    .agg(
        Full_Valid_Days=("AQI", "count"),
        Full_High_Days=("Above_200", "sum"),
        Full_Mean_AQI=("AQI", "mean")
    )
    .reset_index()
)

restricted_data = seasonal_sensitivity_data[
    seasonal_sensitivity_data["AQI"].between(0, 500)
].copy()

seasonal_restricted = (
    restricted_data.groupby("Season")
    .agg(
        Restricted_Valid_Days=("AQI", "count"),
        Restricted_High_Days=("Above_200", "sum"),
        Restricted_Mean_AQI=("AQI", "mean")
    )
    .reset_index()
)

seasonal_sensitivity = seasonal_full.merge(
    seasonal_restricted,
    on="Season",
    how="outer"
)

seasonal_sensitivity["Full_Exposure_Rate"] = (
    seasonal_sensitivity["Full_High_Days"]
    / seasonal_sensitivity["Full_Valid_Days"] * 100
)

seasonal_sensitivity["Restricted_Exposure_Rate"] = (
    seasonal_sensitivity["Restricted_High_Days"]
    / seasonal_sensitivity["Restricted_Valid_Days"] * 100
)

print("SEASONAL SENSITIVITY COMPARISON")
display(seasonal_sensitivity.round(2))

seasonal_sensitivity.to_csv(
    os.path.join(
        report_dir,
        "seasonal_extreme_aqi_sensitivity.csv"
    ),
    index=False
)

print("Seasonal sensitivity report saved.")
```

```python
# ============================================================
# TOPIC: ENVIRONMENTAL RISK MODULE COMPLETION AUDIT
# PURPOSE: Inspect existing risk-analysis datasets and outputs.
# METHOD: Check notebook variables, columns and saved reports.
# OUTPUT: Risk module inventory and missing-component checklist.
# ============================================================

import pandas as pd
import os

# 1. Existing analysis datasets
risk_variables = [
    "city_risk_profile",
    "environmental_risk",
    "risk_indicators",
    "risk_ranking",
    "persistent_hotspots",
    "pollution_episodes",
    "city_episode_summary",
    "seasonal_event_summary",
    "annual_event_summary",
    "seasonal_sensitivity",
    "threshold_sensitivity",
    "coverage_sensitivity",
    "aqi_sensitivity"
]

inventory = []

for name in risk_variables:

    available = name in globals()

    if available:
        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):
            inventory.append({
                "Dataset": name,
                "Available": True,
                "Rows": obj.shape[0],
                "Columns": obj.shape[1],
                "Column_Names": ", ".join(obj.columns)
            })
        else:
            inventory.append({
                "Dataset": name,
                "Available": True,
                "Rows": "-",
                "Columns": "-",
                "Column_Names": "Non-DataFrame object"
            })
    else:
        inventory.append({
            "Dataset": name,
            "Available": False,
            "Rows": "-",
            "Columns": "-",
            "Column_Names": "-"
        })

risk_inventory = pd.DataFrame(inventory)

print("EXISTING RISK DATASET INVENTORY")
display(risk_inventory)

# 2. Check saved reports
expected_reports = [
    "environmental_risk_indicators.csv",
    "environmental_risk_city_rankings.csv",
    "persistent_pollution_hotspots.csv",
    "environmental_risk_executive_summary.csv",
    "city_wise_pollution_risk_profile.csv",
    "city_pollution_hotspot_classification.csv"
]

report_inventory = pd.DataFrame({
    "Report": expected_reports,
    "Available": [
        os.path.isfile(os.path.join(report_dir, filename))
        for filename in expected_reports
    ]
})

print("\nSAVED REPORT INVENTORY")
display(report_inventory)

# 3. Check risk-related visualizations
expected_charts = [
    "environmental_risk_hotspot_analysis.png",
    "city_wise_pollution_exposure_rate.png",
    "pollution_episode_frequency_vs_intensity.png"
]

chart_inventory = pd.DataFrame({
    "Visualization": expected_charts,
    "Available": [
        os.path.isfile(
            os.path.join(visualization_dir, filename)
        )
        for filename in expected_charts
    ]
})

print("\nVISUALIZATION INVENTORY")
display(chart_inventory)
```

```python
# ============================================================
# TOPIC: CITY-WISE SEASONAL ENVIRONMENTAL RISK ASSESSMENT
# PURPOSE: Measure seasonal pollution risk separately for each city.
# METHOD: Aggregate AQI, exposure and extreme-event indicators.
# OUTPUT: City-season environmental risk dataset.
# ============================================================

import pandas as pd
import numpy as np
import os

# Use existing validated observations
seasonal_risk_data = air_final[
    ["City", "Date", "AQI"]
].copy()

seasonal_risk_data["Date"] = pd.to_datetime(
    seasonal_risk_data["Date"],
    errors="coerce"
)

seasonal_risk_data = seasonal_risk_data.dropna(
    subset=["City", "Date", "AQI"]
).copy()

# Assign seasons
seasonal_risk_data["Season"] = (
    seasonal_risk_data["Date"].dt.month.map(assign_season)
)

# Aggregate city-season indicators
city_season_risk = (
    seasonal_risk_data
    .groupby(["City", "Season"])
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Average_AQI=("AQI", "mean"),
        Median_AQI=("AQI", "median"),
        Maximum_AQI=("AQI", "max"),
        High_Pollution_Days=("AQI", lambda x: (x > 200).sum()),
        Extreme_AQI_Days=("AQI", lambda x: (x > 500).sum())
    )
    .reset_index()
)

# Calculate exposure percentages
city_season_risk["High_Pollution_Rate_Percent"] = (
    city_season_risk["High_Pollution_Days"]
    / city_season_risk["Valid_AQI_Days"] * 100
)

city_season_risk["Extreme_AQI_Rate_Percent"] = (
    city_season_risk["Extreme_AQI_Days"]
    / city_season_risk["Valid_AQI_Days"] * 100
)

# Rank cities within each season
city_season_risk["Seasonal_Exposure_Rank"] = (
    city_season_risk.groupby("Season")[
        "High_Pollution_Rate_Percent"
    ].rank(method="min", ascending=False)
)

season_order = [
    "Winter",
    "Pre-Monsoon",
    "Monsoon",
    "Post-Monsoon"
]

city_season_risk["Season"] = pd.Categorical(
    city_season_risk["Season"],
    categories=season_order,
    ordered=True
)

city_season_risk = city_season_risk.sort_values(
    ["Season", "Seasonal_Exposure_Rank"]
)

print("CITY-WISE SEASONAL RISK ASSESSMENT")
display(city_season_risk.round(2))

city_season_risk.to_csv(
    os.path.join(
        processed_dir,
        "city_wise_seasonal_environmental_risk.csv"
    ),
    index=False
)

print("Seasonal environmental risk dataset saved.")
```

```python
# ============================================================
# TOPIC: SEASONAL ENVIRONMENTAL RISK HOTSPOT IDENTIFICATION
# PURPOSE: Identify the highest-exposure city in each season.
# METHOD: Rank city-season pollution exposure percentages.
# OUTPUT: Seasonal environmental hotspot report.
# ============================================================

seasonal_hotspots = (
    city_season_risk
    .sort_values(
        "High_Pollution_Rate_Percent",
        ascending=False
    )
    .groupby("Season", observed=True)
    .head(5)
    .sort_values(
        ["Season", "Seasonal_Exposure_Rank"]
    )
)

print("TOP FIVE SEASONAL POLLUTION HOTSPOTS")

display(
    seasonal_hotspots[
        [
            "Season",
            "City",
            "Valid_AQI_Days",
            "High_Pollution_Days",
            "High_Pollution_Rate_Percent",
            "Average_AQI",
            "Maximum_AQI",
            "Extreme_AQI_Days"
        ]
    ].round(2)
)

seasonal_hotspots.to_csv(
    os.path.join(
        report_dir,
        "seasonal_environmental_hotspots.csv"
    ),
    index=False
)

print("Seasonal hotspot report saved.")
```

```python
# ============================================================
# TOPIC: CITY-SEASON ENVIRONMENTAL RISK HEATMAP
# PURPOSE: Compare seasonal pollution exposure across cities.
# METHOD: Pivot city-season exposure percentages into a matrix.
# OUTPUT: High-resolution seasonal risk heatmap.
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

seasonal_risk_matrix = city_season_risk.pivot(
    index="City",
    columns="Season",
    values="High_Pollution_Rate_Percent"
)

seasonal_risk_matrix = seasonal_risk_matrix.reindex(
    columns=season_order
)

plt.figure(figsize=(12, 11))

sns.heatmap(
    seasonal_risk_matrix,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={
        "label": "High-Pollution Exposure Rate (%)"
    }
)

plt.title(
    "City-wise Seasonal Environmental Risk Heatmap",
    fontsize=16
)

plt.xlabel("Season")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_season_environmental_risk_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: SEASONAL CITY RISK RANKING COMPARISON
# PURPOSE: Identify cities with changing seasonal risk rankings.
# METHOD: Compare each city's highest and lowest seasonal rank.
# OUTPUT: Seasonal risk variability report.
# ============================================================

seasonal_rank_matrix = city_season_risk.pivot(
    index="City",
    columns="Season",
    values="Seasonal_Exposure_Rank"
)

seasonal_rank_matrix = seasonal_rank_matrix.reindex(
    columns=season_order
)

seasonal_rank_variability = pd.DataFrame({
    "Highest_Risk_Rank": seasonal_rank_matrix.min(axis=1),
    "Lowest_Risk_Rank": seasonal_rank_matrix.max(axis=1),
    "Seasonal_Rank_Range": (
        seasonal_rank_matrix.max(axis=1)
        - seasonal_rank_matrix.min(axis=1)
    )
}).reset_index()

seasonal_rank_variability = (
    seasonal_rank_variability.sort_values(
        "Seasonal_Rank_Range",
        ascending=False
    )
)

print("SEASONAL RISK RANKING VARIABILITY")

display(seasonal_rank_variability.round(2))

seasonal_rank_variability.to_csv(
    os.path.join(
        report_dir,
        "seasonal_city_risk_rank_variability.csv"
    ),
    index=False
)

print("Seasonal ranking variability report saved.")
```

```python
# ============================================================
# TOPIC: ANNUAL CITY-WISE POLLUTION EXPOSURE ANALYSIS
# PURPOSE: Measure high-pollution exposure separately for
#          every city and calendar year.
# METHOD: Calculate AQI > 200 frequency and exposure rate.
# OUTPUT: Annual city-level pollution exposure dataset.
# ============================================================

import pandas as pd
import numpy as np
import os

# Use the existing final analytical dataset
annual_city_data = air_final[
    ["City", "Date", "AQI"]
].copy()

annual_city_data["Date"] = pd.to_datetime(
    annual_city_data["Date"],
    errors="coerce"
)

annual_city_data = annual_city_data.dropna(
    subset=["City", "Date", "AQI"]
).copy()

annual_city_data["Year"] = (
    annual_city_data["Date"].dt.year
)

annual_city_risk = (
    annual_city_data.groupby(["City", "Year"])
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max"),
        High_Pollution_Days=(
            "AQI", lambda x: (x > 200).sum()
        ),
        Extreme_AQI_Days=(
            "AQI", lambda x: (x > 500).sum()
        )
    )
    .reset_index()
)

annual_city_risk["High_Pollution_Rate_Percent"] = (
    annual_city_risk["High_Pollution_Days"]
    / annual_city_risk["Valid_AQI_Days"] * 100
)

annual_city_risk["Extreme_AQI_Rate_Percent"] = (
    annual_city_risk["Extreme_AQI_Days"]
    / annual_city_risk["Valid_AQI_Days"] * 100
)

# Flag city-year combinations with sufficient observations
annual_city_risk["Coverage_Eligible"] = (
    annual_city_risk["Valid_AQI_Days"] >= 30
)

print("ANNUAL CITY-WISE POLLUTION EXPOSURE")

display(
    annual_city_risk.sort_values(
        ["Year", "High_Pollution_Rate_Percent"],
        ascending=[True, False]
    ).round(2)
)

annual_city_risk.to_csv(
    os.path.join(
        processed_dir,
        "annual_city_pollution_risk.csv"
    ),
    index=False
)

print("Annual city risk dataset saved.")
```

```python
# ============================================================
# TOPIC: YEAR-WISE POLLUTION HOTSPOT RANKING
# PURPOSE: Identify the highest-exposure cities in each year.
# METHOD: Rank city exposure rates and identify the annual
#         top quartile of eligible cities.
# OUTPUT: Year-wise pollution hotspot classification.
# ============================================================

eligible_annual_risk = annual_city_risk[
    annual_city_risk["Coverage_Eligible"]
].copy()

eligible_annual_risk["Annual_Risk_Rank"] = (
    eligible_annual_risk.groupby("Year")[
        "High_Pollution_Rate_Percent"
    ].rank(method="min", ascending=False)
)

eligible_annual_risk["Annual_Percentile"] = (
    eligible_annual_risk.groupby("Year")[
        "High_Pollution_Rate_Percent"
    ].rank(method="average", pct=True)
)

# Top 25% means percentile rank above 75%
eligible_annual_risk["Annual_Hotspot"] = (
    eligible_annual_risk["Annual_Percentile"] >= 0.75
)

print("ANNUAL POLLUTION HOTSPOTS")

display(
    eligible_annual_risk[
        eligible_annual_risk["Annual_Hotspot"]
    ].sort_values(
        ["Year", "Annual_Risk_Rank"]
    ).round(2)
)

eligible_annual_risk.to_csv(
    os.path.join(
        report_dir,
        "annual_pollution_hotspot_rankings.csv"
    ),
    index=False
)

print("Annual hotspot rankings saved.")
```

```python
# ============================================================
# TOPIC: MULTI-YEAR PERSISTENT POLLUTION HOTSPOT DETECTION
# PURPOSE: Identify cities repeatedly appearing among annual
#          high-pollution exposure hotspots.
# METHOD: Aggregate annual hotspot appearances and eligible years.
# OUTPUT: Multi-year persistent hotspot report.
# ============================================================

persistent_hotspot_summary = (
    eligible_annual_risk.groupby("City")
    .agg(
        Eligible_Years=("Year", "nunique"),
        Hotspot_Years=("Annual_Hotspot", "sum"),
        Average_Annual_Exposure=(
            "High_Pollution_Rate_Percent", "mean"
        ),
        Maximum_Annual_Exposure=(
            "High_Pollution_Rate_Percent", "max"
        ),
        Average_Annual_AQI=("Average_AQI", "mean"),
        Maximum_Recorded_AQI=("Maximum_AQI", "max")
    )
    .reset_index()
)

persistent_hotspot_summary["Hotspot_Consistency_Percent"] = (
    persistent_hotspot_summary["Hotspot_Years"]
    / persistent_hotspot_summary["Eligible_Years"] * 100
)

persistent_hotspot_summary = (
    persistent_hotspot_summary.sort_values(
        [
            "Hotspot_Consistency_Percent",
            "Hotspot_Years",
            "Average_Annual_Exposure"
        ],
        ascending=False
    )
)

print("MULTI-YEAR PERSISTENT POLLUTION HOTSPOTS")

display(persistent_hotspot_summary.round(2))

persistent_hotspot_summary.to_csv(
    os.path.join(
        report_dir,
        "persistent_multi_year_pollution_hotspots.csv"
    ),
    index=False
)

print("Persistent hotspot report saved.")
```

```python
# ============================================================
# TOPIC: MULTI-YEAR POLLUTION HOTSPOT HEATMAP
# PURPOSE: Visualize changes in city-level pollution exposure
#          across the study period.
# METHOD: Create a city-year matrix of high-pollution rates.
# OUTPUT: Multi-year environmental hotspot heatmap.
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

annual_exposure_matrix = eligible_annual_risk.pivot(
    index="City",
    columns="Year",
    values="High_Pollution_Rate_Percent"
)

# Sort cities by average annual exposure
annual_exposure_matrix = annual_exposure_matrix.loc[
    annual_exposure_matrix.mean(axis=1).sort_values(
        ascending=False
    ).index
]

plt.figure(figsize=(14, 11))

sns.heatmap(
    annual_exposure_matrix,
    cmap="YlOrRd",
    annot=True,
    fmt=".1f",
    linewidths=0.4,
    mask=annual_exposure_matrix.isna(),
    cbar_kws={
        "label": "High-Pollution Exposure Rate (%)"
    }
)

plt.title(
    "Multi-Year City-Wise Pollution Hotspot Analysis",
    fontsize=16
)

plt.xlabel("Year")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "multi_year_pollution_hotspot_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: POLLUTION HOTSPOT CONSISTENCY ASSESSMENT
# PURPOSE: Classify cities according to repeated hotspot appearances.
# METHOD: Evaluate hotspot frequency across eligible years.
# OUTPUT: Multi-year hotspot consistency classification.
# ============================================================

def classify_hotspot_consistency(row):

    if row["Eligible_Years"] < 2:
        return "Insufficient Annual Coverage"

    rate = row["Hotspot_Consistency_Percent"]

    if rate >= 75:
        return "Highly Persistent"

    elif rate >= 50:
        return "Frequently Persistent"

    elif rate > 0:
        return "Occasional Hotspot"

    return "Not Repeatedly Identified"

persistent_hotspot_summary["Persistence_Class"] = (
    persistent_hotspot_summary.apply(
        classify_hotspot_consistency,
        axis=1
    )
)

print("HOTSPOT CONSISTENCY CLASSIFICATION")

display(
    persistent_hotspot_summary[
        [
            "City",
            "Eligible_Years",
            "Hotspot_Years",
            "Hotspot_Consistency_Percent",
            "Persistence_Class"
        ]
    ].round(2)
)

persistent_hotspot_summary.to_csv(
    os.path.join(
        report_dir,
        "pollution_hotspot_consistency_classification.csv"
    ),
    index=False
)

print("Hotspot consistency classification saved.")
```

```python
# ============================================================
# TOPIC: EXTREME AQI DATA EXTRACTION
# PURPOSE: Extract observations with AQI greater than 500.
# METHOD: Filter the validated dataset without deleting or
#         modifying the original observations.
# OUTPUT: Extreme AQI records dataset.
# ============================================================

extreme_aqi_data = air_final[
    ["City", "Date", "AQI", "AQI_Bucket"]
].copy()

extreme_aqi_data["Date"] = pd.to_datetime(
    extreme_aqi_data["Date"],
    errors="coerce"
)

extreme_aqi_data = extreme_aqi_data.dropna(
    subset=["City", "Date", "AQI"]
).copy()

extreme_aqi_data["Year"] = (
    extreme_aqi_data["Date"].dt.year
)

extreme_aqi_data["Extreme_AQI_Flag"] = (
    extreme_aqi_data["AQI"] > 500
)

extreme_aqi_records = extreme_aqi_data[
    extreme_aqi_data["Extreme_AQI_Flag"]
].copy()

print("EXTREME AQI DATA SUMMARY")
print("Total valid AQI observations:", len(extreme_aqi_data))
print("Extreme AQI observations:", len(extreme_aqi_records))
print(
    "Extreme AQI percentage:",
    round(
        len(extreme_aqi_records)
        / len(extreme_aqi_data) * 100,
        2
    ),
    "%"
)

display(
    extreme_aqi_records.sort_values(
        "AQI",
        ascending=False
    ).head(20)
)

extreme_aqi_records.to_csv(
    os.path.join(
        processed_dir,
        "extreme_aqi_records.csv"
    ),
    index=False
)

print("Extreme AQI records saved.")
```

```python
# ============================================================
# TOPIC: CITY-WISE EXTREME POLLUTION RISK
# PURPOSE: Measure extreme AQI frequency and exposure rates
#          across cities.
# METHOD: Compare extreme observations with total valid AQI
#         observations for each city.
# OUTPUT: City-wise extreme pollution risk report.
# ============================================================

city_extreme_risk = (
    extreme_aqi_data.groupby("City")
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Extreme_AQI_Days=("Extreme_AQI_Flag", "sum"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

city_extreme_risk["Extreme_AQI_Rate_Percent"] = (
    city_extreme_risk["Extreme_AQI_Days"]
    / city_extreme_risk["Valid_AQI_Days"] * 100
)

city_extreme_risk = city_extreme_risk.sort_values(
    "Extreme_AQI_Rate_Percent",
    ascending=False
)

print("CITY-WISE EXTREME POLLUTION RISK")

display(city_extreme_risk.round(2))

city_extreme_risk.to_csv(
    os.path.join(
        report_dir,
        "city_wise_extreme_pollution_risk.csv"
    ),
    index=False
)

print("City-wise extreme pollution report saved.")
```

```python
# ============================================================
# TOPIC: EXTREME POLLUTION BURDEN RANKING
# PURPOSE: Identify cities contributing the largest number
#          of extreme AQI observations.
# METHOD: Rank cities by extreme AQI observation counts.
# OUTPUT: Extreme pollution burden ranking and chart.
# ============================================================

city_extreme_ranking = city_extreme_risk.sort_values(
    "Extreme_AQI_Days",
    ascending=False
).copy()

display(
    city_extreme_ranking[
        [
            "City",
            "Extreme_AQI_Days",
            "Valid_AQI_Days",
            "Extreme_AQI_Rate_Percent",
            "Maximum_AQI"
        ]
    ].head(15).round(2)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=city_extreme_ranking.head(15),
    x="Extreme_AQI_Days",
    y="City"
)

plt.title(
    "Top 15 Cities by Extreme AQI Observation Count",
    fontsize=15
)

plt.xlabel("Number of AQI Observations Above 500")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_wise_extreme_aqi_burden.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: YEAR-WISE EXTREME POLLUTION ANALYSIS
# PURPOSE: Examine annual changes in extreme AQI frequency.
# METHOD: Calculate yearly extreme observations and their
#         percentage among valid AQI observations.
# OUTPUT: Annual extreme pollution trend report.
# ============================================================

annual_extreme_risk = (
    extreme_aqi_data.groupby("Year")
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Extreme_AQI_Days=("Extreme_AQI_Flag", "sum"),
        Average_AQI=("AQI", "mean"),
        Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

annual_extreme_risk["Extreme_AQI_Rate_Percent"] = (
    annual_extreme_risk["Extreme_AQI_Days"]
    / annual_extreme_risk["Valid_AQI_Days"] * 100
)

print("YEAR-WISE EXTREME POLLUTION ANALYSIS")

display(annual_extreme_risk.round(2))

annual_extreme_risk.to_csv(
    os.path.join(
        report_dir,
        "annual_extreme_pollution_analysis.csv"
    ),
    index=False
)

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=annual_extreme_risk,
    x="Year",
    y="Extreme_AQI_Rate_Percent",
    marker="o",
    linewidth=2
)

plt.title(
    "Annual Extreme AQI Exposure Rate",
    fontsize=15
)

plt.xlabel("Year")
plt.ylabel("Extreme AQI Rate (%)")
plt.xticks(annual_extreme_risk["Year"])

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "annual_extreme_aqi_exposure_trend.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: EXTREME AQI SEVERITY DISTRIBUTION
# PURPOSE: Examine the severity distribution of AQI values
#          exceeding 500.
# METHOD: Group extreme observations into descriptive ranges.
# OUTPUT: Extreme AQI severity distribution report.
# ============================================================

severity_bins = [500, 1000, 1500, 2000, np.inf]

severity_labels = [
    "501-1000",
    "1001-1500",
    "1501-2000",
    "Above 2000"
]

extreme_aqi_records["Extreme_AQI_Range"] = pd.cut(
    extreme_aqi_records["AQI"],
    bins=severity_bins,
    labels=severity_labels,
    right=True
)

extreme_severity_summary = (
    extreme_aqi_records["Extreme_AQI_Range"]
    .value_counts(sort=False)
    .rename_axis("Extreme_AQI_Range")
    .reset_index(name="Observation_Count")
)

extreme_severity_summary["Percentage"] = (
    extreme_severity_summary["Observation_Count"]
    / extreme_severity_summary["Observation_Count"].sum()
    * 100
)

print("EXTREME AQI SEVERITY DISTRIBUTION")

display(extreme_severity_summary.round(2))

extreme_severity_summary.to_csv(
    os.path.join(
        report_dir,
        "extreme_aqi_severity_distribution.csv"
    ),
    index=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=extreme_severity_summary,
    x="Extreme_AQI_Range",
    y="Observation_Count"
)

plt.title(
    "Distribution of Extreme AQI Observations",
    fontsize=15
)

plt.xlabel("AQI Range")
plt.ylabel("Number of Observations")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "extreme_aqi_severity_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: CITY-YEAR EXTREME POLLUTION HOTSPOT ANALYSIS
# PURPOSE: Identify city and year combinations with extreme
#          pollution observations.
# METHOD: Aggregate extreme AQI counts and rates by city-year.
# OUTPUT: City-year extreme pollution hotspot report.
# ============================================================

city_year_extreme_risk = (
    extreme_aqi_data.groupby(["City", "Year"])
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Extreme_AQI_Days=("Extreme_AQI_Flag", "sum"),
        Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

city_year_extreme_risk["Extreme_AQI_Rate_Percent"] = (
    city_year_extreme_risk["Extreme_AQI_Days"]
    / city_year_extreme_risk["Valid_AQI_Days"] * 100
)

city_year_extreme_risk = city_year_extreme_risk.sort_values(
    ["Extreme_AQI_Days", "Extreme_AQI_Rate_Percent"],
    ascending=False
)

print("TOP CITY-YEAR EXTREME POLLUTION RECORDS")

display(city_year_extreme_risk.head(25).round(2))

city_year_extreme_risk.to_csv(
    os.path.join(
        report_dir,
        "city_year_extreme_pollution_hotspots.csv"
    ),
    index=False
)

print("City-year extreme pollution report saved.")
```

```python
# ============================================================
# TOPIC: CITY-YEAR EXTREME POLLUTION HEATMAP
# PURPOSE: Visualize the concentration of extreme AQI events
#          across cities and years.
# METHOD: Pivot city-year extreme observation counts.
# OUTPUT: Extreme pollution concentration heatmap.
# ============================================================

extreme_count_matrix = city_year_extreme_risk.pivot(
    index="City",
    columns="Year",
    values="Extreme_AQI_Days"
).fillna(0)

extreme_count_matrix = extreme_count_matrix.loc[
    extreme_count_matrix.sum(axis=1).sort_values(
        ascending=False
    ).index
]

plt.figure(figsize=(13, 10))

sns.heatmap(
    extreme_count_matrix,
    cmap="Reds",
    annot=True,
    fmt=".0f",
    linewidths=0.4,
    cbar_kws={
        "label": "Extreme AQI Observations"
    }
)

plt.title(
    "City-Year Extreme Pollution Concentration",
    fontsize=15
)

plt.xlabel("Year")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_year_extreme_pollution_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# TOPIC: EXTREME POLLUTION RISK INDICATOR CONSOLIDATION
# PURPOSE: Create a structured city-level summary of extreme
#          pollution burden and exposure.
# METHOD: Combine frequency, exposure rate and severity metrics.
# OUTPUT: Consolidated extreme pollution risk indicators.
# ============================================================

extreme_risk_indicators = city_extreme_risk[
    [
        "City",
        "Valid_AQI_Days",
        "Extreme_AQI_Days",
        "Extreme_AQI_Rate_Percent",
        "Average_AQI",
        "Maximum_AQI"
    ]
].copy()

extreme_risk_indicators["Extreme_Risk_Rank"] = (
    extreme_risk_indicators[
        "Extreme_AQI_Rate_Percent"
    ].rank(
        method="min",
        ascending=False
    ).astype(int)
)

extreme_risk_indicators = extreme_risk_indicators.sort_values(
    "Extreme_Risk_Rank"
)

print("CONSOLIDATED EXTREME POLLUTION RISK INDICATORS")

display(extreme_risk_indicators.round(2))

extreme_risk_indicators.to_csv(
    os.path.join(
        report_dir,
        "extreme_pollution_risk_indicators.csv"
    ),
    index=False
)

print("Consolidated extreme risk indicators saved.")
```

```python
# ============================================================
# TOPIC: POLLUTANT-SPECIFIC DATA PREPARATION
# PURPOSE: Prepare pollutant measurements for city-level
#          environmental risk analysis.
# METHOD: Select the 12 pollutant columns and validate values.
# OUTPUT: Clean pollutant analytical dataset.
# ============================================================

pollutants = [
    "PM2.5", "PM10",
    "NO", "NO2", "NOx", "NH3",
    "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

pollutant_data = air_final[
    ["City", "Date"] + pollutants
].copy()

pollutant_data["Date"] = pd.to_datetime(
    pollutant_data["Date"],
    errors="coerce"
)

for pollutant in pollutants:
    pollutant_data[pollutant] = pd.to_numeric(
        pollutant_data[pollutant],
        errors="coerce"
    )

# Replace infinite values with missing values for analysis.
pollutant_data[pollutants] = (
    pollutant_data[pollutants]
    .replace([np.inf, -np.inf], np.nan)
)

# Retain the original dataset unchanged.
# Negative values are excluded from concentration summaries
# and separately counted for data-quality review.
negative_pollutant_counts = (
    pollutant_data[pollutants].lt(0).sum()
)

pollutant_data[pollutants] = (
    pollutant_data[pollutants].mask(
        pollutant_data[pollutants] < 0
    )
)

print("POLLUTANT ANALYTICAL DATA")
print("Dataset shape:", pollutant_data.shape)
print("Cities:", pollutant_data["City"].nunique())

print("\nNegative values excluded from concentration analysis:")
display(negative_pollutant_counts.to_frame(
    "Negative_Value_Count"
))

display(pollutant_data.head())
```

```python
# ============================================================
# TOPIC: OVERALL POLLUTANT CONCENTRATION PROFILE
# PURPOSE: Understand the distribution and variability of
#          individual pollutant concentrations.
# METHOD: Calculate descriptive statistics pollutant by pollutant.
# OUTPUT: Overall pollutant statistical summary.
# ============================================================

pollutant_statistics = []

for pollutant in pollutants:

    values = pollutant_data[pollutant].dropna()

    pollutant_statistics.append({
        "Pollutant": pollutant,
        "Valid_Observations": values.count(),
        "Missing_Observations": pollutant_data[pollutant].isna().sum(),
        "Mean": values.mean(),
        "Median": values.median(),
        "Standard_Deviation": values.std(),
        "P90": values.quantile(0.90),
        "P95": values.quantile(0.95),
        "Maximum": values.max()
    })

pollutant_statistics = pd.DataFrame(
    pollutant_statistics
)

print("OVERALL POLLUTANT STATISTICS")

display(
    pollutant_statistics.round(2)
)

pollutant_statistics.to_csv(
    os.path.join(
        report_dir,
        "overall_pollutant_statistical_profile.csv"
    ),
    index=False
)

print("Pollutant statistical profile saved.")
```

```python
# ============================================================
# TOPIC: POLLUTANT-SPECIFIC CITY HOTSPOT RANKING
# PURPOSE: Identify the cities with the highest average
#          concentration for each pollutant.
# METHOD: Rank cities independently for every pollutant.
# OUTPUT: Pollutant-specific city hotspot rankings.
# ============================================================

pollutant_city_rankings = []

for pollutant in pollutants:

    mean_column = f"{pollutant}_mean"
    count_column = f"{pollutant}_count"

    temp = city_pollutant_risk[
        ["City", mean_column, count_column]
    ].copy()

    temp = temp.rename(columns={
        mean_column: "Average_Concentration",
        count_column: "Valid_Observations"
    })

    temp["Pollutant"] = pollutant

    temp["Pollutant_Rank"] = (
        temp["Average_Concentration"]
        .rank(method="min", ascending=False)
    )

    pollutant_city_rankings.append(temp)

pollutant_city_rankings = pd.concat(
    pollutant_city_rankings,
    ignore_index=True
)

pollutant_city_rankings = (
    pollutant_city_rankings.sort_values(
        ["Pollutant", "Pollutant_Rank"]
    )
)

print("TOP 5 CITIES FOR EACH POLLUTANT")

display(
    pollutant_city_rankings.groupby(
        "Pollutant",
        group_keys=False
    ).head(5).round(2)
)

pollutant_city_rankings.to_csv(
    os.path.join(
        report_dir,
        "pollutant_specific_city_hotspot_rankings.csv"
    ),
    index=False
)

print("Pollutant-specific hotspot rankings saved.")
```

```python
# ============================================================
# TOPIC: POLLUTANT-SPECIFIC HIGH-CONCENTRATION DETECTION
# PURPOSE: Identify unusually high concentration observations
#          for each pollutant.
# METHOD: Apply a separate dataset-wide 90th-percentile
#         threshold to every pollutant.
# OUTPUT: Pollutant-specific high-concentration flags.
# ============================================================

pollutant_thresholds = {}

for pollutant in pollutants:

    pollutant_thresholds[pollutant] = (
        pollutant_data[pollutant].quantile(0.90)
    )

for pollutant in pollutants:

    threshold = pollutant_thresholds[pollutant]

    pollutant_data[f"{pollutant}_High_Flag"] = (
        pollutant_data[pollutant] > threshold
    )

print("POLLUTANT-SPECIFIC 90TH-PERCENTILE THRESHOLDS")

threshold_report = pd.DataFrame({
    "Pollutant": list(pollutant_thresholds.keys()),
    "P90_Threshold": list(pollutant_thresholds.values())
})

display(threshold_report.round(3))

threshold_report.to_csv(
    os.path.join(
        report_dir,
        "pollutant_high_concentration_thresholds.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: REBUILD CITY-WISE POLLUTANT CONCENTRATION PROFILE
# PURPOSE: Restore the missing city_pollutant_risk variable.
# METHOD: Aggregate pollutant statistics separately by city.
# OUTPUT: city_pollutant_risk DataFrame.
# ============================================================

city_pollutant_risk = (
    pollutant_data.groupby("City")[pollutants]
    .agg(["count", "mean", "median", "max"])
)

city_pollutant_risk.columns = [
    f"{pollutant}_{statistic}"
    for pollutant, statistic in city_pollutant_risk.columns
]

city_pollutant_risk = city_pollutant_risk.reset_index()

print("City-wise pollutant risk profile created.")
print("Shape:", city_pollutant_risk.shape)

display(city_pollutant_risk.head())
```

```python

# ============================================================
# TOPIC: POLLUTANT-SPECIFIC HIGH-CONCENTRATION DETECTION
# PURPOSE: Identify unusually high pollutant measurements.
# METHOD: Calculate a separate 90th-percentile threshold
#         for each pollutant.
# OUTPUT: High-concentration flags and threshold report.
# ============================================================

pollutant_thresholds = {}

for pollutant in pollutants:
    pollutant_thresholds[pollutant] = (
        pollutant_data[pollutant].quantile(0.90)
    )

for pollutant in pollutants:
    threshold = pollutant_thresholds[pollutant]

    pollutant_data[f"{pollutant}_High_Flag"] = (
        pollutant_data[pollutant] > threshold
    )

threshold_report = pd.DataFrame({
    "Pollutant": list(pollutant_thresholds.keys()),
    "P90_Threshold": list(pollutant_thresholds.values())
})

display(threshold_report.round(3))

threshold_report.to_csv(
    os.path.join(
        report_dir,
        "pollutant_high_concentration_thresholds.csv"
    ),
    index=False
)

print("Pollutant-specific thresholds saved.")
```

```python

# ============================================================
# TOPIC: CITY-WISE POLLUTANT EXPOSURE RATE
# PURPOSE: Compare high-concentration frequency across cities.
# METHOD: Calculate flagged observations as a percentage
#         of valid observations for each pollutant.
# OUTPUT: City-wise pollutant exposure report.
# ============================================================

city_pollutant_exposure = []

for pollutant in pollutants:

    flag_column = f"{pollutant}_High_Flag"

    temp = (
        pollutant_data.groupby("City")
        .agg(
            Valid_Observations=(pollutant, "count"),
            High_Concentration_Days=(flag_column, "sum")
        )
        .reset_index()
    )

    temp["High_Concentration_Rate_Percent"] = (
        temp["High_Concentration_Days"]
        / temp["Valid_Observations"] * 100
    )

    temp["Pollutant"] = pollutant

    city_pollutant_exposure.append(temp)

city_pollutant_exposure = pd.concat(
    city_pollutant_exposure,
    ignore_index=True
)

display(
    city_pollutant_exposure.sort_values(
        ["Pollutant", "High_Concentration_Rate_Percent"],
        ascending=[True, False]
    ).round(2)
)

city_pollutant_exposure.to_csv(
    os.path.join(
        report_dir,
        "city_wise_pollutant_high_exposure_rates.csv"
    ),
    index=False
)

print("City-wise pollutant exposure report saved.")
```

```python

# ============================================================
# TOPIC: CITY-WISE POLLUTANT EXPOSURE HEATMAP
# PURPOSE: Compare high-concentration exposure patterns
#          across cities and pollutants.
# METHOD: Convert exposure rates into a matrix.
# OUTPUT: Professional pollutant exposure heatmap.
# ============================================================

pollutant_exposure_matrix = city_pollutant_exposure.pivot(
    index="City",
    columns="Pollutant",
    values="High_Concentration_Rate_Percent"
)

plt.figure(figsize=(15, 10))

sns.heatmap(
    pollutant_exposure_matrix,
    cmap="YlOrRd",
    annot=True,
    fmt=".1f",
    linewidths=0.4,
    cbar_kws={
        "label": "High-Concentration Exposure Rate (%)"
    }
)

plt.title(
    "City-Wise Pollutant Exposure Risk Heatmap",
    fontsize=16
)

plt.xlabel("Pollutant")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_wise_pollutant_exposure_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: POLLUTANT CONCENTRATION VARIABILITY
# PURPOSE: Measure relative variation in pollutant values.
# METHOD: Calculate standard deviation divided by mean.
# OUTPUT: Pollutant variability report and chart.
# ============================================================

pollutant_variability = []

for pollutant in pollutants:

    values = pollutant_data[pollutant].dropna()

    mean_value = values.mean()
    std_value = values.std()

    cv_percent = (
        std_value / mean_value * 100
        if mean_value != 0 else np.nan
    )

    pollutant_variability.append({
        "Pollutant": pollutant,
        "Mean_Concentration": mean_value,
        "Standard_Deviation": std_value,
        "Coefficient_of_Variation_Percent": cv_percent
    })

pollutant_variability = pd.DataFrame(
    pollutant_variability
).sort_values(
    "Coefficient_of_Variation_Percent",
    ascending=False
)

display(pollutant_variability.round(2))

pollutant_variability.to_csv(
    os.path.join(
        report_dir,
        "pollutant_concentration_variability.csv"
    ),
    index=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=pollutant_variability,
    x="Coefficient_of_Variation_Percent",
    y="Pollutant"
)

plt.title(
    "Relative Variability Across Air Pollutants",
    fontsize=15
)

plt.xlabel("Coefficient of Variation (%)")
plt.ylabel("Pollutant")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "pollutant_concentration_variability.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: POLLUTANT RISK INDICATOR CONSOLIDATION
# PURPOSE: Combine statistical, variability and exposure
#          indicators into one pollutant-level report.
# METHOD: Merge the previously generated analytical summaries.
# OUTPUT: Consolidated pollutant risk intelligence report.
# ============================================================

overall_exposure_summary = (
    city_pollutant_exposure.groupby("Pollutant")
    .agg(
        Cities_Analyzed=("City", "nunique"),
        Average_City_Exposure_Rate=(
            "High_Concentration_Rate_Percent", "mean"
        ),
        Maximum_City_Exposure_Rate=(
            "High_Concentration_Rate_Percent", "max"
        )
    )
    .reset_index()
)

pollutant_risk_summary = (
    pollutant_statistics
    .merge(
        pollutant_variability[
            ["Pollutant", "Coefficient_of_Variation_Percent"]
        ],
        on="Pollutant",
        how="left"
    )
    .merge(
        overall_exposure_summary,
        on="Pollutant",
        how="left"
    )
)

display(pollutant_risk_summary.round(2))

pollutant_risk_summary.to_csv(
    os.path.join(
        report_dir,
        "consolidated_pollutant_risk_intelligence.csv"
    ),
    index=False
)

print("Consolidated pollutant risk report saved.")
```

```python

# ============================================================
# TOPIC: COMPOSITE ENVIRONMENTAL RISK DATA PREPARATION
# PURPOSE: Build consistent city-level environmental indicators.
# METHOD: Aggregate AQI, high-pollution and extreme-pollution
#         observations for every city.
# OUTPUT: Base dataset for the composite risk index.
# ============================================================

risk_source = air_final[
    ["City", "Date", "AQI"]
].copy()

risk_source["Date"] = pd.to_datetime(
    risk_source["Date"],
    errors="coerce"
)

risk_source["AQI"] = pd.to_numeric(
    risk_source["AQI"],
    errors="coerce"
)

risk_source = risk_source.dropna(
    subset=["City", "Date"]
).copy()

valid_risk_source = risk_source.dropna(
    subset=["AQI"]
).copy()

city_index = (
    valid_risk_source.groupby("City")
    .agg(
        Valid_AQI_Days=("AQI", "count"),
        Average_AQI=("AQI", "mean"),
        High_Pollution_Days=(
            "AQI", lambda x: (x > 200).sum()
        ),
        Extreme_AQI_Days=(
            "AQI", lambda x: (x > 500).sum()
        ),
        Maximum_AQI=("AQI", "max")
    )
    .reset_index()
)

city_index["High_Pollution_Rate"] = (
    city_index["High_Pollution_Days"]
    / city_index["Valid_AQI_Days"] * 100
)

city_index["Extreme_AQI_Rate"] = (
    city_index["Extreme_AQI_Days"]
    / city_index["Valid_AQI_Days"] * 100
)

print("BASE ENVIRONMENTAL RISK DATASET")
print("Cities:", city_index["City"].nunique())

display(city_index.round(2))

city_index.to_csv(
    os.path.join(
        processed_dir,
        "composite_risk_base_indicators.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: POLLUTANT EXPOSURE INDEX CONSTRUCTION
# PURPOSE: Add pollutant-specific exposure information to
#          the city-level environmental risk dataset.
# METHOD: Average available pollutant-specific high-
#         concentration exposure rates by city.
# OUTPUT: City-level pollutant exposure indicator.
# ============================================================

if "city_pollutant_exposure" not in globals():
    raise NameError(
        "Run Submodule 6.5 Step 6 before continuing."
    )

city_pollutant_index = (
    city_pollutant_exposure.groupby("City")
    .agg(
        Pollutants_Available=("Pollutant", "nunique"),
        Pollutant_Exposure_Index=(
            "High_Concentration_Rate_Percent", "mean"
        )
    )
    .reset_index()
)

city_index = city_index.merge(
    city_pollutant_index,
    on="City",
    how="left"
)

print("POLLUTANT EXPOSURE INDICATORS")

display(
    city_index[
        [
            "City",
            "Pollutants_Available",
            "Pollutant_Exposure_Index"
        ]
    ].round(2)
)

city_index.to_csv(
    os.path.join(
        processed_dir,
        "composite_risk_with_pollutant_indicators.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: MULTI-YEAR HOTSPOT PERSISTENCE INTEGRATION
# PURPOSE: Incorporate repeated pollution hotspot appearances.
# METHOD: Merge the hotspot consistency percentage from
#         Submodule 6.3.
# OUTPUT: Consolidated city-level risk indicators.
# ============================================================

if "persistent_hotspot_summary" not in globals():
    raise NameError(
        "Run Submodule 6.3 Step 3 before continuing."
    )

persistence_index = persistent_hotspot_summary[
    [
        "City",
        "Eligible_Years",
        "Hotspot_Years",
        "Hotspot_Consistency_Percent"
    ]
].copy()

city_index = city_index.merge(
    persistence_index,
    on="City",
    how="left"
)

city_index["Hotspot_Consistency_Percent"] = (
    city_index["Hotspot_Consistency_Percent"].fillna(0)
)

city_index["Eligible_Years"] = (
    city_index["Eligible_Years"].fillna(0).astype(int)
)

city_index["Hotspot_Years"] = (
    city_index["Hotspot_Years"].fillna(0).astype(int)
)

print("MULTI-YEAR PERSISTENCE INTEGRATION")

display(
    city_index[
        [
            "City",
            "Eligible_Years",
            "Hotspot_Years",
            "Hotspot_Consistency_Percent"
        ]
    ].round(2)
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK MONITORING COVERAGE AUDIT
# PURPOSE: Assess the availability of pollutant measurements.
# METHOD: Calculate pollutant coverage across the 12
#         monitored pollutants for every city.
# OUTPUT: City-level monitoring coverage report.
# ============================================================

coverage_source = pollutant_data[
    ["City"] + pollutants
].copy()

city_coverage = (
    coverage_source.groupby("City")[pollutants]
    .apply(lambda group: group.notna().mean().mean() * 100)
    .rename("Pollutant_Monitoring_Coverage_Percent")
    .reset_index()
)

city_index = city_index.merge(
    city_coverage,
    on="City",
    how="left"
)

coverage_audit = city_index[
    [
        "City",
        "Valid_AQI_Days",
        "Pollutants_Available",
        "Pollutant_Monitoring_Coverage_Percent"
    ]
].copy()

print("MONITORING COVERAGE AUDIT")

display(coverage_audit.round(2))

coverage_audit.to_csv(
    os.path.join(
        report_dir,
        "composite_risk_monitoring_coverage_audit.csv"
    ),
    index=False
)

print("Coverage audit saved.")
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK INDICATOR NORMALIZATION
# PURPOSE: Convert different environmental indicators into
#          comparable percentile scores.
# METHOD: Apply ascending percentile ranking.
# OUTPUT: Normalized environmental risk indicators.
# ============================================================

risk_indicators = {
    "Average_AQI": "Average_AQI_Normalized",
    "High_Pollution_Rate": "High_Exposure_Normalized",
    "Extreme_AQI_Rate": "Extreme_Exposure_Normalized",
    "Pollutant_Exposure_Index": "Pollutant_Exposure_Normalized",
    "Hotspot_Consistency_Percent": "Persistence_Normalized"
}

for source_column, normalized_column in risk_indicators.items():

    city_index[normalized_column] = (
        city_index[source_column]
        .rank(method="average", pct=True) * 100
    )

print("NORMALIZED ENVIRONMENTAL RISK INDICATORS")

display(
    city_index[
        ["City"] + list(risk_indicators.values())
    ].round(2)
)

city_index.to_csv(
    os.path.join(
        processed_dir,
        "normalized_environmental_risk_indicators.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: COMPOSITE ENVIRONMENTAL RISK INDEX CALCULATION
# PURPOSE: Calculate one integrated environmental risk score
#          for every city.
# METHOD: Apply transparent weighted aggregation to the
#         normalized indicators.
# OUTPUT: City-level Composite Environmental Risk Index.
# ============================================================

risk_weights = {
    "Average_AQI_Normalized": 0.25,
    "High_Exposure_Normalized": 0.30,
    "Extreme_Exposure_Normalized": 0.15,
    "Pollutant_Exposure_Normalized": 0.20,
    "Persistence_Normalized": 0.10
}

city_index["Composite_Environmental_Risk_Index"] = sum(
    city_index[column] * weight
    for column, weight in risk_weights.items()
)

city_index["Environmental_Risk_Rank"] = (
    city_index["Composite_Environmental_Risk_Index"]
    .rank(method="min", ascending=False)
    .astype(int)
)

city_index = city_index.sort_values(
    "Composite_Environmental_Risk_Index",
    ascending=False
)

print("COMPOSITE ENVIRONMENTAL RISK INDEX")

display(
    city_index[
        [
            "City",
            "Composite_Environmental_Risk_Index",
            "Environmental_Risk_Rank"
        ]
    ].round(2)
)

city_index.to_csv(
    os.path.join(
        report_dir,
        "composite_environmental_risk_index.csv"
    ),
    index=False
)

print("Composite environmental risk index saved.")
```

```python

# ============================================================
# TOPIC: COMPOSITE ENVIRONMENTAL RISK CLASSIFICATION
# PURPOSE: Categorize cities according to their composite
#          environmental risk scores.
# METHOD: Apply predefined project-level score intervals.
# OUTPUT: City-level environmental risk classification.
# ============================================================

def classify_environmental_risk(score):

    if score >= 80:
        return "Very High"

    elif score >= 60:
        return "High"

    elif score >= 40:
        return "Moderate"

    elif score >= 20:
        return "Low"

    return "Very Low"


city_index["Environmental_Risk_Class"] = (
    city_index["Composite_Environmental_Risk_Index"]
    .apply(classify_environmental_risk)
)

print("ENVIRONMENTAL RISK CLASSIFICATION")

display(
    city_index[
        [
            "City",
            "Composite_Environmental_Risk_Index",
            "Environmental_Risk_Rank",
            "Environmental_Risk_Class"
        ]
    ].round(2)
)

city_index.to_csv(
    os.path.join(
        report_dir,
        "city_environmental_risk_classification.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: COMPOSITE ENVIRONMENTAL RISK RANKING VISUALIZATION
# PURPOSE: Compare integrated environmental risk scores
#          across cities.
# METHOD: Sort cities by their composite risk index.
# OUTPUT: Professional city-level risk ranking chart.
# ============================================================

plt.figure(figsize=(13, 9))

sns.barplot(
    data=city_index,
    x="Composite_Environmental_Risk_Index",
    y="City"
)

plt.title(
    "City-Wise Composite Environmental Risk Index",
    fontsize=16
)

plt.xlabel("Composite Environmental Risk Index (0-100)")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "composite_environmental_risk_ranking.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: MULTI-DIMENSIONAL ENVIRONMENTAL RISK COMPARISON
# PURPOSE: Compare the five normalized risk dimensions.
# METHOD: Plot a city-by-indicator heatmap.
# OUTPUT: Environmental risk dimension heatmap.
# ============================================================

risk_dimension_columns = list(risk_indicators.values())

risk_dimension_matrix = city_index.set_index("City")[
    risk_dimension_columns
].copy()

risk_dimension_matrix.columns = [
    "Average AQI",
    "High Exposure",
    "Extreme Exposure",
    "Pollutant Exposure",
    "Persistence"
]

plt.figure(figsize=(13, 10))

sns.heatmap(
    risk_dimension_matrix,
    cmap="YlOrRd",
    annot=True,
    fmt=".1f",
    linewidths=0.4,
    vmin=0,
    vmax=100
)

plt.title(
    "Multi-Dimensional Environmental Risk Profile",
    fontsize=16
)

plt.xlabel("Environmental Risk Dimension")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "environmental_risk_dimension_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK RANK STABILITY
# PURPOSE: Evaluate agreement between alternative index
#          weighting scenarios.
# METHOD: Calculate Spearman rank correlations.
# OUTPUT: Risk ranking stability matrix.
# ============================================================

rank_columns = [
    "Baseline_Rank",
    "Equal_Weights_Rank",
    "Exposure_Focused_Rank"
]

rank_stability = sensitivity_results[
    rank_columns
].corr(method="spearman")

rank_stability.index = [
    "Baseline",
    "Equal Weights",
    "Exposure Focused"
]

rank_stability.columns = [
    "Baseline",
    "Equal Weights",
    "Exposure Focused"
]

print("RANKING STABILITY MATRIX")

display(rank_stability.round(3))

rank_stability.to_csv(
    os.path.join(
        report_dir,
        "composite_risk_rank_stability.csv"
    )
)
```

```python

# ============================================================
# TOPIC: REBUILD COMPOSITE RISK WEIGHT SENSITIVITY RESULTS
# PURPOSE: Restore the missing sensitivity_results variable.
# METHOD: Recalculate three alternative index weighting scenarios.
# OUTPUT: sensitivity_results DataFrame.
# ============================================================

normalized_columns = [
    "Average_AQI_Normalized",
    "High_Exposure_Normalized",
    "Extreme_Exposure_Normalized",
    "Pollutant_Exposure_Normalized",
    "Persistence_Normalized"
]

scenario_weights = {
    "Baseline": {
        "Average_AQI_Normalized": 0.25,
        "High_Exposure_Normalized": 0.30,
        "Extreme_Exposure_Normalized": 0.15,
        "Pollutant_Exposure_Normalized": 0.20,
        "Persistence_Normalized": 0.10
    },
    "Equal_Weights": {
        column: 0.20 for column in normalized_columns
    },
    "Exposure_Focused": {
        "Average_AQI_Normalized": 0.15,
        "High_Exposure_Normalized": 0.35,
        "Extreme_Exposure_Normalized": 0.25,
        "Pollutant_Exposure_Normalized": 0.15,
        "Persistence_Normalized": 0.10
    }
}

sensitivity_results = city_index[["City"]].copy()

for scenario, weights in scenario_weights.items():

    sensitivity_results[scenario] = sum(
        city_index[column] * weight
        for column, weight in weights.items()
    )

    sensitivity_results[f"{scenario}_Rank"] = (
        sensitivity_results[scenario]
        .rank(method="min", ascending=False)
        .astype(int)
    )

print("Sensitivity analysis successfully recreated.")
print("Shape:", sensitivity_results.shape)

display(
    sensitivity_results.sort_values(
        "Baseline",
        ascending=False
    ).round(2)
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK RANK STABILITY
# PURPOSE: Evaluate agreement between alternative index
#          weighting scenarios.
# METHOD: Calculate Spearman rank correlations.
# OUTPUT: Risk ranking stability matrix.
# ============================================================

rank_columns = [
    "Baseline_Rank",
    "Equal_Weights_Rank",
    "Exposure_Focused_Rank"
]

rank_stability = sensitivity_results[
    rank_columns
].corr(method="spearman")

rank_stability.index = [
    "Baseline",
    "Equal Weights",
    "Exposure Focused"
]

rank_stability.columns = [
    "Baseline",
    "Equal Weights",
    "Exposure Focused"
]

print("RISK RANKING STABILITY MATRIX")

display(rank_stability.round(3))

rank_stability.to_csv(
    os.path.join(
        report_dir,
        "composite_risk_rank_stability.csv"
    )
)

print("Risk ranking stability report saved.")
```

```python

# ============================================================
# TOPIC: FINAL ENVIRONMENTAL RISK INDEX EXPORT
# PURPOSE: Prepare the complete city-level composite risk
#          dataset for project reporting.
# METHOD: Consolidate risk scores, indicators, rankings,
#         classifications and monitoring coverage.
# OUTPUT: Final environmental risk index CSV files.
# ============================================================

final_risk_columns = [
    "City",
    "Valid_AQI_Days",
    "Average_AQI",
    "High_Pollution_Rate",
    "Extreme_AQI_Rate",
    "Pollutants_Available",
    "Pollutant_Monitoring_Coverage_Percent",
    "Pollutant_Exposure_Index",
    "Eligible_Years",
    "Hotspot_Years",
    "Hotspot_Consistency_Percent",
    "Composite_Environmental_Risk_Index",
    "Environmental_Risk_Rank",
    "Environmental_Risk_Class"
]

final_environmental_risk_index = (
    city_index[final_risk_columns].copy()
)

final_environmental_risk_index = (
    final_environmental_risk_index.sort_values(
        "Environmental_Risk_Rank"
    )
)

final_environmental_risk_index.to_csv(
    os.path.join(
        processed_dir,
        "final_environmental_risk_index.csv"
    ),
    index=False
)

final_environmental_risk_index.to_csv(
    os.path.join(
        report_dir,
        "final_environmental_risk_index_report.csv"
    ),
    index=False
)

print("FINAL ENVIRONMENTAL RISK INDEX CREATED")
print("Number of cities:", len(final_environmental_risk_index))

display(
    final_environmental_risk_index.round(2)
)
```

```python

# ============================================================
# TOPIC: COMPOSITE ENVIRONMENTAL RISK RANKING CHART
# PURPOSE: Display the overall environmental risk ranking
#          of all analyzed cities.
# METHOD: Sort cities using the composite risk index.
# OUTPUT: Final city ranking visualization.
# ============================================================

risk_plot_data = final_environmental_risk_index.sort_values(
    "Composite_Environmental_Risk_Index",
    ascending=True
)

plt.figure(figsize=(13, 10))

sns.barplot(
    data=risk_plot_data,
    x="Composite_Environmental_Risk_Index",
    y="City"
)

plt.title(
    "Urban Air Quality — Composite Environmental Risk Ranking",
    fontsize=16
)

plt.xlabel("Composite Environmental Risk Index (0–100)")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "01_composite_environmental_risk_ranking.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK CLASS DISTRIBUTION
# PURPOSE: Examine the distribution of cities across risk classes.
# METHOD: Count cities in each composite risk classification.
# OUTPUT: Environmental risk classification chart.
# ============================================================

risk_class_order = [
    "Very Low",
    "Low",
    "Moderate",
    "High",
    "Very High"
]

risk_class_distribution = (
    final_environmental_risk_index[
        "Environmental_Risk_Class"
    ]
    .value_counts()
    .reindex(risk_class_order, fill_value=0)
    .rename_axis("Risk_Class")
    .reset_index(name="City_Count")
)

display(risk_class_distribution)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=risk_class_distribution,
    x="Risk_Class",
    y="City_Count",
    order=risk_class_order
)

plt.title(
    "Distribution of Cities by Environmental Risk Class",
    fontsize=15
)

plt.xlabel("Environmental Risk Class")
plt.ylabel("Number of Cities")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "02_environmental_risk_class_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: MULTI-DIMENSIONAL ENVIRONMENTAL RISK HEATMAP
# PURPOSE: Compare normalized risk dimensions across cities.
# METHOD: Plot the five normalized environmental indicators.
# OUTPUT: City-level multi-dimensional risk heatmap.
# ============================================================

risk_dimension_columns = [
    "Average_AQI_Normalized",
    "High_Exposure_Normalized",
    "Extreme_Exposure_Normalized",
    "Pollutant_Exposure_Normalized",
    "Persistence_Normalized"
]

risk_dimension_matrix = city_index.set_index("City")[
    risk_dimension_columns
].copy()

risk_dimension_matrix.columns = [
    "Average AQI",
    "High Exposure",
    "Extreme Exposure",
    "Pollutant Exposure",
    "Persistence"
]

plt.figure(figsize=(14, 10))

sns.heatmap(
    risk_dimension_matrix,
    cmap="YlOrRd",
    annot=True,
    fmt=".1f",
    linewidths=0.4,
    vmin=0,
    vmax=100
)

plt.title(
    "Multi-Dimensional Environmental Risk Profile",
    fontsize=16
)

plt.xlabel("Environmental Risk Dimension")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "03_multi_dimensional_environmental_risk_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: HIGH AND EXTREME POLLUTION EXPOSURE COMPARISON
# PURPOSE: Examine the relationship between ordinary
#          high-pollution and extreme-pollution exposure.
# METHOD: Create a city-level scatter plot.
# OUTPUT: Environmental exposure comparison chart.
# ============================================================

plt.figure(figsize=(11, 8))

sns.scatterplot(
    data=final_environmental_risk_index,
    x="High_Pollution_Rate",
    y="Extreme_AQI_Rate",
    size="Valid_AQI_Days",
    hue="Environmental_Risk_Class",
    sizes=(60, 400)
)

plt.title(
    "High Pollution vs Extreme Pollution Exposure",
    fontsize=15
)

plt.xlabel("High Pollution Exposure Rate (%)")
plt.ylabel("Extreme AQI Exposure Rate (%)")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "04_high_vs_extreme_pollution_exposure.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK RANKING SCENARIO COMPARISON
# PURPOSE: Visualize how alternative index weights affect
#          city-level risk rankings.
# METHOD: Compare the three sensitivity scenarios.
# OUTPUT: Risk ranking sensitivity visualization.
# ============================================================

scenario_rank_data = sensitivity_results.melt(
    id_vars="City",
    value_vars=[
        "Baseline_Rank",
        "Equal_Weights_Rank",
        "Exposure_Focused_Rank"
    ],
    var_name="Scenario",
    value_name="Risk_Rank"
)

scenario_rank_data["Scenario"] = (
    scenario_rank_data["Scenario"].replace({
        "Baseline_Rank": "Baseline",
        "Equal_Weights_Rank": "Equal Weights",
        "Exposure_Focused_Rank": "Exposure Focused"
    })
)

plt.figure(figsize=(14, 9))

sns.lineplot(
    data=scenario_rank_data,
    x="Scenario",
    y="Risk_Rank",
    hue="City",
    marker="o",
    legend=False
)

plt.title(
    "Environmental Risk Ranking Across Weighting Scenarios",
    fontsize=15
)

plt.xlabel("Weighting Scenario")
plt.ylabel("City Risk Rank")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "05_environmental_risk_scenario_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: CITY-WISE ENVIRONMENTAL RISK INDICATOR COMPARISON
# PURPOSE: Compare average AQI, exposure rates and pollutant
#          exposure across cities.
# METHOD: Create a normalized indicator comparison chart.
# OUTPUT: Environmental risk indicator comparison.
# ============================================================

indicator_comparison = city_index[
    [
        "City",
        "Average_AQI_Normalized",
        "High_Exposure_Normalized",
        "Extreme_Exposure_Normalized",
        "Pollutant_Exposure_Normalized",
        "Persistence_Normalized"
    ]
].melt(
    id_vars="City",
    var_name="Risk_Indicator",
    value_name="Normalized_Score"
)

plt.figure(figsize=(15, 8))

sns.barplot(
    data=indicator_comparison,
    x="Risk_Indicator",
    y="Normalized_Score",
    hue="City",
    errorbar=None
)

plt.title(
    "Environmental Risk Indicator Comparison",
    fontsize=15
)

plt.xlabel("Risk Indicator")
plt.ylabel("Normalized Score (0–100)")
plt.xticks(rotation=20)

plt.legend(
    title="City",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "06_environmental_risk_indicator_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK VISUALIZATION DATA EXPORT
# PURPOSE: Save the underlying data used for the final charts.
# METHOD: Consolidate risk rankings, classifications and
#         indicator scores.
# OUTPUT: Visualization-ready environmental risk dataset.
# ============================================================

visualization_summary = city_index[
    [
        "City",
        "Composite_Environmental_Risk_Index",
        "Environmental_Risk_Rank",
        "Environmental_Risk_Class",
        "Average_AQI_Normalized",
        "High_Exposure_Normalized",
        "Extreme_Exposure_Normalized",
        "Pollutant_Exposure_Normalized",
        "Persistence_Normalized"
    ]
].copy()

visualization_summary.to_csv(
    os.path.join(
        processed_dir,
        "environmental_risk_visualization_data.csv"
    ),
    index=False
)

print("Visualization summary exported.")
display(visualization_summary.round(2))
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK VISUALIZATION FILE AUDIT
# PURPOSE: Verify that the required Module 6 charts were saved.
# METHOD: Check each expected image filename.
# OUTPUT: Visualization completion audit.
# ============================================================

expected_visualizations = [
    "01_composite_environmental_risk_ranking.png",
    "02_environmental_risk_class_distribution.png",
    "03_multi_dimensional_environmental_risk_heatmap.png",
    "04_high_vs_extreme_pollution_exposure.png",
    "05_environmental_risk_scenario_comparison.png",
    "06_environmental_risk_indicator_comparison.png"
]

visualization_audit = pd.DataFrame({
    "Visualization_File": expected_visualizations,
    "File_Exists": [
        os.path.exists(
            os.path.join(visualization_dir, filename)
        )
        for filename in expected_visualizations
    ]
})

display(visualization_audit)

print(
    "Charts successfully saved:",
    visualization_audit["File_Exists"].sum(),
    "of",
    len(expected_visualizations)
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL INTELLIGENCE REPORT DATA PREPARATION
# PURPOSE: Consolidate the main analytical outputs from
#          Advanced Module 6.
# METHOD: Validate the availability of previous result datasets.
# OUTPUT: Report-ready environmental risk dataset.
# ============================================================

required_variables = [
    "final_environmental_risk_index",
    "city_index",
    "annual_city_risk",
    "city_pollutant_exposure",
    "extreme_aqi_records"
]

missing_variables = [
    variable for variable in required_variables
    if variable not in globals()
]

print("ENVIRONMENTAL REPORT DATA AUDIT")

if missing_variables:
    print("Missing variables:", missing_variables)
else:
    print("All primary report datasets are available.")

display(
    final_environmental_risk_index.head()
)
```

```python


# ============================================================
# TOPIC: CITY ENVIRONMENTAL RISK LEADERBOARD
# PURPOSE: Identify cities with the highest and lowest
#          composite environmental risk scores.
# METHOD: Rank cities using the final composite risk index.
# OUTPUT: Highest-risk and lowest-risk city report.
# ============================================================

city_risk_leaderboard = (
    final_environmental_risk_index.sort_values(
        "Composite_Environmental_Risk_Index",
        ascending=False
    ).copy()
)

top_risk_cities = city_risk_leaderboard.head(10)

lowest_risk_cities = (
    city_risk_leaderboard.tail(10)
    .sort_values("Composite_Environmental_Risk_Index")
)

print("TOP 10 ENVIRONMENTAL RISK CITIES")

display(
    top_risk_cities[
        [
            "City",
            "Composite_Environmental_Risk_Index",
            "Environmental_Risk_Rank",
            "Environmental_Risk_Class"
        ]
    ].round(2)
)

print("10 CITIES WITH LOWEST OBSERVED COMPOSITE RISK")

display(
    lowest_risk_cities[
        [
            "City",
            "Composite_Environmental_Risk_Index",
            "Environmental_Risk_Rank",
            "Environmental_Risk_Class"
        ]
    ].round(2)
)

top_risk_cities.to_csv(
    os.path.join(
        report_dir,
        "top_10_environmental_risk_cities.csv"
    ),
    index=False
)

lowest_risk_cities.to_csv(
    os.path.join(
        report_dir,
        "lowest_10_environmental_risk_cities.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL RISK CLASSIFICATION SUMMARY
# PURPOSE: Summarize the number and percentage of cities
#          in each composite environmental risk class.
# METHOD: Aggregate city classifications.
# OUTPUT: Environmental risk classification summary.
# ============================================================

risk_class_summary = (
    final_environmental_risk_index.groupby(
        "Environmental_Risk_Class"
    )
    .agg(
        City_Count=("City", "nunique"),
        Average_Risk_Score=(
            "Composite_Environmental_Risk_Index", "mean"
        )
    )
    .reset_index()
)

risk_class_summary["City_Percentage"] = (
    risk_class_summary["City_Count"]
    / len(final_environmental_risk_index) * 100
)

print("ENVIRONMENTAL RISK CLASS SUMMARY")

display(risk_class_summary.round(2))

risk_class_summary.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_class_summary.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: PERSISTENT ENVIRONMENTAL HOTSPOT SUMMARY
# PURPOSE: Summarize cities repeatedly appearing among
#          annual pollution hotspots.
# METHOD: Rank cities by hotspot consistency and eligible years.
# OUTPUT: Persistent environmental hotspot report.
# ============================================================

persistent_report = persistent_hotspot_summary.copy()

persistent_report = persistent_report.sort_values(
    [
        "Hotspot_Consistency_Percent",
        "Hotspot_Years"
    ],
    ascending=False
)

print("PERSISTENT POLLUTION HOTSPOTS")

display(
    persistent_report[
        [
            "City",
            "Eligible_Years",
            "Hotspot_Years",
            "Hotspot_Consistency_Percent",
            "Persistence_Class"
        ]
    ].head(15).round(2)
)

persistent_report.to_csv(
    os.path.join(
        report_dir,
        "environmental_persistent_hotspot_findings.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: SEASONAL ENVIRONMENTAL RISK SUMMARY
# PURPOSE: Identify seasons associated with higher city-level
#          pollution exposure.
# METHOD: Summarize city-season exposure rates.
# OUTPUT: Seasonal environmental intelligence report.
# ============================================================

seasonal_report = (
    city_season_risk.groupby("Season")
    .agg(
        Cities_Analyzed=("City", "nunique"),
        Average_Exposure_Rate=(
            "High_Pollution_Rate_Percent", "mean"
        ),
        Average_AQI=("Average_AQI", "mean"),
        Maximum_AQI=("Maximum_AQI", "max")
    )
    .reset_index()
)

seasonal_report = seasonal_report.sort_values(
    "Average_Exposure_Rate",
    ascending=False
)

print("SEASONAL ENVIRONMENTAL RISK FINDINGS")

display(seasonal_report.round(2))

seasonal_report.to_csv(
    os.path.join(
        report_dir,
        "environmental_seasonal_risk_findings.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: EXTREME POLLUTION INTELLIGENCE SUMMARY
# PURPOSE: Consolidate the frequency, severity and city
#          distribution of AQI observations above 500.
# METHOD: Summarize extreme AQI records and affected cities.
# OUTPUT: Extreme pollution executive findings.
# ============================================================

extreme_city_summary = (
    extreme_aqi_records.groupby("City")
    .agg(
        Extreme_Observations=("AQI", "count"),
        Average_Extreme_AQI=("AQI", "mean"),
        Maximum_Extreme_AQI=("AQI", "max")
    )
    .reset_index()
    .sort_values(
        "Extreme_Observations",
        ascending=False
    )
)

extreme_summary = pd.DataFrame({
    "Indicator": [
        "Total Extreme AQI Observations",
        "Cities With Extreme Observations",
        "Average Extreme AQI",
        "Maximum Recorded AQI"
    ],
    "Value": [
        len(extreme_aqi_records),
        extreme_aqi_records["City"].nunique(),
        round(extreme_aqi_records["AQI"].mean(), 2),
        extreme_aqi_records["AQI"].max()
    ]
})

print("EXTREME POLLUTION SUMMARY")

display(extreme_summary)
display(extreme_city_summary.head(10))

extreme_summary.to_csv(
    os.path.join(
        report_dir,
        "extreme_pollution_executive_summary.csv"
    ),
    index=False
)

extreme_city_summary.to_csv(
    os.path.join(
        report_dir,
        "extreme_pollution_city_findings.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: POLLUTANT-SPECIFIC ENVIRONMENTAL FINDINGS
# PURPOSE: Identify pollutants and cities with comparatively
#          high observed concentration exposure.
# METHOD: Rank pollutant-specific city exposure rates.
# OUTPUT: Pollutant intelligence findings report.
# ============================================================

pollutant_findings = (
    city_pollutant_exposure.sort_values(
        ["Pollutant", "High_Concentration_Rate_Percent"],
        ascending=[True, False]
    )
    .groupby("Pollutant", group_keys=False)
    .head(3)
    .copy()
)

print("TOP THREE CITY EXPOSURE RATES BY POLLUTANT")

display(
    pollutant_findings[
        [
            "Pollutant",
            "City",
            "Valid_Observations",
            "High_Concentration_Days",
            "High_Concentration_Rate_Percent"
        ]
    ].round(2)
)

pollutant_findings.to_csv(
    os.path.join(
        report_dir,
        "pollutant_specific_environmental_findings.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: EVIDENCE-BASED ENVIRONMENTAL RECOMMENDATIONS
# PURPOSE: Translate measured environmental risk indicators
#          into practical monitoring and planning priorities.
# METHOD: Apply transparent rules to observed risk indicators.
# OUTPUT: City-wise environmental recommendation report.
# ============================================================

recommendation_records = []

for _, row in final_environmental_risk_index.iterrows():

    recommendations = []

    if row["High_Pollution_Rate"] >= 25:
        recommendations.append(
            "Prioritize investigation of recurring high-AQI days"
        )

    if row["Extreme_AQI_Rate"] > 0:
        recommendations.append(
            "Review extreme AQI records and strengthen monitoring"
        )

    if row["Hotspot_Consistency_Percent"] >= 50:
        recommendations.append(
            "Investigate repeated multi-year pollution hotspots"
        )

    if row["Pollutant_Monitoring_Coverage_Percent"] < 50:
        recommendations.append(
            "Improve pollutant monitoring coverage"
        )

    if not recommendations:
        recommendations.append(
            "Continue monitoring and periodic environmental assessment"
        )

    recommendation_records.append({
        "City": row["City"],
        "Environmental_Risk_Score": (
            row["Composite_Environmental_Risk_Index"]
        ),
        "Risk_Class": row["Environmental_Risk_Class"],
        "Recommendations": "; ".join(recommendations)
    })

environmental_recommendations = pd.DataFrame(
    recommendation_records
)

display(environmental_recommendations)

environmental_recommendations.to_csv(
    os.path.join(
        report_dir,
        "city_wise_environmental_recommendations.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL INTELLIGENCE EXECUTIVE SUMMARY
# PURPOSE: Consolidate the major findings from Advanced Module 6.
# METHOD: Extract the highest and lowest risk indicators,
#         city coverage and extreme pollution statistics.
# OUTPUT: Executive summary CSV report.
# ============================================================

highest_risk_city = (
    final_environmental_risk_index.iloc[0]
)

lowest_risk_city = (
    final_environmental_risk_index.iloc[-1]
)

executive_summary = pd.DataFrame({
    "Metric": [
        "Cities Analyzed",
        "Highest Risk City",
        "Highest Composite Risk Score",
        "Lowest Risk City",
        "Lowest Composite Risk Score",
        "Total Extreme AQI Observations",
        "Cities With Extreme AQI Observations",
        "Highest Persistent Hotspot Consistency"
    ],
    "Result": [
        len(final_environmental_risk_index),
        highest_risk_city["City"],
        round(
            highest_risk_city[
                "Composite_Environmental_Risk_Index"
            ], 2
        ),
        lowest_risk_city["City"],
        round(
            lowest_risk_city[
                "Composite_Environmental_Risk_Index"
            ], 2
        ),
        len(extreme_aqi_records),
        extreme_aqi_records["City"].nunique(),
        round(
            persistent_hotspot_summary[
                "Hotspot_Consistency_Percent"
            ].max(), 2
        )
    ]
})

print("ENVIRONMENTAL INTELLIGENCE EXECUTIVE SUMMARY")

display(executive_summary)

executive_summary.to_csv(
    os.path.join(
        report_dir,
        "environmental_intelligence_executive_summary.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL INTELLIGENCE REPORT INVENTORY
# PURPOSE: Verify the main report files generated by Module 6.
# METHOD: Check expected report filenames in reports directory.
# OUTPUT: Final report completion audit.
# ============================================================

expected_reports = [
    "final_environmental_risk_index_report.csv",
    "top_10_environmental_risk_cities.csv",
    "lowest_10_environmental_risk_cities.csv",
    "environmental_risk_class_summary.csv",
    "environmental_persistent_hotspot_findings.csv",
    "environmental_seasonal_risk_findings.csv",
    "extreme_pollution_executive_summary.csv",
    "extreme_pollution_city_findings.csv",
    "pollutant_specific_environmental_findings.csv",
    "city_wise_environmental_recommendations.csv",
    "environmental_intelligence_executive_summary.csv"
]

report_audit = pd.DataFrame({
    "Report_File": expected_reports,
    "Exists": [
        os.path.exists(os.path.join(report_dir, filename))
        for filename in expected_reports
    ]
})

display(report_audit)

print(
    "Reports found:",
    report_audit["Exists"].sum(),
    "of",
    len(report_audit)
)

report_audit.to_csv(
    os.path.join(
        report_dir,
        "module_6_report_completion_audit.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: MODULE 6 VARIABLE AVAILABILITY AUDIT
# PURPOSE: Verify that the major Module 6 analytical outputs
#          are available in the current notebook session.
# METHOD: Check required variables using globals().
# OUTPUT: Variable availability audit.
# ============================================================

module_6_variables = [
    "air_final",
    "city_risk_profile",
    "city_season_risk",
    "annual_city_risk",
    "eligible_annual_risk",
    "persistent_hotspot_summary",
    "extreme_aqi_records",
    "city_extreme_risk",
    "city_pollutant_exposure",
    "pollutant_risk_summary",
    "city_index",
    "sensitivity_results",
    "rank_stability",
    "final_environmental_risk_index",
    "environmental_recommendations",
    "executive_summary"
]

module_6_variable_audit = pd.DataFrame({
    "Variable": module_6_variables,
    "Available": [
        variable in globals()
        for variable in module_6_variables
    ]
})

display(module_6_variable_audit)

print(
    "Available:",
    module_6_variable_audit["Available"].sum(),
    "of",
    len(module_6_variable_audit)
)
```

```python

# ============================================================
# TOPIC: COMPOSITE RISK INDEX VALIDATION
# PURPOSE: Check the structural and numerical validity of
#          the final environmental risk index.
# METHOD: Inspect duplicates, missing values and score ranges.
# OUTPUT: Composite risk index validation report.
# ============================================================

risk_validation = {
    "Total_Cities": len(final_environmental_risk_index),
    "Unique_Cities": final_environmental_risk_index["City"].nunique(),
    "Duplicate_Cities": final_environmental_risk_index["City"].duplicated().sum(),
    "Missing_Risk_Scores": final_environmental_risk_index[
        "Composite_Environmental_Risk_Index"
    ].isna().sum(),
    "Minimum_Risk_Score": final_environmental_risk_index[
        "Composite_Environmental_Risk_Index"
    ].min(),
    "Maximum_Risk_Score": final_environmental_risk_index[
        "Composite_Environmental_Risk_Index"
    ].max(),
    "Missing_Risk_Classes": final_environmental_risk_index[
        "Environmental_Risk_Class"
    ].isna().sum()
}

risk_validation_report = pd.DataFrame(
    list(risk_validation.items()),
    columns=["Validation_Check", "Result"]
)

display(risk_validation_report)

risk_validation_report.to_csv(
    os.path.join(
        report_dir,
        "module_6_composite_index_validation.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: COMPOSITE RISK WEIGHT VALIDATION
# PURPOSE: Confirm that the five index weights sum to 100%.
# METHOD: Calculate the total and verify individual weights.
# OUTPUT: Risk index weight validation report.
# ============================================================

weight_validation = pd.DataFrame({
    "Indicator": list(risk_weights.keys()),
    "Weight": list(risk_weights.values())
})

weight_validation["Weight_Percent"] = (
    weight_validation["Weight"] * 100
)

total_weight = weight_validation["Weight"].sum()

print("TOTAL INDEX WEIGHT:", total_weight)

display(weight_validation)

print(
    "Weight validation:",
    "PASS" if np.isclose(total_weight, 1.0) else "CHECK REQUIRED"
)
```

```python

# ============================================================
# TOPIC: ENVIRONMENTAL MONITORING COVERAGE VALIDATION
# PURPOSE: Identify cities with limited pollutant measurement
#          availability.
# METHOD: Group cities by monitoring coverage intervals.
# OUTPUT: Monitoring coverage quality report.
# ============================================================

coverage_validation = (
    final_environmental_risk_index[
        [
            "City",
            "Valid_AQI_Days",
            "Pollutants_Available",
            "Pollutant_Monitoring_Coverage_Percent"
        ]
    ].copy()
)

coverage_validation["Coverage_Status"] = pd.cut(
    coverage_validation["Pollutant_Monitoring_Coverage_Percent"],
    bins=[-np.inf, 25, 50, 75, 100],
    labels=[
        "Very Limited",
        "Limited",
        "Moderate",
        "Relatively High"
    ]
)

display(
    coverage_validation.sort_values(
        "Pollutant_Monitoring_Coverage_Percent"
    ).round(2)
)

coverage_validation.to_csv(
    os.path.join(
        report_dir,
        "module_6_monitoring_coverage_validation.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: CITY-CLUSTER BOOTSTRAP DATA PREPARATION
# PURPOSE: Prepare pollutant and AQI observations for
#          city-level cluster resampling.
# METHOD: Retain complete pollutant-AQI pairs and organize
#         observations by city.
# OUTPUT: Cluster bootstrap analytical dataset.
# ============================================================

from scipy.stats import spearmanr
import numpy as np
import pandas as pd
import os

cluster_pollutants = [
    "PM2.5", "PM10",
    "NO", "NO2", "NOx", "NH3",
    "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

cluster_data = air_final[
    ["City", "AQI"] + cluster_pollutants
].copy()

cluster_data["AQI"] = pd.to_numeric(
    cluster_data["AQI"],
    errors="coerce"
)

for pollutant in cluster_pollutants:
    cluster_data[pollutant] = pd.to_numeric(
        cluster_data[pollutant],
        errors="coerce"
    )

cluster_data = cluster_data.replace(
    [np.inf, -np.inf],
    np.nan
)

cluster_cities = sorted(
    cluster_data["City"].dropna().unique()
)

print("CITY-CLUSTER BOOTSTRAP PREPARATION")
print("Number of city clusters:", len(cluster_cities))
print("Total source records:", len(cluster_data))

display(
    cluster_data.groupby("City").size()
    .describe()
    .to_frame("Observations_Per_City")
)
```

```python

# ============================================================
# TOPIC: ORIGINAL POLLUTANT-AQI CORRELATION BASELINE
# PURPOSE: Establish reference Spearman correlations.
# METHOD: Calculate pooled correlations using complete pairs.
# OUTPUT: Baseline pollutant-AQI correlation report.
# ============================================================

cluster_baseline_results = []

for pollutant in cluster_pollutants:

    temp = cluster_data[
        ["AQI", pollutant]
    ].dropna()

    correlation, p_value = spearmanr(
        temp["AQI"],
        temp[pollutant]
    )

    cluster_baseline_results.append({
        "Pollutant": pollutant,
        "Valid_Observations": len(temp),
        "Original_Spearman": correlation,
        "Original_P_Value": p_value
    })

cluster_baseline_df = pd.DataFrame(
    cluster_baseline_results
)

display(cluster_baseline_df.round(5))
```

```python

# ============================================================
# TOPIC: CITY-CLUSTER BOOTSTRAP CORRELATION ANALYSIS
# PURPOSE: Estimate uncertainty in pollutant-AQI correlations
#          while preserving within-city observation clusters.
# METHOD: Resample entire cities with replacement and
#         recalculate pooled Spearman correlations.
# OUTPUT: Cluster-bootstrap correlation distributions.
# ============================================================

N_BOOTSTRAPS = 500

rng = np.random.default_rng(42)

cluster_bootstrap_results = []

for pollutant in cluster_pollutants:

    city_groups = {}

    for city in cluster_cities:

        city_group = cluster_data.loc[
            cluster_data["City"] == city,
            ["AQI", pollutant]
        ].dropna()

        if len(city_group) >= 2:
            city_groups[city] = city_group

    eligible_cities = list(city_groups.keys())

    bootstrap_correlations = []

    for iteration in range(N_BOOTSTRAPS):

        sampled_cities = rng.choice(
            eligible_cities,
            size=len(eligible_cities),
            replace=True
        )

        sampled_frames = [
            city_groups[city]
            for city in sampled_cities
        ]

        bootstrap_sample = pd.concat(
            sampled_frames,
            ignore_index=True
        )

        if (
            bootstrap_sample["AQI"].nunique() < 2
            or bootstrap_sample[pollutant].nunique() < 2
        ):
            continue

        correlation, _ = spearmanr(
            bootstrap_sample["AQI"],
            bootstrap_sample[pollutant]
        )

        if np.isfinite(correlation):
            bootstrap_correlations.append(correlation)

    cluster_bootstrap_results.append({
        "Pollutant": pollutant,
        "Eligible_City_Clusters": len(eligible_cities),
        "Successful_Bootstraps": len(bootstrap_correlations),
        "Bootstrap_Correlations": bootstrap_correlations
    })

print("City-cluster bootstrap calculations completed.")

display(
    pd.DataFrame([
        {
            "Pollutant": result["Pollutant"],
            "Eligible_City_Clusters": result["Eligible_City_Clusters"],
            "Successful_Bootstraps": result["Successful_Bootstraps"]
        }
        for result in cluster_bootstrap_results
    ])
)
```

```python

# ============================================================
# TOPIC: CITY-CLUSTER BOOTSTRAP CONFIDENCE INTERVALS
# PURPOSE: Estimate 95% confidence intervals for pooled
#          pollutant-AQI Spearman correlations.
# METHOD: Use the percentile method on bootstrap correlations.
# OUTPUT: Cluster-bootstrap confidence interval report.
# ============================================================

cluster_ci_records = []

for result in cluster_bootstrap_results:

    correlations = np.asarray(
        result["Bootstrap_Correlations"]
    )

    cluster_ci_records.append({
        "Pollutant": result["Pollutant"],
        "Eligible_City_Clusters": (
            result["Eligible_City_Clusters"]
        ),
        "Successful_Bootstraps": len(correlations),
        "Cluster_CI_Lower": (
            np.percentile(correlations, 2.5)
            if len(correlations) else np.nan
        ),
        "Cluster_CI_Upper": (
            np.percentile(correlations, 97.5)
            if len(correlations) else np.nan
        ),
        "Cluster_Bootstrap_Median": (
            np.median(correlations)
            if len(correlations) else np.nan
        )
    })

cluster_ci_df = pd.DataFrame(cluster_ci_records)

cluster_ci_df = cluster_baseline_df.merge(
    cluster_ci_df,
    on="Pollutant",
    how="left"
)

display(cluster_ci_df.round(5))

cluster_ci_df.to_csv(
    os.path.join(
        report_dir,
        "city_cluster_bootstrap_correlation_intervals.csv"
    ),
    index=False
)

print("Cluster-bootstrap confidence intervals saved.")
```

```python

# ============================================================
# TOPIC: ROW-LEVEL VS CITY-CLUSTER BOOTSTRAP COMPARISON
# PURPOSE: Compare uncertainty estimates from two bootstrap
#          sampling strategies.
# METHOD: Merge previously calculated row-level intervals
#         with the new city-cluster intervals.
# OUTPUT: Bootstrap robustness comparison report.
# ============================================================

if "bootstrap_ci_df" not in globals():
    raise NameError(
        "Previous row-level bootstrap_ci_df is unavailable. "
        "Restore the original bootstrap results before comparison."
    )

print("ROW-LEVEL BOOTSTRAP COLUMNS:")
print(bootstrap_ci_df.columns.tolist())

print("\nCITY-CLUSTER BOOTSTRAP COLUMNS:")
print(cluster_ci_df.columns.tolist())

display(cluster_ci_df)
```

```python

# ============================================================
# TOPIC: CITY-WISE DAILY AQI TIME-SERIES PREPARATION
# PURPOSE: Prepare AQI observations for temporal dependence
#          analysis within individual cities.
# METHOD: Sort observations by city and date and identify
#         consecutive calendar-day observations.
# OUTPUT: Consecutive-day AQI analytical dataset.
# ============================================================

from scipy.stats import pearsonr, spearmanr
from statsmodels.stats.multitest import multipletests

temporal_data = air_final[
    ["City", "Date", "AQI"]
].copy()

temporal_data["Date"] = pd.to_datetime(
    temporal_data["Date"],
    errors="coerce"
)

temporal_data["AQI"] = pd.to_numeric(
    temporal_data["AQI"],
    errors="coerce"
)

temporal_data = temporal_data.dropna(
    subset=["City", "Date", "AQI"]
).copy()

temporal_data = temporal_data.sort_values(
    ["City", "Date"]
)

# Create the previous observed AQI and date within each city.
temporal_data["Previous_AQI"] = (
    temporal_data.groupby("City")["AQI"].shift(1)
)

temporal_data["Previous_Date"] = (
    temporal_data.groupby("City")["Date"].shift(1)
)

temporal_data["Date_Gap_Days"] = (
    temporal_data["Date"] - temporal_data["Previous_Date"]
).dt.days

# Retain only pairs exactly one calendar day apart.
consecutive_aqi = temporal_data[
    temporal_data["Date_Gap_Days"] == 1
].copy()

consecutive_aqi = consecutive_aqi.rename(
    columns={"AQI": "Current_AQI"}
)

print("TEMPORAL DATA PREPARATION")
print("Valid AQI observations:", len(temporal_data))
print("Consecutive-day pairs:", len(consecutive_aqi))
print("Cities:", temporal_data["City"].nunique())

display(consecutive_aqi.head())
```

```python

# ============================================================
# TOPIC: CITY-WISE LAG-1 AQI AUTOCORRELATION
# PURPOSE: Measure temporal relationships between consecutive
#          daily AQI observations in each city.
# METHOD: Calculate Pearson and Spearman correlations between
#         previous-day and current-day AQI.
# OUTPUT: City-wise temporal autocorrelation report.
# ============================================================

temporal_city_results = []

for city, group in consecutive_aqi.groupby("City"):

    group = group.dropna(
        subset=["Previous_AQI", "Current_AQI"]
    )

    pair_count = len(group)

    if (
        pair_count >= 30
        and group["Previous_AQI"].nunique() > 1
        and group["Current_AQI"].nunique() > 1
    ):

        pearson_corr, pearson_p = pearsonr(
            group["Previous_AQI"],
            group["Current_AQI"]
        )

        spearman_corr, spearman_p = spearmanr(
            group["Previous_AQI"],
            group["Current_AQI"]
        )

    else:
        pearson_corr = np.nan
        pearson_p = np.nan
        spearman_corr = np.nan
        spearman_p = np.nan

    temporal_city_results.append({
        "City": city,
        "Consecutive_Day_Pairs": pair_count,
        "Lag1_Pearson": pearson_corr,
        "Lag1_Pearson_P_Value": pearson_p,
        "Lag1_Spearman": spearman_corr,
        "Lag1_Spearman_P_Value": spearman_p
    })

city_temporal_dependence = pd.DataFrame(
    temporal_city_results
)

city_temporal_dependence = (
    city_temporal_dependence.sort_values(
        "Lag1_Spearman",
        ascending=False
    )
)

display(city_temporal_dependence.round(4))

city_temporal_dependence.to_csv(
    os.path.join(
        report_dir,
        "city_wise_aqi_temporal_dependence.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: TEMPORAL AUTOCORRELATION SIGNIFICANCE VALIDATION
# PURPOSE: Adjust city-wise temporal correlation tests for
#          multiple comparisons.
# METHOD: Apply Benjamini-Hochberg FDR correction.
# OUTPUT: Adjusted temporal significance report.
# ============================================================

temporal_significance = city_temporal_dependence.dropna(
    subset=["Lag1_Spearman_P_Value"]
).copy()

if len(temporal_significance) > 0:

    reject, adjusted_p, _, _ = multipletests(
        temporal_significance["Lag1_Spearman_P_Value"],
        method="fdr_bh"
    )

    temporal_significance["FDR_Adjusted_P_Value"] = adjusted_p

    temporal_significance["FDR_Significant"] = reject

else:
    temporal_significance["FDR_Adjusted_P_Value"] = np.nan
    temporal_significance["FDR_Significant"] = False

print("CITY-WISE TEMPORAL SIGNIFICANCE RESULTS")

display(
    temporal_significance[
        [
            "City",
            "Consecutive_Day_Pairs",
            "Lag1_Spearman",
            "Lag1_Spearman_P_Value",
            "FDR_Adjusted_P_Value",
            "FDR_Significant"
        ]
    ].round(5)
)

temporal_significance.to_csv(
    os.path.join(
        report_dir,
        "city_wise_temporal_significance_fdr.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: CITY-WISE TEMPORAL DEPENDENCE CLASSIFICATION
# PURPOSE: Classify the magnitude of consecutive-day AQI
#          relationships across cities.
# METHOD: Apply descriptive absolute-correlation intervals.
# OUTPUT: Temporal dependence classification report.
# ============================================================

def classify_temporal_dependence(value):

    if pd.isna(value):
        return "Insufficient Data"

    magnitude = abs(value)

    if magnitude >= 0.70:
        return "Very Strong"

    elif magnitude >= 0.50:
        return "Strong"

    elif magnitude >= 0.30:
        return "Moderate"

    elif magnitude >= 0.10:
        return "Weak"

    return "Very Weak"


city_temporal_dependence["Dependence_Strength"] = (
    city_temporal_dependence["Lag1_Spearman"]
    .apply(classify_temporal_dependence)
)

city_temporal_dependence["Dependence_Direction"] = (
    city_temporal_dependence["Lag1_Spearman"]
    .apply(
        lambda x:
        "Positive" if x > 0
        else "Negative" if x < 0
        else "Unavailable"
        if pd.isna(x)
        else "Near Zero"
    )
)

display(
    city_temporal_dependence[
        [
            "City",
            "Consecutive_Day_Pairs",
            "Lag1_Spearman",
            "Dependence_Strength",
            "Dependence_Direction"
        ]
    ].round(4)
)

city_temporal_dependence.to_csv(
    os.path.join(
        report_dir,
        "city_temporal_dependence_classification.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: CITY-WISE TEMPORAL AUTOCORRELATION VISUALIZATION
# PURPOSE: Compare consecutive-day AQI dependence across cities.
# METHOD: Plot city-level lag-1 Spearman correlations.
# OUTPUT: City-wise temporal dependence chart.
# ============================================================

temporal_plot_data = (
    city_temporal_dependence.dropna(
        subset=["Lag1_Spearman"]
    ).sort_values("Lag1_Spearman")
)

plt.figure(figsize=(12, 9))

sns.barplot(
    data=temporal_plot_data,
    x="Lag1_Spearman",
    y="City"
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "City-Wise Consecutive-Day AQI Dependence",
    fontsize=15
)

plt.xlabel("Lag-1 Spearman Correlation")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_wise_aqi_temporal_dependence.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: YEAR-WISE CITY TEMPORAL DEPENDENCE
# PURPOSE: Examine whether AQI persistence changes over time.
# METHOD: Calculate lag-1 Spearman correlation separately
#         for each city and year.
# OUTPUT: Annual city-level temporal dependence report.
# ============================================================

consecutive_aqi["Year"] = (
    consecutive_aqi["Date"].dt.year
)

annual_temporal_results = []

for (city, year), group in consecutive_aqi.groupby(
    ["City", "Year"]
):

    group = group.dropna(
        subset=["Previous_AQI", "Current_AQI"]
    )

    if (
        len(group) >= 30
        and group["Previous_AQI"].nunique() > 1
        and group["Current_AQI"].nunique() > 1
    ):

        correlation, p_value = spearmanr(
            group["Previous_AQI"],
            group["Current_AQI"]
        )

    else:
        correlation = np.nan
        p_value = np.nan

    annual_temporal_results.append({
        "City": city,
        "Year": year,
        "Consecutive_Day_Pairs": len(group),
        "Lag1_Spearman": correlation,
        "P_Value": p_value
    })

annual_temporal_dependence = pd.DataFrame(
    annual_temporal_results
)

annual_temporal_summary = (
    annual_temporal_dependence.groupby("Year")
    .agg(
        Cities_Analyzed=("City", "nunique"),
        Average_Lag1_Correlation=("Lag1_Spearman", "mean"),
        Median_Lag1_Correlation=("Lag1_Spearman", "median"),
        Total_Consecutive_Pairs=("Consecutive_Day_Pairs", "sum")
    )
    .reset_index()
)

print("YEAR-WISE TEMPORAL DEPENDENCE")

display(annual_temporal_summary.round(4))

annual_temporal_dependence.to_csv(
    os.path.join(
        report_dir,
        "annual_city_temporal_dependence.csv"
    ),
    index=False
)

annual_temporal_summary.to_csv(
    os.path.join(
        report_dir,
        "annual_temporal_dependence_summary.csv"
    ),
    index=False
)
```

```python

# ============================================================
# TOPIC: ANNUAL AQI TEMPORAL DEPENDENCE TREND
# PURPOSE: Visualize how average city-level lag-1 AQI
#          correlation varies by year.
# METHOD: Plot annual mean and median city correlations.
# OUTPUT: Annual temporal dependence visualization.
# ============================================================

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=annual_temporal_summary,
    x="Year",
    y="Average_Lag1_Correlation",
    marker="o",
    label="Average"
)

sns.lineplot(
    data=annual_temporal_summary,
    x="Year",
    y="Median_Lag1_Correlation",
    marker="s",
    label="Median"
)

plt.title(
    "Annual Variation in City-Level AQI Temporal Dependence",
    fontsize=15
)

plt.xlabel("Year")
plt.ylabel("Lag-1 Spearman Correlation")

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "annual_aqi_temporal_dependence.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ============================================================
# TOPIC: TEMPORAL DEPENDENCE INTELLIGENCE SUMMARY
# PURPOSE: Consolidate city-level and annual temporal
#          dependence indicators.
# METHOD: Summarize correlation distributions, coverage
#         and significance results.
# OUTPUT: Final temporal dependence report.
# ============================================================

temporal_overall_summary = pd.DataFrame({
    "Metric": [
        "Cities With Calculated Lag-1 Correlation",
        "Cities With Positive Lag-1 Correlation",
        "Cities With Negative Lag-1 Correlation",
        "Average City Lag-1 Spearman Correlation",
        "Median City Lag-1 Spearman Correlation",
        "Total Consecutive-Day Pairs",
        "Cities Significant After FDR"
    ],
    "Value": [
        city_temporal_dependence["Lag1_Spearman"].notna().sum(),
        (
            city_temporal_dependence["Lag1_Spearman"] > 0
        ).sum(),
        (
            city_temporal_dependence["Lag1_Spearman"] < 0
        ).sum(),
        city_temporal_dependence["Lag1_Spearman"].mean(),
        city_temporal_dependence["Lag1_Spearman"].median(),
        len(consecutive_aqi),
        (
            temporal_significance["FDR_Significant"].sum()
            if len(temporal_significance) else 0
        )
    ]
})

display(temporal_overall_summary)

temporal_overall_summary.to_csv(
    os.path.join(
        report_dir,
        "temporal_dependence_intelligence_summary.csv"
    ),
    index=False
)
```

```python
# ============================================================
# MODULE 7.3 — ROBUST POLLUTANT–AQI ASSOCIATION ANALYSIS
# STEP 1: PREPARE CITY-WISE POLLUTANT DATA
# Purpose:
#   Prepare AQI and pollutant observations for city-level
#   association analysis without imputing missing values.
# Method:
#   Select analytical columns and standardize the date-independent
#   city/pollutant structure.
# Output:
#   association_data
# ============================================================

import pandas as pd
import numpy as np
from scipy.stats import spearmanr
from statsmodels.stats.multitest import multipletests

association_pollutants = [
    "PM2.5", "PM10", "NO", "NO2", "NOx", "NH3",
    "CO", "SO2", "O3", "Benzene", "Toluene", "Xylene"
]

association_data = air_final[
    ["City", "AQI"] + association_pollutants
].copy()

association_data["City"] = association_data["City"].astype(str)

print("Association dataset shape:", association_data.shape)
print("Number of cities:", association_data["City"].nunique())
print("Number of pollutants:", len(association_pollutants))
```

```python
# ============================================================
# STEP 2: CITY-WISE SPEARMAN CORRELATION
# Purpose:
#   Measure the association between each pollutant and AQI
#   separately for every city.
# Method:
#   Spearman rank correlation using at least 30 valid pairs.
# Output:
#   city_pollutant_correlations
# ============================================================

city_pollutant_results = []

minimum_pairs = 30

for city, city_data in association_data.groupby("City"):

    for pollutant in association_pollutants:

        valid_data = city_data[["AQI", pollutant]].dropna()

        pair_count = len(valid_data)

        if (
            pair_count >= minimum_pairs
            and valid_data["AQI"].nunique() > 1
            and valid_data[pollutant].nunique() > 1
        ):

            correlation, p_value = spearmanr(
                valid_data["AQI"],
                valid_data[pollutant]
            )

            city_pollutant_results.append({
                "City": city,
                "Pollutant": pollutant,
                "Valid_Pairs": pair_count,
                "Spearman_Correlation": correlation,
                "P_Value": p_value
            })

        else:

            city_pollutant_results.append({
                "City": city,
                "Pollutant": pollutant,
                "Valid_Pairs": pair_count,
                "Spearman_Correlation": np.nan,
                "P_Value": np.nan
            })

city_pollutant_correlations = pd.DataFrame(
    city_pollutant_results
)

print("City-pollutant analysis shape:",
      city_pollutant_correlations.shape)

display(city_pollutant_correlations.head(15))
```

```python
# ============================================================
# STEP 3: FALSE DISCOVERY RATE CORRECTION
# Purpose:
#   Adjust p-values for multiple city-pollutant tests.
# Method:
#   Benjamini-Hochberg FDR correction across valid tests.
# Output:
#   city_pollutant_correlations with adjusted p-values
# ============================================================

city_pollutant_correlations["FDR_Adjusted_P_Value"] = np.nan
city_pollutant_correlations["FDR_Significant"] = False

valid_tests = city_pollutant_correlations["P_Value"].notna()

if valid_tests.sum() > 0:

    rejected, adjusted_p_values, _, _ = multipletests(
        city_pollutant_correlations.loc[
            valid_tests, "P_Value"
        ],
        method="fdr_bh",
        alpha=0.05
    )

    city_pollutant_correlations.loc[
        valid_tests, "FDR_Adjusted_P_Value"
    ] = adjusted_p_values

    city_pollutant_correlations.loc[
        valid_tests, "FDR_Significant"
    ] = rejected

print("Valid statistical tests:", valid_tests.sum())
print("FDR-significant tests:",
      city_pollutant_correlations["FDR_Significant"].sum())

display(
    city_pollutant_correlations.sort_values(
        "FDR_Adjusted_P_Value"
    ).head(15)
)
```

```python
# ============================================================
# STEP 4: CROSS-CITY POLLUTANT ASSOCIATION SUMMARY
# Purpose:
#   Summarize city-specific correlations for each pollutant.
# Method:
#   Median, mean, standard deviation, IQR and significance counts.
# Output:
#   pollutant_city_association_summary
# ============================================================

pollutant_city_association_summary = (
    city_pollutant_correlations
    .groupby("Pollutant")
    .agg(
        Eligible_Cities=("Spearman_Correlation", "count"),
        Median_Correlation=("Spearman_Correlation", "median"),
        Mean_Correlation=("Spearman_Correlation", "mean"),
        Correlation_Std=("Spearman_Correlation", "std"),
        Minimum_Correlation=("Spearman_Correlation", "min"),
        Maximum_Correlation=("Spearman_Correlation", "max"),
        Median_Valid_Pairs=("Valid_Pairs", "median"),
        FDR_Significant_Cities=("FDR_Significant", "sum")
    )
    .reset_index()
)

pollutant_city_association_summary["Correlation_IQR"] = (
    city_pollutant_correlations
    .groupby("Pollutant")["Spearman_Correlation"]
    .quantile(0.75).reindex(
        pollutant_city_association_summary["Pollutant"]
    ).to_numpy()
    -
    city_pollutant_correlations
    .groupby("Pollutant")["Spearman_Correlation"]
    .quantile(0.25).reindex(
        pollutant_city_association_summary["Pollutant"]
    ).to_numpy()
)

pollutant_city_association_summary = (
    pollutant_city_association_summary.sort_values(
        "Median_Correlation",
        ascending=False
    )
)

display(pollutant_city_association_summary)

pollutant_city_association_summary.to_csv(
    os.path.join(
        report_dir,
        "pollutant_city_association_summary.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 5: CITY-LEVEL ASSOCIATION VARIABILITY
# Purpose:
#   Identify cities with unusually strong or weak pollutant
#   associations with AQI.
# Method:
#   Rank city-pollutant correlations and retain FDR results.
# Output:
#   strongest_city_pollutant_associations
#   weakest_city_pollutant_associations
# ============================================================

strongest_city_pollutant_associations = (
    city_pollutant_correlations
    .dropna(subset=["Spearman_Correlation"])
    .sort_values(
        "Spearman_Correlation",
        ascending=False
    )
    .head(20)
)

weakest_city_pollutant_associations = (
    city_pollutant_correlations
    .dropna(subset=["Spearman_Correlation"])
    .sort_values(
        "Spearman_Correlation",
        ascending=True
    )
    .head(20)
)

print("Strongest city-pollutant associations")
display(strongest_city_pollutant_associations)

print("Weakest city-pollutant associations")
display(weakest_city_pollutant_associations)

strongest_city_pollutant_associations.to_csv(
    os.path.join(
        report_dir,
        "strongest_city_pollutant_associations.csv"
    ),
    index=False
)

weakest_city_pollutant_associations.to_csv(
    os.path.join(
        report_dir,
        "weakest_city_pollutant_associations.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 6: CITY-WISE CORRELATION HEATMAP
# Purpose:
#   Visually compare pollutant-AQI relationships across cities.
# Method:
#   Pivot city-specific Spearman correlations into a heatmap.
# Output:
#   city_pollutant_correlation_heatmap.png
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

city_pollutant_matrix = (
    city_pollutant_correlations
    .pivot(
        index="City",
        columns="Pollutant",
        values="Spearman_Correlation"
    )
)

plt.figure(figsize=(16, 11))

sns.heatmap(
    city_pollutant_matrix,
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.3,
    linecolor="white"
)

plt.title(
    "City-Wise Pollutant–AQI Spearman Correlations",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("Pollutant")
plt.ylabel("City")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_pollutant_correlation_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 7: CROSS-CITY CORRELATION DISTRIBUTIONS
# Purpose:
#   Compare how pollutant-AQI correlations vary across cities.
# Method:
#   Boxplots of city-specific Spearman correlations.
# Output:
#   pollutant_correlation_distribution.png
# ============================================================

plt.figure(figsize=(16, 8))

sns.boxplot(
    data=city_pollutant_correlations,
    x="Pollutant",
    y="Spearman_Correlation"
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Distribution of City-Specific Pollutant–AQI Correlations",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Pollutant")
plt.ylabel("Spearman Correlation")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "pollutant_correlation_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 8: POOLED VERSUS CITY-LEVEL CORRELATION
# Purpose:
#   Compare overall pooled correlations with median city-level
#   correlations.
# Method:
#   Calculate pooled Spearman correlation for each pollutant
#   and compare it with the median of city-specific correlations.
# Output:
#   pooled_vs_city_association
# ============================================================

pooled_association_results = []

for pollutant in association_pollutants:

    valid_data = association_data[
        ["AQI", pollutant]
    ].dropna()

    if (
        len(valid_data) >= minimum_pairs
        and valid_data["AQI"].nunique() > 1
        and valid_data[pollutant].nunique() > 1
    ):

        pooled_corr, pooled_p = spearmanr(
            valid_data["AQI"],
            valid_data[pollutant]
        )

    else:
        pooled_corr = np.nan
        pooled_p = np.nan

    city_median = (
        city_pollutant_correlations.loc[
            city_pollutant_correlations["Pollutant"] == pollutant,
            "Spearman_Correlation"
        ].median()
    )

    pooled_association_results.append({
        "Pollutant": pollutant,
        "Pooled_Spearman": pooled_corr,
        "Pooled_P_Value": pooled_p,
        "Median_City_Spearman": city_median,
        "Pooled_Minus_City_Median": (
            pooled_corr - city_median
            if pd.notna(pooled_corr) and pd.notna(city_median)
            else np.nan
        )
    })

pooled_vs_city_association = pd.DataFrame(
    pooled_association_results
)

display(
    pooled_vs_city_association.sort_values(
        "Pooled_Spearman",
        ascending=False
    )
)

pooled_vs_city_association.to_csv(
    os.path.join(
        report_dir,
        "pooled_vs_city_pollutant_associations.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 9: CONSOLIDATED ASSOCIATION INTELLIGENCE REPORT
# Purpose:
#   Export the full city-wise association analysis and its
#   pollutant-level summary.
# Method:
#   Save detailed tests, summary statistics and pooled comparison.
# Output:
#   Three CSV reports for project documentation.
# ============================================================

city_pollutant_correlations.to_csv(
    os.path.join(
        report_dir,
        "city_wise_pollutant_aqi_correlations.csv"
    ),
    index=False
)

pollutant_city_association_summary.to_csv(
    os.path.join(
        report_dir,
        "pollutant_city_association_summary.csv"
    ),
    index=False
)

pooled_vs_city_association.to_csv(
    os.path.join(
        report_dir,
        "pooled_vs_city_pollutant_associations.csv"
    ),
    index=False
)

print("Submodule 7.3 report exports completed.")
print("Detailed city-pollutant tests:",
      len(city_pollutant_correlations))
print("Eligible correlation tests:",
      city_pollutant_correlations[
          "Spearman_Correlation"
      ].notna().sum())
print("FDR-significant tests:",
      city_pollutant_correlations[
          "FDR_Significant"
      ].sum())
```

```python
# ============================================================
# MODULE 7.4 — STATISTICAL STABILITY AND SENSITIVITY ANALYSIS
# STEP 1: DEFINE SENSITIVITY SETTINGS
# Purpose:
#   Establish alternative sample-size and correlation settings.
# Method:
#   Compare multiple minimum-pair thresholds and two
#   non-parametric correlation methods.
# Output:
#   sensitivity_minimum_pairs
#   sensitivity_methods
# ============================================================

sensitivity_minimum_pairs = [30, 50, 100]

sensitivity_methods = [
    "spearman",
    "kendall"
]

print("Minimum pair thresholds:", sensitivity_minimum_pairs)
print("Correlation methods:", sensitivity_methods)

```

```python
# ============================================================
# STEP 2: MULTI-THRESHOLD CORRELATION ANALYSIS
# Purpose:
#   Determine how pollutant-AQI associations change when
#   minimum sample-size requirements become stricter.
# Method:
#   Calculate city-specific Spearman correlations at
#   30, 50 and 100 valid pairs.
# Output:
#   threshold_sensitivity_results
# ============================================================

threshold_results = []

for city, city_data in association_data.groupby("City"):

    for pollutant in association_pollutants:

        valid_data = city_data[["AQI", pollutant]].dropna()

        pair_count = len(valid_data)

        for threshold in sensitivity_minimum_pairs:

            if (
                pair_count >= threshold
                and valid_data["AQI"].nunique() > 1
                and valid_data[pollutant].nunique() > 1
            ):

                corr, p_value = spearmanr(
                    valid_data["AQI"],
                    valid_data[pollutant]
                )

            else:
                corr = np.nan
                p_value = np.nan

            threshold_results.append({
                "City": city,
                "Pollutant": pollutant,
                "Minimum_Pairs": threshold,
                "Valid_Pairs": pair_count,
                "Spearman_Correlation": corr,
                "P_Value": p_value
            })

threshold_sensitivity_results = pd.DataFrame(
    threshold_results
)

print("Sensitivity analysis records:",
      len(threshold_sensitivity_results))

display(threshold_sensitivity_results.head(15))
```

```python
# ============================================================
# STEP 3: SAMPLE-SIZE SENSITIVITY SUMMARY
# Purpose:
#   Compare the number and strength of eligible city-level
#   associations at different minimum sample sizes.
# Method:
#   Summarize median correlations, eligible cities and
#   positive/negative association counts.
# Output:
#   threshold_sensitivity_summary
# ============================================================

threshold_sensitivity_summary = (
    threshold_sensitivity_results
    .groupby(["Pollutant", "Minimum_Pairs"])
    .agg(
        Eligible_Cities=("Spearman_Correlation", "count"),
        Median_Correlation=("Spearman_Correlation", "median"),
        Mean_Correlation=("Spearman_Correlation", "mean"),
        Correlation_Std=("Spearman_Correlation", "std"),
        Positive_Associations=(
            "Spearman_Correlation",
            lambda x: (x > 0).sum()
        ),
        Negative_Associations=(
            "Spearman_Correlation",
            lambda x: (x < 0).sum()
        )
    )
    .reset_index()
)

display(
    threshold_sensitivity_summary.sort_values(
        ["Pollutant", "Minimum_Pairs"]
    )
)

threshold_sensitivity_summary.to_csv(
    os.path.join(
        report_dir,
        "threshold_sensitivity_summary.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 4: ALTERNATIVE CORRELATION METHOD COMPARISON
# Purpose:
#   Test whether city-specific associations are similar
#   under Spearman and Kendall correlation methods.
# Method:
#   Calculate both methods using at least 30 valid pairs.
# Output:
#   correlation_method_sensitivity
# ============================================================

method_results = []

for city, city_data in association_data.groupby("City"):

    for pollutant in association_pollutants:

        valid_data = city_data[["AQI", pollutant]].dropna()

        pair_count = len(valid_data)

        if (
            pair_count >= 30
            and valid_data["AQI"].nunique() > 1
            and valid_data[pollutant].nunique() > 1
        ):

            spearman_corr, spearman_p = spearmanr(
                valid_data["AQI"],
                valid_data[pollutant]
            )

            kendall_corr, kendall_p = (
                valid_data["AQI"].corr(
                    valid_data[pollutant],
                    method="kendall"
                ),
                np.nan
            )

        else:
            spearman_corr = np.nan
            spearman_p = np.nan
            kendall_corr = np.nan
            kendall_p = np.nan

        method_results.append({
            "City": city,
            "Pollutant": pollutant,
            "Valid_Pairs": pair_count,
            "Spearman_Correlation": spearman_corr,
            "Spearman_P_Value": spearman_p,
            "Kendall_Correlation": kendall_corr,
            "Kendall_P_Value": kendall_p
        })

correlation_method_sensitivity = pd.DataFrame(
    method_results
)

correlation_method_sensitivity[
    "Absolute_Method_Difference"
] = (
    correlation_method_sensitivity["Spearman_Correlation"]
    - correlation_method_sensitivity["Kendall_Correlation"]
).abs()

display(correlation_method_sensitivity.head(15))
```

```python
# ============================================================
# STEP 5: CORRELATION METHOD AGREEMENT
# Purpose:
#   Measure agreement between Spearman and Kendall results.
# Method:
#   Compare direction agreement and average absolute
#   difference for each pollutant.
# Output:
#   correlation_method_agreement
# ============================================================

method_comparison = correlation_method_sensitivity.dropna(
    subset=[
        "Spearman_Correlation",
        "Kendall_Correlation"
    ]
).copy()

method_comparison["Direction_Agreement"] = (
    np.sign(method_comparison["Spearman_Correlation"])
    ==
    np.sign(method_comparison["Kendall_Correlation"])
)

correlation_method_agreement = (
    method_comparison
    .groupby("Pollutant")
    .agg(
        Eligible_Cities=("City", "count"),
        Direction_Agreement_Rate=(
            "Direction_Agreement", "mean"
        ),
        Mean_Absolute_Method_Difference=(
            "Absolute_Method_Difference", "mean"
        ),
        Median_Absolute_Method_Difference=(
            "Absolute_Method_Difference", "median"
        )
    )
    .reset_index()
)

correlation_method_agreement[
    "Direction_Agreement_Percent"
] = (
    correlation_method_agreement["Direction_Agreement_Rate"] * 100
)

display(correlation_method_agreement)

correlation_method_agreement.to_csv(
    os.path.join(
        report_dir,
        "correlation_method_agreement.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 6: CITY-WISE MONITORING COVERAGE
# Purpose:
#   Measure available paired AQI-pollutant observations
#   for each city.
# Method:
#   Calculate valid pairs and coverage percentages.
# Output:
#   city_pollutant_coverage
# ============================================================

coverage_results = []

for city, city_data in association_data.groupby("City"):

    total_city_rows = len(city_data)

    for pollutant in association_pollutants:

        valid_pairs = city_data[
            ["AQI", pollutant]
        ].dropna().shape[0]

        coverage_percent = (
            valid_pairs / total_city_rows * 100
            if total_city_rows > 0
            else np.nan
        )

        coverage_results.append({
            "City": city,
            "Pollutant": pollutant,
            "Total_City_Records": total_city_rows,
            "Valid_Pairs": valid_pairs,
            "Coverage_Percent": coverage_percent
        })

city_pollutant_coverage = pd.DataFrame(
    coverage_results
)

display(city_pollutant_coverage.head(15))

city_pollutant_coverage.to_csv(
    os.path.join(
        report_dir,
        "city_pollutant_monitoring_coverage.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 6: CITY-WISE MONITORING COVERAGE
# Purpose:
#   Measure available paired AQI-pollutant observations
#   for each city.
# Method:
#   Calculate valid pairs and coverage percentages.
# Output:
#   city_pollutant_coverage
# ============================================================

coverage_results = []

for city, city_data in association_data.groupby("City"):

    total_city_rows = len(city_data)

    for pollutant in association_pollutants:

        valid_pairs = city_data[
            ["AQI", pollutant]
        ].dropna().shape[0]

        coverage_percent = (
            valid_pairs / total_city_rows * 100
            if total_city_rows > 0
            else np.nan
        )

        coverage_results.append({
            "City": city,
            "Pollutant": pollutant,
            "Total_City_Records": total_city_rows,
            "Valid_Pairs": valid_pairs,
            "Coverage_Percent": coverage_percent
        })

city_pollutant_coverage = pd.DataFrame(
    coverage_results
)

display(city_pollutant_coverage.head(15))

city_pollutant_coverage.to_csv(
    os.path.join(
        report_dir,
        "city_pollutant_monitoring_coverage.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 7: BETTER-COVERED CITY SENSITIVITY ANALYSIS
# Purpose:
#   Check whether pollutant-AQI associations persist among
#   cities with stronger paired observation coverage.
# Method:
#   Require at least 50% coverage and 30 valid pairs.
# Output:
#   high_coverage_association_results
# ============================================================

high_coverage_results = []

for city, city_data in association_data.groupby("City"):

    total_city_rows = len(city_data)

    for pollutant in association_pollutants:

        valid_data = city_data[
            ["AQI", pollutant]
        ].dropna()

        pair_count = len(valid_data)

        coverage_percent = (
            pair_count / total_city_rows * 100
            if total_city_rows > 0
            else np.nan
        )

        eligible = (
            pair_count >= 30
            and coverage_percent >= 50
            and valid_data["AQI"].nunique() > 1
            and valid_data[pollutant].nunique() > 1
        )

        if eligible:

            corr, p_value = spearmanr(
                valid_data["AQI"],
                valid_data[pollutant]
            )

        else:
            corr = np.nan
            p_value = np.nan

        high_coverage_results.append({
            "City": city,
            "Pollutant": pollutant,
            "Valid_Pairs": pair_count,
            "Coverage_Percent": coverage_percent,
            "High_Coverage_Eligible": eligible,
            "Spearman_Correlation": corr,
            "P_Value": p_value
        })

high_coverage_association_results = pd.DataFrame(
    high_coverage_results
)

display(high_coverage_association_results.head(15))
```

```python
# ============================================================
# STEP 8: ORIGINAL VERSUS HIGH-COVERAGE COMPARISON
# Purpose:
#   Determine whether the median city-level correlation
#   changes when lower-coverage city-pollutant combinations
#   are excluded.
# Method:
#   Compare median Spearman correlations and eligible counts.
# Output:
#   coverage_sensitivity_comparison
# ============================================================

original_summary = (
    city_pollutant_correlations
    .groupby("Pollutant")
    .agg(
        Original_Eligible_Cities=(
            "Spearman_Correlation", "count"
        ),
        Original_Median_Correlation=(
            "Spearman_Correlation", "median"
        )
    )
    .reset_index()
)

high_coverage_summary = (
    high_coverage_association_results
    .groupby("Pollutant")
    .agg(
        High_Coverage_Cities=(
            "Spearman_Correlation", "count"
        ),
        High_Coverage_Median_Correlation=(
            "Spearman_Correlation", "median"
        )
    )
    .reset_index()
)

coverage_sensitivity_comparison = original_summary.merge(
    high_coverage_summary,
    on="Pollutant",
    how="outer"
)

coverage_sensitivity_comparison[
    "Median_Correlation_Difference"
] = (
    coverage_sensitivity_comparison[
        "High_Coverage_Median_Correlation"
    ]
    -
    coverage_sensitivity_comparison[
        "Original_Median_Correlation"
    ]
)

display(coverage_sensitivity_comparison)

coverage_sensitivity_comparison.to_csv(
    os.path.join(
        report_dir,
        "coverage_sensitivity_comparison.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 9: SAMPLE-SIZE SENSITIVITY VISUALIZATION
# Purpose:
#   Compare pollutant-level median correlations across
#   minimum sample-size thresholds.
# Method:
#   Line plot with one line per pollutant.
# Output:
#   pollutant_threshold_sensitivity.png
# ============================================================

plt.figure(figsize=(15, 8))

sns.lineplot(
    data=threshold_sensitivity_summary,
    x="Minimum_Pairs",
    y="Median_Correlation",
    hue="Pollutant",
    marker="o"
)

plt.axhline(0, linestyle="--", linewidth=1)

plt.title(
    "Pollutant–AQI Correlation Stability Across Sample Sizes",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Minimum Valid Pairs")
plt.ylabel("Median City-Level Spearman Correlation")
plt.xticks(sensitivity_minimum_pairs)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "pollutant_threshold_sensitivity.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 10: CONSOLIDATED SENSITIVITY REPORT
# Purpose:
#   Export all major sensitivity-analysis results.
# Method:
#   Save threshold, method and coverage comparisons.
# Output:
#   Consolidated Module 7.4 CSV reports.
# ============================================================

threshold_sensitivity_results.to_csv(
    os.path.join(
        report_dir,
        "threshold_sensitivity_detailed.csv"
    ),
    index=False
)

threshold_sensitivity_summary.to_csv(
    os.path.join(
        report_dir,
        "threshold_sensitivity_summary.csv"
    ),
    index=False
)

correlation_method_sensitivity.to_csv(
    os.path.join(
        report_dir,
        "correlation_method_sensitivity_detailed.csv"
    ),
    index=False
)

correlation_method_agreement.to_csv(
    os.path.join(
        report_dir,
        "correlation_method_agreement.csv"
    ),
    index=False
)

high_coverage_association_results.to_csv(
    os.path.join(
        report_dir,
        "high_coverage_association_results.csv"
    ),
    index=False
)

coverage_sensitivity_comparison.to_csv(
    os.path.join(
        report_dir,
        "coverage_sensitivity_comparison.csv"
    ),
    index=False
)

print("Module 7.4 sensitivity reports exported.")
print("Threshold analysis records:",
      len(threshold_sensitivity_results))
print("Method comparison records:",
      len(correlation_method_sensitivity))
print("High-coverage eligible combinations:",
      high_coverage_association_results[
          "High_Coverage_Eligible"
      ].sum())
```

```python
# ============================================================
# MODULE 7.5 — STATISTICAL VALIDATION OF ENVIRONMENTAL RISK RANKINGS
# STEP 1: INSPECT MODULE 6 RISK INDEX
# Purpose:
#   Verify the existing city-level indicators and baseline weights.
# Method:
#   Inspect city_index and risk_weights.
# Output:
#   Confirmed indicator names and available columns.
# ============================================================

print("City index columns:")
print(city_index.columns.tolist())

print("\nOriginal Module 6 risk weights:")
print(risk_weights)

print("\nCity index shape:", city_index.shape)

display(city_index.head())
```

```python
# ============================================================
# STEP 2: PREPARE BASELINE INDICATOR MATRIX
# Purpose:
#   Extract the existing normalized indicators used in Module 6.
# Method:
#   Match risk_weights keys against city_index columns.
# Output:
#   risk_indicator_matrix and baseline_weights
# ============================================================

risk_indicator_columns = list(risk_weights.keys())

missing_indicators = [
    col for col in risk_indicator_columns
    if col not in city_index.columns
]

if missing_indicators:
    raise KeyError(
        "These weighted indicators are missing from city_index: "
        f"{missing_indicators}. Check the Step 1 column output."
    )

risk_indicator_matrix = (
    city_index[["City"] + risk_indicator_columns]
    .copy()
)

baseline_weights = pd.Series(
    risk_weights,
    dtype=float
).reindex(risk_indicator_columns)

if baseline_weights.isna().any() or (baseline_weights < 0).any():
    raise ValueError("Baseline weights must be present and non-negative.")

baseline_weights = baseline_weights / baseline_weights.sum()

print("Indicators used:", risk_indicator_columns)
print("\nNormalized baseline weights:")
display(baseline_weights.rename("Baseline_Weight").to_frame())

display(risk_indicator_matrix.head())
```

```python
# ============================================================
# STEP 3: RECONSTRUCT BASELINE ENVIRONMENTAL RISK INDEX
# Purpose:
#   Recalculate the original weighted risk score.
# Method:
#   Weighted sum of the normalized Module 6 indicators.
# Output:
#   baseline_ranking
# ============================================================

baseline_ranking = risk_indicator_matrix[["City"]].copy()

baseline_ranking["Baseline_Recalculated_Score"] = (
    risk_indicator_matrix[risk_indicator_columns]
    .mul(baseline_weights, axis=1)
    .sum(axis=1)
)

baseline_ranking["Baseline_Rank"] = (
    baseline_ranking["Baseline_Recalculated_Score"]
    .rank(method="min", ascending=False)
    .astype(int)
)

baseline_ranking = baseline_ranking.sort_values(
    "Baseline_Rank"
).reset_index(drop=True)

display(baseline_ranking)

baseline_ranking.to_csv(
    os.path.join(
        report_dir,
        "risk_ranking_baseline_recalculated.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 4: DEFINE ALTERNATIVE RISK-WEIGHT SCENARIOS
# Purpose:
#   Test how alternative indicator priorities affect city rankings.
# Method:
#   Construct normalized alternative weights.
# Output:
#   scenario_weights
# ============================================================

if len(risk_indicator_columns) != 5:
    raise ValueError(
        "Expected the five Module 6 risk indicators. "
        "Check risk_indicator_columns before defining scenarios."
    )

scenario_weights = {
    "Baseline": baseline_weights.to_numpy(),

    "Equal_Weight": np.array([
        0.20, 0.20, 0.20, 0.20, 0.20
    ]),

    "AQI_Priority": np.array([
        0.40, 0.20, 0.10, 0.20, 0.10
    ]),

    "Pollution_Exposure_Priority": np.array([
        0.15, 0.35, 0.25, 0.20, 0.05
    ]),

    "Persistence_Priority": np.array([
        0.15, 0.20, 0.10, 0.15, 0.40
    ])
}

scenario_weights = {
    name: weights / weights.sum()
    for name, weights in scenario_weights.items()
}

scenario_weight_table = pd.DataFrame(
    scenario_weights,
    index=risk_indicator_columns
)

print("Alternative weighting scenarios:")
display(scenario_weight_table)
```

```python
# ============================================================
# STEP 5: CALCULATE SCENARIO-SPECIFIC CITY RANKINGS
# Purpose:
#   Recalculate environmental risk scores under each weighting
#   scenario.
# Method:
#   Apply each weight vector to the same normalized indicators.
# Output:
#   scenario_ranking_results
# ============================================================

scenario_ranking_results = risk_indicator_matrix[["City"]].copy()

for scenario_name, weights in scenario_weights.items():

    scenario_ranking_results[
        f"{scenario_name}_Score"
    ] = (
        risk_indicator_matrix[risk_indicator_columns]
        .mul(weights, axis=1)
        .sum(axis=1)
    )

    scenario_ranking_results[
        f"{scenario_name}_Rank"
    ] = (
        scenario_ranking_results[
            f"{scenario_name}_Score"
        ]
        .rank(method="min", ascending=False)
        .astype(int)
    )

display(
    scenario_ranking_results.sort_values(
        "Baseline_Rank"
    )
)

scenario_ranking_results.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_scenario_rankings.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 6: RANK AGREEMENT ANALYSIS
# Purpose:
#   Measure agreement between baseline and alternative rankings.
# Method:
#   Spearman and Kendall rank correlations.
# Output:
#   ranking_agreement_results
# ============================================================

from scipy.stats import kendalltau

ranking_agreement_records = []

baseline_rank_values = scenario_ranking_results[
    "Baseline_Rank"
]

for scenario_name in scenario_weights:

    scenario_rank_values = scenario_ranking_results[
        f"{scenario_name}_Rank"
    ]

    spearman_corr, spearman_p = spearmanr(
        baseline_rank_values,
        scenario_rank_values
    )

    kendall_corr, kendall_p = kendalltau(
        baseline_rank_values,
        scenario_rank_values
    )

    ranking_agreement_records.append({
        "Scenario": scenario_name,
        "Spearman_Rank_Correlation": spearman_corr,
        "Spearman_P_Value": spearman_p,
        "Kendall_Rank_Correlation": kendall_corr,
        "Kendall_P_Value": kendall_p
    })

ranking_agreement_results = pd.DataFrame(
    ranking_agreement_records
)

display(ranking_agreement_results)

ranking_agreement_results.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_ranking_agreement.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 7: CITY-LEVEL RANK MOVEMENT
# Purpose:
#   Identify cities whose rankings change most under
#   alternative weighting scenarios.
# Method:
#   Calculate rank differences from the baseline.
# Output:
#   city_ranking_stability
# ============================================================

city_ranking_stability = scenario_ranking_results[
    ["City", "Baseline_Rank"]
].copy()

alternative_scenarios = [
    name for name in scenario_weights
    if name != "Baseline"
]

rank_difference_columns = []

for scenario_name in alternative_scenarios:

    difference_column = f"{scenario_name}_Rank_Change"

    city_ranking_stability[difference_column] = (
        scenario_ranking_results[
            f"{scenario_name}_Rank"
        ].to_numpy()
        -
        scenario_ranking_results[
            "Baseline_Rank"
        ].to_numpy()
    )

    rank_difference_columns.append(difference_column)

city_ranking_stability["Mean_Absolute_Rank_Change"] = (
    city_ranking_stability[rank_difference_columns]
    .abs()
    .mean(axis=1)
)

city_ranking_stability["Maximum_Absolute_Rank_Change"] = (
    city_ranking_stability[rank_difference_columns]
    .abs()
    .max(axis=1)
)

city_ranking_stability = city_ranking_stability.sort_values(
    "Mean_Absolute_Rank_Change",
    ascending=False
).reset_index(drop=True)

display(city_ranking_stability)

city_ranking_stability.to_csv(
    os.path.join(
        report_dir,
        "city_environmental_ranking_stability.csv"
    ),
    index=False
)
```

```python



# ============================================================
# STEP 8: RANKING STABILITY CLASSIFICATION
# Purpose:
#   Group cities according to their average rank movement.
# Method:
#   Apply project-defined thresholds to mean absolute
#   rank changes.
# Output:
#   city_ranking_stability
# ============================================================

def classify_rank_stability(rank_change):

    if rank_change <= 2:
        return "Highly Stable"

    elif rank_change <= 5:
        return "Moderately Stable"

    else:
        return "Sensitive to Weights"


city_ranking_stability["Ranking_Stability_Class"] = (
    city_ranking_stability[
        "Mean_Absolute_Rank_Change"
    ].apply(classify_rank_stability)
)

ranking_stability_summary = (
    city_ranking_stability
    ["Ranking_Stability_Class"]
    .value_counts()
    .rename_axis("Stability_Class")
    .reset_index(name="City_Count")
)

display(ranking_stability_summary)

ranking_stability_summary.to_csv(
    os.path.join(
        report_dir,
        "environmental_ranking_stability_class_summary.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 9: RANKING STABILITY VISUALIZATION
# Purpose:
#   Visualize how much city rankings move under alternative
#   weighting scenarios.
# Method:
#   Horizontal bar chart of mean absolute rank changes.
# Output:
#   environmental_ranking_stability.png
# ============================================================

plt.figure(figsize=(12, 9))

plot_data = city_ranking_stability.sort_values(
    "Mean_Absolute_Rank_Change",
    ascending=True
)

sns.barplot(
    data=plot_data,
    x="Mean_Absolute_Rank_Change",
    y="City"
)

plt.title(
    "Environmental Risk Ranking Stability Across Cities",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Mean Absolute Rank Change")
plt.ylabel("City")
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "environmental_ranking_stability.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 10: CONSOLIDATED RANKING VALIDATION EXPORT
# Purpose:
#   Save the complete Module 7.5 ranking sensitivity results.
# Method:
#   Export scenario scores, rank agreement and city stability.
# Output:
#   Module 7.5 CSV reports.
# ============================================================

scenario_weight_table.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_scenario_weights.csv"
    )
)

scenario_ranking_results.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_scenario_rankings.csv"
    ),
    index=False
)

ranking_agreement_results.to_csv(
    os.path.join(
        report_dir,
        "environmental_risk_ranking_agreement.csv"
    ),
    index=False
)

city_ranking_stability.to_csv(
    os.path.join(
        report_dir,
        "city_environmental_ranking_stability.csv"
    ),
    index=False
)

print("Module 7.5 ranking validation exports completed.")
print("Cities evaluated:", len(scenario_ranking_results))
print("Scenarios evaluated:", len(scenario_weights))
print("\nRanking agreement:")
display(ranking_agreement_results)

print("\nRanking stability classes:")
display(ranking_stability_summary)
```

```python
# ============================================================
# STEP 3: TEMPORAL DEPENDENCE AUDIT
# Purpose:
#   Check city-level and annual temporal-dependence outputs.
# Method:
#   Validate required fields, correlation ranges and
#   consecutive-day pair counts.
# Output:
#   temporal_audit_summary
# ============================================================

temporal_audit_records = []

if (
    "city_temporal_dependence" in globals()
    and isinstance(city_temporal_dependence, pd.DataFrame)
):

    temporal_df = city_temporal_dependence.copy()

    temporal_audit_records.append({
        "Analysis": "City-Level Temporal Dependence",
        "Rows": len(temporal_df),
        "Columns": len(temporal_df.columns),
        "Missing_Values": int(temporal_df.isna().sum().sum()),
        "Required_Columns_Available": all(
            col in temporal_df.columns
            for col in ["City"]
        )
    })

    correlation_columns = [
        col for col in temporal_df.columns
        if "Correlation" in col
    ]

    for column in correlation_columns:

        invalid_values = (
            ~temporal_df[column].dropna().between(-1, 1)
        ).sum()

        print(
            column,
            "- values outside [-1, 1]:",
            invalid_values
        )

else:
    temporal_audit_records.append({
        "Analysis": "City-Level Temporal Dependence",
        "Rows": 0,
        "Columns": 0,
        "Missing_Values": None,
        "Required_Columns_Available": False
    })

if (
    "annual_temporal_dependence" in globals()
    and isinstance(annual_temporal_dependence, pd.DataFrame)
):

    annual_df = annual_temporal_dependence.copy()

    temporal_audit_records.append({
        "Analysis": "Annual Temporal Dependence",
        "Rows": len(annual_df),
        "Columns": len(annual_df.columns),
        "Missing_Values": int(annual_df.isna().sum().sum()),
        "Required_Columns_Available": "City" in annual_df.columns
    })

else:
    temporal_audit_records.append({
        "Analysis": "Annual Temporal Dependence",
        "Rows": 0,
        "Columns": 0,
        "Missing_Values": None,
        "Required_Columns_Available": False
    })

temporal_audit_summary = pd.DataFrame(
    temporal_audit_records
)

display(temporal_audit_summary)
```

```python
# ============================================================
# STEP 4: POLLUTANT ASSOCIATION AUDIT
# Purpose:
#   Validate city-specific pollutant-AQI correlation results.
# Method:
#   Check correlation bounds, valid sample counts,
#   FDR-adjusted p-values and duplicate city-pollutant pairs.
# Output:
#   association_audit_summary
# ============================================================

association_audit_records = []

if (
    "city_pollutant_correlations" in globals()
    and isinstance(city_pollutant_correlations, pd.DataFrame)
):

    association_df = city_pollutant_correlations.copy()

    required_association_columns = [
        "City",
        "Pollutant",
        "Valid_Pairs",
        "Spearman_Correlation",
        "P_Value",
        "FDR_Adjusted_P_Value",
        "FDR_Significant"
    ]

    missing_columns = [
        col for col in required_association_columns
        if col not in association_df.columns
    ]

    duplicate_pairs = association_df.duplicated(
        subset=["City", "Pollutant"]
    ).sum()

    invalid_correlations = (
        ~association_df["Spearman_Correlation"]
        .dropna()
        .between(-1, 1)
    ).sum()

    invalid_p_values = (
        ~association_df["FDR_Adjusted_P_Value"]
        .dropna()
        .between(0, 1)
    ).sum()

    association_audit_summary = pd.DataFrame([{
        "Total_Records": len(association_df),
        "Missing_Required_Columns": len(missing_columns),
        "Duplicate_City_Pollutant_Pairs": int(duplicate_pairs),
        "Invalid_Correlations": int(invalid_correlations),
        "Invalid_FDR_P_Values": int(invalid_p_values),
        "Minimum_Valid_Pairs": (
            association_df["Valid_Pairs"].min()
            if len(association_df) else np.nan
        ),
        "Maximum_Valid_Pairs": (
            association_df["Valid_Pairs"].max()
            if len(association_df) else np.nan
        )
    }])

    print("Missing required columns:", missing_columns)

else:

    association_audit_summary = pd.DataFrame([{
        "Total_Records": 0,
        "Missing_Required_Columns": None,
        "Duplicate_City_Pollutant_Pairs": None,
        "Invalid_Correlations": None,
        "Invalid_FDR_P_Values": None,
        "Minimum_Valid_Pairs": None,
        "Maximum_Valid_Pairs": None
    }])

display(association_audit_summary)
```

```python
# ============================================================
# STEP 4: POLLUTANT ASSOCIATION AUDIT
# Purpose:
#   Validate city-specific pollutant-AQI correlation results.
# Method:
#   Check correlation bounds, valid sample counts,
#   FDR-adjusted p-values and duplicate city-pollutant pairs.
# Output:
#   association_audit_summary
# ============================================================

association_audit_records = []

if (
    "city_pollutant_correlations" in globals()
    and isinstance(city_pollutant_correlations, pd.DataFrame)
):

    association_df = city_pollutant_correlations.copy()

    required_association_columns = [
        "City",
        "Pollutant",
        "Valid_Pairs",
        "Spearman_Correlation",
        "P_Value",
        "FDR_Adjusted_P_Value",
        "FDR_Significant"
    ]

    missing_columns = [
        col for col in required_association_columns
        if col not in association_df.columns
    ]

    duplicate_pairs = association_df.duplicated(
        subset=["City", "Pollutant"]
    ).sum()

    invalid_correlations = (
        ~association_df["Spearman_Correlation"]
        .dropna()
        .between(-1, 1)
    ).sum()

    invalid_p_values = (
        ~association_df["FDR_Adjusted_P_Value"]
        .dropna()
        .between(0, 1)
    ).sum()

    association_audit_summary = pd.DataFrame([{
        "Total_Records": len(association_df),
        "Missing_Required_Columns": len(missing_columns),
        "Duplicate_City_Pollutant_Pairs": int(duplicate_pairs),
        "Invalid_Correlations": int(invalid_correlations),
        "Invalid_FDR_P_Values": int(invalid_p_values),
        "Minimum_Valid_Pairs": (
            association_df["Valid_Pairs"].min()
            if len(association_df) else np.nan
        ),
        "Maximum_Valid_Pairs": (
            association_df["Valid_Pairs"].max()
            if len(association_df) else np.nan
        )
    }])

    print("Missing required columns:", missing_columns)

else:

    association_audit_summary = pd.DataFrame([{
        "Total_Records": 0,
        "Missing_Required_Columns": None,
        "Duplicate_City_Pollutant_Pairs": None,
        "Invalid_Correlations": None,
        "Invalid_FDR_P_Values": None,
        "Minimum_Valid_Pairs": None,
        "Maximum_Valid_Pairs": None
    }])

display(association_audit_summary)
```

```python
# ============================================================
# STEP 5: SENSITIVITY ANALYSIS AUDIT
# Purpose:
#   Validate the sample-size and correlation-method sensitivity
#   outputs.
# Method:
#   Check thresholds, methods and required result tables.
# Output:
#   sensitivity_audit_summary
# ============================================================

sensitivity_objects = {
    "Threshold Sensitivity": "threshold_sensitivity_results",
    "Threshold Summary": "threshold_sensitivity_summary",
    "Method Sensitivity": "correlation_method_sensitivity",
    "Method Agreement": "correlation_method_agreement",
    "Monitoring Coverage": "city_pollutant_coverage",
    "High Coverage Analysis": "high_coverage_association_results",
    "Coverage Comparison": "coverage_sensitivity_comparison"
}

sensitivity_audit_records = []

for analysis_name, variable_name in sensitivity_objects.items():

    available = (
        variable_name in globals()
        and isinstance(globals()[variable_name], pd.DataFrame)
    )

    if available:

        result_df = globals()[variable_name]

        sensitivity_audit_records.append({
            "Analysis": analysis_name,
            "Available": True,
            "Rows": len(result_df),
            "Columns": len(result_df.columns),
            "Missing_Cells": int(result_df.isna().sum().sum())
        })

    else:

        sensitivity_audit_records.append({
            "Analysis": analysis_name,
            "Available": False,
            "Rows": None,
            "Columns": None,
            "Missing_Cells": None
        })

sensitivity_audit_summary = pd.DataFrame(
    sensitivity_audit_records
)

display(sensitivity_audit_summary)
```

```python
# ============================================================
# STEP 6: ENVIRONMENTAL RANKING AUDIT
# Purpose:
#   Validate scenario rankings and rank-stability results.
# Method:
#   Check rank completeness, duplicate city entries,
#   and rank correlation bounds.
# Output:
#   ranking_audit_summary
# ============================================================

ranking_audit_records = []

if (
    "scenario_ranking_results" in globals()
    and isinstance(scenario_ranking_results, pd.DataFrame)
):

    ranking_df = scenario_ranking_results.copy()

    rank_columns = [
        col for col in ranking_df.columns
        if col.endswith("_Rank")
    ]

    for column in rank_columns:

        ranking_audit_records.append({
            "Rank_Column": column,
            "Missing_Ranks": int(ranking_df[column].isna().sum()),
            "Unique_Ranks": int(ranking_df[column].nunique()),
            "Minimum_Rank": ranking_df[column].min(),
            "Maximum_Rank": ranking_df[column].max()
        })

ranking_audit_summary = pd.DataFrame(
    ranking_audit_records
)

display(ranking_audit_summary)

if (
    "ranking_agreement_results" in globals()
    and isinstance(ranking_agreement_results, pd.DataFrame)
):

    print("Ranking agreement validation:")

    for column in [
        "Spearman_Rank_Correlation",
        "Kendall_Rank_Correlation"
    ]:

        if column in ranking_agreement_results.columns:

            invalid_count = (
                ~ranking_agreement_results[column]
                .dropna()
                .between(-1, 1)
            ).sum()

            print(
                column,
                "- invalid values:",
                invalid_count
            )
```

```python
# ============================================================
# STEP 7: CONSOLIDATED MODULE 7 AUDIT
# Purpose:
#   Bring together the availability of the major analytical
#   outputs across Submodules 7.1 to 7.5.
# Method:
#   Generate a consolidated audit status table.
# Output:
#   module7_comprehensive_audit
# ============================================================

module7_section_variables = {
    "7.1 City-Cluster Bootstrap": [
        "cluster_ci_df"
    ],

    "7.2 Temporal Dependence": [
        "city_temporal_dependence",
        "annual_temporal_dependence",
        "temporal_overall_summary"
    ],

    "7.3 Pollutant Association": [
        "city_pollutant_correlations",
        "pollutant_city_association_summary",
        "pooled_vs_city_association"
    ],

    "7.4 Sensitivity Analysis": [
        "threshold_sensitivity_results",
        "correlation_method_sensitivity",
        "city_pollutant_coverage",
        "coverage_sensitivity_comparison"
    ],

    "7.5 Risk Ranking Validation": [
        "scenario_ranking_results",
        "ranking_agreement_results",
        "city_ranking_stability"
    ]
}

module7_section_records = []

for section, variables in module7_section_variables.items():

    available_count = sum(
        variable in globals()
        and globals()[variable] is not None
        for variable in variables
    )

    total_count = len(variables)

    module7_section_records.append({
        "Module_Section": section,
        "Expected_Outputs": total_count,
        "Available_Outputs": available_count,
        "Unavailable_Outputs": total_count - available_count,
        "Availability_Percent": (
            available_count / total_count * 100
        )
    })

module7_comprehensive_audit = pd.DataFrame(
    module7_section_records
)

display(module7_comprehensive_audit)

print(
    "Overall expected output availability:",
    round(
        module7_comprehensive_audit["Available_Outputs"].sum()
        / module7_comprehensive_audit["Expected_Outputs"].sum()
        * 100,
        2
    ),
    "%"
)
```

```python
# ============================================================
# MODULE 8.1 — LONG-TERM AQI TREND AND SEASONAL DECOMPOSITION
# STEP 1: PREPARE MONTHLY NATIONAL AQI SERIES
# Purpose:
#   Aggregate daily city-level AQI observations into a
#   monthly national average.
# Method:
#   Calculate the mean AQI across available city observations
#   for each calendar month.
# Output:
#   national_monthly_aqi
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

time_series_data = air_final[
    ["Date", "City", "AQI"]
].copy()

time_series_data["Date"] = pd.to_datetime(
    time_series_data["Date"],
    errors="coerce"
)

time_series_data = time_series_data.dropna(
    subset=["Date", "AQI"]
)

time_series_data["Month"] = (
    time_series_data["Date"].dt.to_period("M")
)

national_monthly_aqi = (
    time_series_data
    .groupby("Month")
    .agg(
        Mean_AQI=("AQI", "mean"),
        Median_AQI=("AQI", "median"),
        City_Observations=("AQI", "count"),
        Cities_Reporting=("City", "nunique")
    )
    .reset_index()
)

national_monthly_aqi["Month_Start"] = (
    national_monthly_aqi["Month"].dt.to_timestamp()
)

national_monthly_aqi = national_monthly_aqi.sort_values(
    "Month_Start"
).reset_index(drop=True)

print("Monthly national AQI records:",
      len(national_monthly_aqi))

print(
    "Period:",
    national_monthly_aqi["Month_Start"].min(),
    "to",
    national_monthly_aqi["Month_Start"].max()
)

display(national_monthly_aqi.head())
```

```python
# ============================================================
# STEP 2: MONTHLY CONTINUITY AND COVERAGE AUDIT
# Purpose:
#   Check whether the national monthly series has a complete
#   and continuous monthly time index.
# Method:
#   Compare observed months against a complete monthly calendar.
# Output:
#   monthly_continuity_audit
# ============================================================

complete_month_range = pd.period_range(
    start=national_monthly_aqi["Month"].min(),
    end=national_monthly_aqi["Month"].max(),
    freq="M"
)

observed_months = set(
    national_monthly_aqi["Month"]
)

missing_months = [
    month for month in complete_month_range
    if month not in observed_months
]

monthly_continuity_audit = pd.DataFrame([{
    "Expected_Months": len(complete_month_range),
    "Observed_Months": len(observed_months),
    "Missing_Months": len(missing_months),
    "Continuous_Monthly_Series": len(missing_months) == 0,
    "Minimum_City_Observations": (
        national_monthly_aqi["City_Observations"].min()
    ),
    "Maximum_City_Observations": (
        national_monthly_aqi["City_Observations"].max()
    ),
    "Minimum_Cities_Reporting": (
        national_monthly_aqi["Cities_Reporting"].min()
    ),
    "Maximum_Cities_Reporting": (
        national_monthly_aqi["Cities_Reporting"].max()
    )
}])

display(monthly_continuity_audit)

print("Missing months:", missing_months)
```

```python
# ============================================================
# STEP 3: MONTHLY NATIONAL AQI TREND
# Purpose:
#   Visualize the long-term movement in national monthly AQI.
# Method:
#   Plot monthly mean and median AQI.
# Output:
#   national_monthly_aqi_trend.png
# ============================================================

plt.figure(figsize=(16, 7))

plt.plot(
    national_monthly_aqi["Month_Start"],
    national_monthly_aqi["Mean_AQI"],
    label="Monthly Mean AQI",
    linewidth=1.8
)

plt.plot(
    national_monthly_aqi["Month_Start"],
    national_monthly_aqi["Median_AQI"],
    label="Monthly Median AQI",
    linewidth=1.5,
    alpha=0.8
)

plt.title(
    "Long-Term Monthly National AQI Trend",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("AQI")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "national_monthly_aqi_trend.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 4: ROLLING AQI AVERAGES
# Purpose:
#   Smooth short-term variation and examine longer-term AQI
#   movements.
# Method:
#   Calculate 3-, 6- and 12-month rolling averages.
# Output:
#   national_monthly_aqi with rolling-average columns.
# ============================================================

for window in [3, 6, 12]:

    national_monthly_aqi[
        f"AQI_Rolling_{window}M"
    ] = (
        national_monthly_aqi["Mean_AQI"]
        .rolling(
            window=window,
            min_periods=window
        )
        .mean()
    )

display(national_monthly_aqi.tail(15))

plt.figure(figsize=(16, 7))

plt.plot(
    national_monthly_aqi["Month_Start"],
    national_monthly_aqi["Mean_AQI"],
    label="Monthly AQI",
    alpha=0.45
)

for window in [3, 6, 12]:

    plt.plot(
        national_monthly_aqi["Month_Start"],
        national_monthly_aqi[f"AQI_Rolling_{window}M"],
        label=f"{window}-Month Rolling Average"
    )

plt.title(
    "National AQI — Rolling Average Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("AQI")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "national_aqi_rolling_averages.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 5: STL SEASONAL DECOMPOSITION
# Purpose:
#   Separate the monthly AQI series into trend, seasonal
#   and residual components.
# Method:
#   STL decomposition with annual period of 12 months.
# Output:
#   national_stl_result
# ============================================================

from statsmodels.tsa.seasonal import STL

monthly_series = (
    national_monthly_aqi
    .set_index("Month_Start")["Mean_AQI"]
    .asfreq("MS")
)

if monthly_series.isna().any():

    print(
        "STL decomposition skipped: the monthly series "
        "contains missing months or missing AQI values."
    )

elif len(monthly_series) < 36:

    print(
        "STL decomposition skipped: at least 36 monthly "
        "observations are required for this analysis."
    )

else:

    national_stl_result = STL(
        monthly_series,
        period=12,
        robust=True
    ).fit()

    print("STL decomposition completed.")

    print(
        "Number of monthly observations:",
        len(monthly_series)
    )
```

```python
# ============================================================
# STEP 6: STL COMPONENT VISUALIZATION
# Purpose:
#   Visualize observed AQI, trend, seasonality and residuals.
# Method:
#   Plot the four STL decomposition components.
# Output:
#   national_aqi_stl_decomposition.png
# ============================================================

if "national_stl_result" in globals():

    fig, axes = plt.subplots(
        4,
        1,
        figsize=(16, 12),
        sharex=True
    )

    components = [
        (monthly_series, "Observed AQI"),
        (national_stl_result.trend, "Trend"),
        (national_stl_result.seasonal, "Seasonal"),
        (national_stl_result.resid, "Residual")
    ]

    for ax, (component, label) in zip(
        axes,
        components
    ):

        ax.plot(component.index, component.values)
        ax.set_ylabel(label)
        ax.grid(alpha=0.25)

    axes[0].set_title(
        "STL Decomposition of Monthly National AQI",
        fontsize=15,
        fontweight="bold"
    )

    axes[-1].set_xlabel("Month")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            visualization_dir,
            "national_aqi_stl_decomposition.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:

    print(
        "No STL result available. Complete Step 5 "
        "successfully before running this visualization."
    )
```

```python
# ============================================================
# STEP 7: SEASONAL AND TREND STRENGTH
# Purpose:
#   Quantify the strength of the seasonal and trend components.
# Method:
#   Use variance-based descriptive decomposition measures.
# Output:
#   national_decomposition_strength
# ============================================================

if "national_stl_result" in globals():

    seasonal_component = national_stl_result.seasonal
    trend_component = national_stl_result.trend
    residual_component = national_stl_result.resid

    seasonal_strength = max(
        0,
        1 - (
            np.var(residual_component)
            /
            np.var(seasonal_component + residual_component)
        )
    )

    trend_strength = max(
        0,
        1 - (
            np.var(residual_component)
            /
            np.var(trend_component + residual_component)
        )
    )

    national_decomposition_strength = pd.DataFrame([{
        "Seasonal_Strength": seasonal_strength,
        "Trend_Strength": trend_strength,
        "Residual_Variance": np.var(residual_component),
        "Seasonal_Variance": np.var(seasonal_component),
        "Trend_Variance": np.var(trend_component)
    }])

    display(national_decomposition_strength)

    national_decomposition_strength.to_csv(
        os.path.join(
            report_dir,
            "national_aqi_decomposition_strength.csv"
        ),
        index=False
    )

else:

    print("STL decomposition result unavailable.")
```

```python
# ============================================================
# STEP 8: CITY-LEVEL MONTHLY AQI SERIES
# Purpose:
#   Prepare city-specific monthly AQI observations.
# Method:
#   Aggregate daily AQI observations by city and month.
# Output:
#   city_monthly_aqi
# ============================================================

city_monthly_aqi = (
    time_series_data
    .assign(
        Month_Start=lambda df:
            df["Date"].dt.to_period("M").dt.to_timestamp()
    )
    .groupby(["City", "Month_Start"])
    .agg(
        Mean_AQI=("AQI", "mean"),
        Valid_Daily_Observations=("AQI", "count")
    )
    .reset_index()
)

city_monthly_aqi = city_monthly_aqi.sort_values(
    ["City", "Month_Start"]
).reset_index(drop=True)

city_monthly_coverage = (
    city_monthly_aqi
    .groupby("City")
    .agg(
        Monthly_Observations=("Month_Start", "count"),
        First_Month=("Month_Start", "min"),
        Last_Month=("Month_Start", "max")
    )
    .reset_index()
)

display(city_monthly_coverage)

city_monthly_aqi.to_csv(
    os.path.join(
        processed_dir,
        "city_monthly_aqi.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 9: CITY-LEVEL STL DECOMPOSITION
# Purpose:
#   Identify trend and seasonal patterns for eligible cities.
# Method:
#   Apply annual STL decomposition only to complete monthly
#   city-level series with at least 36 observations.
# Output:
#   city_stl_summary
# ============================================================

city_stl_results = []

for city, city_data in city_monthly_aqi.groupby("City"):

    city_series = (
        city_data
        .set_index("Month_Start")["Mean_AQI"]
        .asfreq("MS")
    )

    if (
        len(city_series) >= 36
        and city_series.notna().all()
    ):

        city_stl = STL(
            city_series,
            period=12,
            robust=True
        ).fit()

        city_seasonal_strength = max(
            0,
            1 - (
                np.var(city_stl.resid)
                /
                np.var(
                    city_stl.seasonal + city_stl.resid
                )
            )
        )

        city_trend_strength = max(
            0,
            1 - (
                np.var(city_stl.resid)
                /
                np.var(
                    city_stl.trend + city_stl.resid
                )
            )
        )

        city_stl_results.append({
            "City": city,
            "Monthly_Observations": len(city_series),
            "Seasonal_Strength": city_seasonal_strength,
            "Trend_Strength": city_trend_strength,
            "Residual_Variance": np.var(city_stl.resid),
            "Decomposition_Status": "Completed"
        })

    else:

        city_stl_results.append({
            "City": city,
            "Monthly_Observations": len(city_series),
            "Seasonal_Strength": np.nan,
            "Trend_Strength": np.nan,
            "Residual_Variance": np.nan,
            "Decomposition_Status": "Insufficient or incomplete series"
        })

city_stl_summary = pd.DataFrame(
    city_stl_results
)

display(
    city_stl_summary.sort_values(
        "Seasonal_Strength",
        ascending=False
    )
)

city_stl_summary.to_csv(
    os.path.join(
        report_dir,
        "city_level_stl_decomposition_summary.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 10: EXPORT LONG-TERM TREND AND DECOMPOSITION RESULTS
# Purpose:
#   Preserve the monthly AQI analysis and decomposition audit.
# Method:
#   Export monthly data, continuity audit and city-level results.
# Output:
#   Submodule 8.1 report files.
# ============================================================

national_monthly_aqi.to_csv(
    os.path.join(
        report_dir,
        "national_monthly_aqi_analysis.csv"
    ),
    index=False
)

monthly_continuity_audit.to_csv(
    os.path.join(
        report_dir,
        "national_monthly_continuity_audit.csv"
    ),
    index=False
)

city_monthly_coverage.to_csv(
    os.path.join(
        report_dir,
        "city_monthly_aqi_coverage.csv"
    ),
    index=False
)

city_stl_summary.to_csv(
    os.path.join(
        report_dir,
        "city_level_stl_decomposition_summary.csv"
    ),
    index=False
)

print("Submodule 8.1 exports completed.")
print("Reports directory:", report_dir)
print("Visualization directory:", visualization_dir)
```

```python
# ============================================================
# MODULE 8.2 — STATIONARITY AND TIME-SERIES DIAGNOSTICS
# STEP 1: PREPARE MONTHLY AQI SERIES
# Purpose:
#   Prepare the national monthly AQI series for stationarity tests.
# Method:
#   Validate monthly frequency and missing observations.
# Output:
#   stationarity_series
# ============================================================

stationarity_series = (
    national_monthly_aqi
    .set_index("Month_Start")["Mean_AQI"]
    .asfreq("MS")
    .astype(float)
)

print("Observations:", len(stationarity_series))
print("Missing values:", stationarity_series.isna().sum())
print("Start:", stationarity_series.index.min())
print("End:", stationarity_series.index.max())

if stationarity_series.isna().any():
    print("Warning: Missing months must be investigated before testing.")
else:
    print("Monthly series is ready for stationarity analysis.")

display(stationarity_series.head())
```

```python
# ============================================================
# STEP 2: ROLLING STATISTICS
# Purpose:
#   Examine whether the mean and variability change over time.
# Method:
#   Calculate 12-month rolling mean and standard deviation.
# Output:
#   rolling_stationarity_data
# ============================================================

rolling_stationarity_data = pd.DataFrame({
    "AQI": stationarity_series,
    "Rolling_Mean_12M": stationarity_series.rolling(
        window=12
    ).mean(),
    "Rolling_Std_12M": stationarity_series.rolling(
        window=12
    ).std()
})

fig, axes = plt.subplots(2, 1, figsize=(15, 9), sharex=True)

axes[0].plot(
    rolling_stationarity_data.index,
    rolling_stationarity_data["AQI"],
    label="Monthly AQI"
)

axes[0].plot(
    rolling_stationarity_data.index,
    rolling_stationarity_data["Rolling_Mean_12M"],
    label="12-Month Rolling Mean"
)

axes[0].set_title("Monthly AQI and Rolling Mean")
axes[0].set_ylabel("AQI")
axes[0].legend()
axes[0].grid(alpha=0.25)

axes[1].plot(
    rolling_stationarity_data.index,
    rolling_stationarity_data["Rolling_Std_12M"],
    label="12-Month Rolling Standard Deviation"
)

axes[1].set_title("Rolling Standard Deviation")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Standard Deviation")
axes[1].legend()
axes[1].grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "monthly_aqi_rolling_stationarity.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 3: AUGMENTED DICKEY-FULLER TEST
# Purpose:
#   Test the monthly AQI series for a unit root.
# Method:
#   ADF test with automatic lag selection using AIC.
# Output:
#   adf_original_results
# ============================================================

from statsmodels.tsa.stattools import adfuller

if stationarity_series.isna().any():
    raise ValueError("Resolve missing monthly observations first.")

if len(stationarity_series) < 24:
    raise ValueError("Insufficient observations for this test.")

adf_result = adfuller(
    stationarity_series,
    autolag="AIC"
)

adf_original_results = pd.DataFrame([{
    "Test": "Augmented Dickey-Fuller",
    "ADF_Statistic": adf_result[0],
    "P_Value": adf_result[1],
    "Used_Lags": adf_result[2],
    "Observations": adf_result[3],
    "Critical_Value_1%": adf_result[4]["1%"],
    "Critical_Value_5%": adf_result[4]["5%"],
    "Critical_Value_10%": adf_result[4]["10%"],
    "Conclusion_At_5_Percent": (
        "Evidence against unit root"
        if adf_result[1] < 0.05
        else "Insufficient evidence against unit root"
    )
}])

display(adf_original_results)
```

```python
# ============================================================
# STEP 4: KPSS STATIONARITY TEST
# Purpose:
#   Test whether the monthly AQI series is level-stationary.
# Method:
#   KPSS test with constant-only specification.
# Output:
#   kpss_original_results
# ============================================================

from statsmodels.tsa.stattools import kpss

kpss_statistic, kpss_p_value, kpss_lags, kpss_critical = (
    kpss(
        stationarity_series,
        regression="c",
        nlags="auto"
    )
)

kpss_original_results = pd.DataFrame([{
    "Test": "KPSS",
    "KPSS_Statistic": kpss_statistic,
    "P_Value": kpss_p_value,
    "Used_Lags": kpss_lags,
    "Critical_Value_1%": kpss_critical["1%"],
    "Critical_Value_5%": kpss_critical["5%"],
    "Critical_Value_10%": kpss_critical["10%"],
    "Conclusion_At_5_Percent": (
        "Evidence against stationarity"
        if kpss_p_value < 0.05
        else "Insufficient evidence against stationarity"
    )
}])

display(kpss_original_results)
```

```python
# ============================================================
# STEP 5: FIRST-ORDER DIFFERENCING
# Purpose:
#   Transform monthly AQI into month-to-month changes.
# Method:
#   Calculate the first difference of the original series.
# Output:
#   differenced_aqi
# ============================================================

differenced_aqi = stationarity_series.diff().dropna()

print("Original observations:", len(stationarity_series))
print("Differenced observations:", len(differenced_aqi))
print("Differenced mean:", differenced_aqi.mean())
print("Differenced standard deviation:", differenced_aqi.std())

display(differenced_aqi.head())
```

```python
# ============================================================
# STEP 6: DIFFERENCED AQI VISUALIZATION
# Purpose:
#   Examine month-to-month AQI changes after differencing.
# Method:
#   Plot first-order differences and a zero reference line.
# Output:
#   monthly_aqi_first_difference.png
# ============================================================

plt.figure(figsize=(15, 6))

plt.plot(
    differenced_aqi.index,
    differenced_aqi.values,
    label="First Difference of AQI"
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Month-to-Month Change in National AQI",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("AQI Change")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "monthly_aqi_first_difference.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 7: DIFFERENCED SERIES STATIONARITY TESTS
# Purpose:
#   Examine whether first-order differencing improves
#   stationarity characteristics.
# Method:
#   Apply ADF and KPSS to the differenced series.
# Output:
#   differenced_stationarity_results
# ============================================================

adf_diff = adfuller(
    differenced_aqi,
    autolag="AIC"
)

kpss_diff_stat, kpss_diff_p, kpss_diff_lags, kpss_diff_critical = (
    kpss(
        differenced_aqi,
        regression="c",
        nlags="auto"
    )
)

differenced_stationarity_results = pd.DataFrame([
    {
        "Series": "First-Differenced AQI",
        "Test": "ADF",
        "Statistic": adf_diff[0],
        "P_Value": adf_diff[1],
        "Used_Lags": adf_diff[2],
        "Conclusion_At_5_Percent": (
            "Evidence against unit root"
            if adf_diff[1] < 0.05
            else "Insufficient evidence against unit root"
        )
    },
    {
        "Series": "First-Differenced AQI",
        "Test": "KPSS",
        "Statistic": kpss_diff_stat,
        "P_Value": kpss_diff_p,
        "Used_Lags": kpss_diff_lags,
        "Conclusion_At_5_Percent": (
            "Evidence against stationarity"
            if kpss_diff_p < 0.05
            else "Insufficient evidence against stationarity"
        )
    }
])

display(differenced_stationarity_results)
```

```python
# ============================================================
# STEP 8: STATIONARITY COMPARISON
# Purpose:
#   Consolidate the ADF and KPSS results for both series.
# Method:
#   Combine original and differenced test outputs.
# Output:
#   stationarity_comparison
# ============================================================

original_adf = pd.DataFrame([{
    "Series": "Original AQI",
    "Test": "ADF",
    "Statistic": adf_result[0],
    "P_Value": adf_result[1],
    "Used_Lags": adf_result[2]
}])

original_kpss = pd.DataFrame([{
    "Series": "Original AQI",
    "Test": "KPSS",
    "Statistic": kpss_statistic,
    "P_Value": kpss_p_value,
    "Used_Lags": kpss_lags
}])

differenced_comparison = differenced_stationarity_results[
    ["Series", "Test", "Statistic", "P_Value", "Used_Lags"]
]

stationarity_comparison = pd.concat(
    [
        original_adf,
        original_kpss,
        differenced_comparison
    ],
    ignore_index=True
)

display(stationarity_comparison)

stationarity_comparison.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_stationarity_comparison.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 9: EXPORT STATIONARITY ANALYSIS
# Purpose:
#   Save stationarity diagnostics and differenced AQI data.
# Method:
#   Export rolling statistics, test results and differences.
# Output:
#   Submodule 8.2 reports.
# ============================================================

rolling_stationarity_data.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_rolling_stationarity.csv"
    )
)

differenced_aqi.rename(
    "AQI_First_Difference"
).to_csv(
    os.path.join(
        processed_dir,
        "national_aqi_first_difference.csv"
)

)

stationarity_comparison.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_stationarity_comparison.csv"
    ),
    index=False
)

print("Submodule 8.2 exports completed.")
print("Reports directory:", report_dir)
```

```python
# ============================================================
# MODULE 8.3 — AUTOCORRELATION AND PARTIAL AUTOCORRELATION
# STEP 1: PREPARE TIME-SERIES INPUTS
# Purpose:
#   Prepare original and differenced AQI for lag analysis.
# Method:
#   Validate monthly continuity and select a safe lag count.
# Output:
#   acf_series, pacf_series, maximum_lags
# ============================================================

acf_series = stationarity_series.dropna().astype(float)
pacf_series = differenced_aqi.dropna().astype(float)

if len(acf_series) < 12:
    raise ValueError("At least 12 monthly observations are required.")

if len(pacf_series) < 12:
    raise ValueError("At least 12 differenced observations are required.")

maximum_lags = min(
    24,
    len(acf_series) // 2 - 1,
    len(pacf_series) // 2 - 1
)

print("Original observations:", len(acf_series))
print("Differenced observations:", len(pacf_series))
print("Maximum lags:", maximum_lags)
```

```python
# ============================================================
# STEP 2: ORIGINAL AQI AUTOCORRELATION
# Purpose:
#   Examine monthly AQI dependence at different time lags.
# Method:
#   Plot the Autocorrelation Function.
# Output:
#   original_aqi_acf.png
# ============================================================

from statsmodels.graphics.tsaplots import plot_acf

fig, ax = plt.subplots(figsize=(13, 6))

plot_acf(
    acf_series,
    lags=maximum_lags,
    alpha=0.05,
    ax=ax
)

ax.set_title(
    "Autocorrelation Function — Original Monthly AQI",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Monthly Lag")
ax.set_ylabel("Autocorrelation")
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "original_aqi_acf.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 3: ORIGINAL AQI PARTIAL AUTOCORRELATION
# Purpose:
#   Identify direct monthly lag relationships.
# Method:
#   Plot the Partial Autocorrelation Function using Yule-Walker.
# Output:
#   original_aqi_pacf.png
# ============================================================

from statsmodels.graphics.tsaplots import plot_pacf

fig, ax = plt.subplots(figsize=(13, 6))

plot_pacf(
    acf_series,
    lags=maximum_lags,
    method="ywm",
    alpha=0.05,
    ax=ax
)

ax.set_title(
    "Partial Autocorrelation Function — Original Monthly AQI",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Monthly Lag")
ax.set_ylabel("Partial Autocorrelation")
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "original_aqi_pacf.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 3: ORIGINAL AQI PARTIAL AUTOCORRELATION
# Purpose:
#   Identify direct monthly lag relationships.
# Method:
#   Plot the Partial Autocorrelation Function using Yule-Walker.
# Output:
#   original_aqi_pacf.png
# ============================================================

from statsmodels.graphics.tsaplots import plot_pacf

fig, ax = plt.subplots(figsize=(13, 6))

plot_pacf(
    acf_series,
    lags=maximum_lags,
    method="ywm",
    alpha=0.05,
    ax=ax
)

ax.set_title(
    "Partial Autocorrelation Function — Original Monthly AQI",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Monthly Lag")
ax.set_ylabel("Partial Autocorrelation")
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "original_aqi_pacf.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 4: DIFFERENCED AQI AUTOCORRELATION
# Purpose:
#   Examine autocorrelation after removing first-order changes.
# Method:
#   Plot ACF of the differenced monthly AQI series.
# Output:
#   differenced_aqi_acf.png
# ============================================================

differenced_lags = min(
    24,
    len(pacf_series) // 2 - 1
)

fig, ax = plt.subplots(figsize=(13, 6))

plot_acf(
    pacf_series,
    lags=differenced_lags,
    alpha=0.05,
    ax=ax
)

ax.set_title(
    "Autocorrelation Function — First-Differenced AQI",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Monthly Lag")
ax.set_ylabel("Autocorrelation")
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "differenced_aqi_acf.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 5: DIFFERENCED AQI PARTIAL AUTOCORRELATION
# Purpose:
#   Identify direct lag relationships in monthly AQI changes.
# Method:
#   Plot PACF of the differenced series.
# Output:
#   differenced_aqi_pacf.png
# ============================================================

fig, ax = plt.subplots(figsize=(13, 6))

plot_pacf(
    pacf_series,
    lags=differenced_lags,
    method="ywm",
    alpha=0.05,
    ax=ax
)

ax.set_title(
    "Partial Autocorrelation Function — First-Differenced AQI",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Monthly Lag")
ax.set_ylabel("Partial Autocorrelation")
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "differenced_aqi_pacf.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 6: NUMERICAL AUTOCORRELATION ANALYSIS
# Purpose:
#   Extract numerical ACF and PACF values for reporting.
# Method:
#   Calculate correlations at each selected lag.
# Output:
#   monthly_acf_pacf_results
# ============================================================

from statsmodels.tsa.stattools import acf, pacf

original_acf_values = acf(
    acf_series,
    nlags=maximum_lags,
    fft=False
)

original_pacf_values = pacf(
    acf_series,
    nlags=maximum_lags,
    method="ywm"
)

differenced_acf_values = acf(
    pacf_series,
    nlags=differenced_lags,
    fft=False
)

differenced_pacf_values = pacf(
    pacf_series,
    nlags=differenced_lags,
    method="ywm"
)

monthly_acf_pacf_results = pd.DataFrame({
    "Lag": range(maximum_lags + 1),
    "Original_ACF": original_acf_values,
    "Original_PACF": original_pacf_values
})

monthly_acf_pacf_results["Differenced_ACF"] = np.nan
monthly_acf_pacf_results["Differenced_PACF"] = np.nan

monthly_acf_pacf_results.loc[
    :differenced_lags,
    "Differenced_ACF"
] = differenced_acf_values

monthly_acf_pacf_results.loc[
    :differenced_lags,
    "Differenced_PACF"
] = differenced_pacf_values

display(monthly_acf_pacf_results)
```

```python
# ============================================================
# STEP 7: ANNUAL SEASONAL LAG ANALYSIS
# Purpose:
#   Examine annual recurring AQI relationships.
# Method:
#   Extract ACF and PACF at seasonal lags 12 and 24.
# Output:
#   annual_seasonal_autocorrelation
# ============================================================

seasonal_lags = [12, 24]

annual_seasonal_autocorrelation = (
    monthly_acf_pacf_results[
        monthly_acf_pacf_results["Lag"].isin(seasonal_lags)
    ]
    .copy()
)

display(annual_seasonal_autocorrelation)
```

```python
# ============================================================
# STEP 8: LJUNG-BOX SERIAL CORRELATION TEST
# Purpose:
#   Test whether multiple monthly autocorrelations are
#   jointly statistically significant.
# Method:
#   Apply Ljung-Box tests at selected lag groups.
# Output:
#   ljung_box_results
# ============================================================

from statsmodels.stats.diagnostic import acorr_ljungbox

ljung_box_records = []

for series_name, series in [
    ("Original AQI", acf_series),
    ("Differenced AQI", pacf_series)
]:

    test_lags = [
        lag for lag in [6, 12, 18]
        if lag < len(series) // 2
    ]

    if test_lags:

        test_result = acorr_ljungbox(
            series,
            lags=test_lags,
            return_df=True
        )

        for lag, row in test_result.iterrows():

            ljung_box_records.append({
                "Series": series_name,
                "Lag": int(lag),
                "Ljung_Box_Statistic": row["lb_stat"],
                "P_Value": row["lb_pvalue"],
                "Conclusion_At_5_Percent": (
                    "Evidence of serial correlation"
                    if row["lb_pvalue"] < 0.05
                    else "Insufficient evidence of serial correlation"
                )
            })

ljung_box_results = pd.DataFrame(
    ljung_box_records
)

display(ljung_box_results)
```

```python
# ============================================================
# STEP 9: CITY-LEVEL MONTHLY LAG-1 AUTOCORRELATION
# Purpose:
#   Compare short-term AQI persistence across cities.
# Method:
#   Calculate Pearson correlation between consecutive monthly
#   AQI observations for each city.
#   Require at least 12 consecutive monthly pairs.
# Output:
#   city_monthly_lag1_autocorrelation
# ============================================================

city_lag1_records = []

for city, city_data in city_monthly_aqi.groupby("City"):

    city_series = (
        city_data
        .set_index("Month_Start")["Mean_AQI"]
        .sort_index()
        .asfreq("MS")
    )

    lagged_data = pd.DataFrame({
        "Current_AQI": city_series,
        "Previous_Month_AQI": city_series.shift(1)
    }).dropna()

    valid_pairs = len(lagged_data)

    if (
        valid_pairs >= 12
        and lagged_data["Current_AQI"].nunique() > 1
        and lagged_data["Previous_Month_AQI"].nunique() > 1
    ):

        lag1_correlation = lagged_data[
            "Current_AQI"
        ].corr(
            lagged_data["Previous_Month_AQI"]
        )

    else:
        lag1_correlation = np.nan

    city_lag1_records.append({
        "City": city,
        "Consecutive_Month_Pairs": valid_pairs,
        "Lag1_Autocorrelation": lag1_correlation
    })

city_monthly_lag1_autocorrelation = pd.DataFrame(
    city_lag1_records
).sort_values(
    "Lag1_Autocorrelation",
    ascending=False
)

display(city_monthly_lag1_autocorrelation)
```

```python
# ============================================================
# STEP 10: CITY-LEVEL AUTOCORRELATION VISUALIZATION
# Purpose:
#   Compare monthly AQI persistence across cities.
# Method:
#   Horizontal bar chart of city-level lag-1 correlations.
# Output:
#   city_monthly_lag1_autocorrelation.png
# ============================================================

city_plot_data = (
    city_monthly_lag1_autocorrelation
    .dropna(subset=["Lag1_Autocorrelation"])
    .sort_values("Lag1_Autocorrelation")
)

plt.figure(figsize=(12, 9))

sns.barplot(
    data=city_plot_data,
    x="Lag1_Autocorrelation",
    y="City"
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "City-Level Monthly AQI Lag-1 Autocorrelation",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Lag-1 Autocorrelation")
plt.ylabel("City")
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "city_monthly_lag1_autocorrelation.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 11: EXPORT AUTOCORRELATION ANALYSIS
# Purpose:
#   Preserve national and city-level lag-analysis results.
# Method:
#   Export ACF/PACF, seasonal lag, Ljung-Box and city results.
# Output:
#   Submodule 8.3 CSV reports.
# ============================================================

monthly_acf_pacf_results.to_csv(
    os.path.join(
        report_dir,
        "national_monthly_acf_pacf_results.csv"
    ),
    index=False
)

annual_seasonal_autocorrelation.to_csv(
    os.path.join(
        report_dir,
        "annual_seasonal_autocorrelation.csv"
    ),
    index=False
)

ljung_box_results.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_ljung_box_results.csv"
    ),
    index=False
)

city_monthly_lag1_autocorrelation.to_csv(
    os.path.join(
        report_dir,
        "city_monthly_lag1_autocorrelation.csv"
    ),
    index=False
)

print("Submodule 8.3 exports completed.")
print("Reports directory:", report_dir)
print("Visualization directory:", visualization_dir)
```

```python
# ============================================================
# MODULE 8.4 — STATISTICAL TIME-SERIES FORECASTING
# STEP 1: PREPARE FORECASTING DATA
# Purpose:
#   Prepare the national monthly AQI series for forecasting.
# Method:
#   Validate monthly continuity and reserve the final
#   12 months for out-of-sample testing.
# Output:
#   forecast_series, train_series, test_series
# ============================================================

forecast_series = (
    national_monthly_aqi
    .set_index("Month_Start")["Mean_AQI"]
    .asfreq("MS")
    .astype(float)
)

if forecast_series.isna().any():
    raise ValueError(
        "Missing monthly observations detected. "
        "Resolve the monthly series before forecasting."
    )

if len(forecast_series) < 36:
    raise ValueError(
        "At least 36 monthly observations are required."
    )

forecast_horizon = 12

train_series = forecast_series.iloc[:-forecast_horizon]
test_series = forecast_series.iloc[-forecast_horizon:]

print("Total observations:", len(forecast_series))
print("Training observations:", len(train_series))
print("Testing observations:", len(test_series))
print("Training period:", train_series.index.min(),
      "to", train_series.index.max())
print("Testing period:", test_series.index.min(),
      "to", test_series.index.max())
```

```python
# ============================================================
# STEP 2: CHRONOLOGICAL TRAIN-TEST VISUALIZATION
# Purpose:
#   Confirm that training observations precede test observations.
# Method:
#   Plot the chronological split.
# Output:
#   national_aqi_forecasting_split.png
# ============================================================

plt.figure(figsize=(15, 6))

plt.plot(
    train_series.index,
    train_series.values,
    label="Training Data"
)

plt.plot(
    test_series.index,
    test_series.values,
    label="Testing Data",
    linewidth=2
)

plt.axvline(
    test_series.index[0],
    linestyle="--",
    linewidth=1,
    label="Forecast Evaluation Start"
)

plt.title(
    "National Monthly AQI — Forecasting Train-Test Split",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("Mean AQI")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "national_aqi_forecasting_split.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 3: NAIVE FORECASTING BENCHMARK
# Purpose:
#   Establish a simple reference forecasting method.
# Method:
#   Repeat the final training observation across the
#   12-month forecast horizon.
# Output:
#   naive_forecast
# ============================================================

naive_forecast = pd.Series(
    train_series.iloc[-1],
    index=test_series.index,
    name="Naive_Forecast"
)

display(naive_forecast.to_frame())

print(
    "Last observed training AQI:",
    train_series.iloc[-1]
)
```

```python
# ============================================================
# STEP 4: SEASONAL NAIVE FORECAST
# Purpose:
#   Establish a benchmark that preserves annual seasonal patterns.
# Method:
#   Repeat the final 12 training observations.
# Output:
#   seasonal_naive_forecast
# ============================================================

seasonal_naive_forecast = pd.Series(
    train_series.iloc[-12:].to_numpy(),
    index=test_series.index,
    name="Seasonal_Naive_Forecast"
)

display(seasonal_naive_forecast.to_frame())
```

```python
# ============================================================
# STEP 5: HOLT-WINTERS EXPONENTIAL SMOOTHING
# Purpose:
#   Model AQI level, trend and annual seasonal patterns.
# Method:
#   Additive trend and additive seasonal components.
# Output:
#   holt_winters_model
#   holt_winters_forecast
# ============================================================

from statsmodels.tsa.holtwinters import ExponentialSmoothing

holt_winters_model = ExponentialSmoothing(
    train_series,
    trend="add",
    seasonal="add",
    seasonal_periods=12,
    initialization_method="estimated"
).fit(optimized=True)

holt_winters_forecast = holt_winters_model.forecast(
    forecast_horizon
)

holt_winters_forecast.name = "Holt_Winters_Forecast"

display(holt_winters_forecast.to_frame())

print("Holt-Winters model fitted.")
```

```python
# ============================================================
# STEP 6: SEASONAL ARIMA MODEL
# Purpose:
#   Fit a classical seasonal autoregressive integrated
#   moving-average model.
# Method:
#   SARIMA(1,1,1)(1,0,0,12).
# Output:
#   sarima_model
#   sarima_forecast
# ============================================================

from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_model = SARIMAX(
    train_series,
    order=(1, 1, 1),
    seasonal_order=(1, 0, 0, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
).fit(
    disp=False,
    maxiter=200
)

sarima_forecast_object = sarima_model.get_forecast(
    steps=forecast_horizon
)

sarima_forecast = sarima_forecast_object.predicted_mean

sarima_forecast.index = test_series.index
sarima_forecast.name = "SARIMA_Forecast"

display(sarima_forecast.to_frame())

print("SARIMA model fitted.")
print(sarima_model.summary())
```

```python
# ============================================================
# STEP 7: CONSOLIDATE FORECAST RESULTS
# Purpose:
#   Combine actual AQI and all statistical forecasts.
# Method:
#   Align each forecast to the same 12-month test period.
# Output:
#   forecast_comparison
# ============================================================

forecast_comparison = pd.DataFrame({
    "Actual_AQI": test_series,
    "Naive_Forecast": naive_forecast,
    "Seasonal_Naive_Forecast": seasonal_naive_forecast,
    "Holt_Winters_Forecast": holt_winters_forecast,
    "SARIMA_Forecast": sarima_forecast
})

display(forecast_comparison)

print("Forecast comparison shape:", forecast_comparison.shape)
```

```python
# ============================================================
# STEP 8: ACTUAL VERSUS FORECAST VISUALIZATION
# Purpose:
#   Compare all statistical forecasts against observed AQI.
# Method:
#   Plot the actual test series and four forecast methods.
# Output:
#   national_aqi_forecast_comparison.png
# ============================================================

plt.figure(figsize=(15, 7))

plt.plot(
    train_series.index,
    train_series.values,
    label="Training AQI",
    alpha=0.65
)

plt.plot(
    test_series.index,
    test_series.values,
    label="Actual Test AQI",
    linewidth=2.5
)

for column in [
    "Naive_Forecast",
    "Seasonal_Naive_Forecast",
    "Holt_Winters_Forecast",
    "SARIMA_Forecast"
]:

    plt.plot(
        forecast_comparison.index,
        forecast_comparison[column],
        marker="o",
        label=column
    )

plt.axvline(
    test_series.index[0],
    linestyle="--",
    linewidth=1
)

plt.title(
    "National Monthly AQI — Statistical Forecast Comparison",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("AQI")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "national_aqi_forecast_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 9: SARIMA FORECAST CONFIDENCE INTERVALS
# Purpose:
#   Quantify forecast uncertainty from the fitted SARIMA model.
# Method:
#   Extract 95% forecast confidence intervals.
# Output:
#   sarima_forecast_intervals
# ============================================================

sarima_interval = (
    sarima_forecast_object
    .conf_int(alpha=0.05)
)

sarima_interval.index = test_series.index

sarima_forecast_intervals = pd.DataFrame({
    "Forecast": sarima_forecast,
    "Lower_95": sarima_interval.iloc[:, 0],
    "Upper_95": sarima_interval.iloc[:, 1]
})

display(sarima_forecast_intervals)

plt.figure(figsize=(15, 6))

plt.plot(
    test_series.index,
    test_series.values,
    marker="o",
    label="Actual AQI"
)

plt.plot(
    sarima_forecast.index,
    sarima_forecast.values,
    marker="o",
    label="SARIMA Forecast"
)

plt.fill_between(
    sarima_forecast_intervals.index,
    sarima_forecast_intervals["Lower_95"],
    sarima_forecast_intervals["Upper_95"],
    alpha=0.2,
    label="95% Forecast Interval"
)

plt.title(
    "SARIMA AQI Forecast with 95% Confidence Intervals",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("AQI")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_dir,
        "sarima_aqi_forecast_intervals.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# STEP 10: EXPORT STATISTICAL FORECASTING RESULTS
# Purpose:
#   Preserve model forecasts and forecast uncertainty.
# Method:
#   Export the forecast comparison and SARIMA intervals.
# Output:
#   Submodule 8.4 reports.
# ============================================================

forecast_comparison.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_statistical_forecast_comparison.csv"
    ),
    index=True
)

sarima_forecast_intervals.to_csv(
    os.path.join(
        report_dir,
        "national_aqi_sarima_forecast_intervals.csv"
    ),
    index=True
)

print("Submodule 8.4 forecasting exports completed.")
print("Forecast horizon:", forecast_horizon, "months")
print("Training observations:", len(train_series))
print("Testing observations:", len(test_series))
print("Reports directory:", report_dir)
```
